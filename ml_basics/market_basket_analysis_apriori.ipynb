```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
df = pd.read_csv('Retail_Data.csv')
df = df.iloc[:,1:]
df.head()
df.describe()
```

```python
df1 = pd.get_dummies(df)
df1.head()
df1.describe()
```

```python
df1.shape
```

```python
# find rules which have at least 5% support level
frequent_items = apriori(df1, min_support=0.02, use_colnames=True)
type(frequent_items)
frequent_items.shape
frequent_items

# build rules with minimum Lift of 1
rules = association_rules(frequent_items, metric="lift", min_threshold=1)
type(rules)
rules.shape
rules
```

```python
# Top 10 rules with highest support
rules.sort_values('support', ascending = False).head(10)

# Top 10 rules with highest confidence
rules.sort_values('confidence', ascending = False). head(10)

# Top 10 rules with highest confidence
rules.sort_values('lift', ascending = False).head(10)
```

```python
# Print many statements at same time using the below command
from IPython. core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Import packages
import pandas as pd
from mlxtend. frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
# data is from http://archive.ics.uci.edu/ml/datasets/online+retail
df = pd.read_excel('Online_Retail.xlsx')
df.head()
```

```python
df.shape
```

```python
# find out # unique countries
df.Country.nunique()
df.Country.unique()
```

```python
# Separate out Germany for further analysis
df1 = df[df.Country == "Japan"]
df1.shape
```

```python
# Reorganize data
bskt = pd.pivot_table(df1,index='InvoiceNo', columns='Description',values='Quantity', fill_value=0)
bskt
```

```python
bskt.shape
```

```python
# Write a function to do 0 or 1 coding for all items

def cod(x):
    if x <= 0:
        return 0
    if x >= 1:
        return 1

bskt_sets = bskt.apply(lambda col: col.map(cod))
```

```python
#applying apriori algorithm to generate rules which have support more than 0.05
frequent_itemsets = apriori(bskt_sets.astype(bool), min_support=0.05, use_colnames=True)
frequent_itemsets.head()
```

```python
# find rules with Lift more than one
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
rules.head()
```
