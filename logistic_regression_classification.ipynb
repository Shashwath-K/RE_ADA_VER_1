# Logistic Regression Classification

## Introduction

This notebook implements Logistic Regression for both binary and multiclass classification tasks. Logistic Regression is a fundamental classification algorithm that estimates class probabilities using a logistic (sigmoid) function.

## Problem Statement

Binary and multiclass classification are common problems in machine learning. We apply Logistic Regression to two classic datasets and compare its performance across both settings.

## Datasets

### Binary Classification: Breast Cancer Wisconsin Dataset
- **Why this dataset?** It is a real-world medical dataset for binary classification (malignant vs. benign tumors). It has well-defined features and moderate dimensionality, making it ideal for demonstrating Logistic Regression.
- **Target:** 0 = Malignant, 1 = Benign
- **Features:** 30 numeric features computed from cell nuclei images

### Multiclass Classification: Iris Dataset
- **Why this dataset?** It is the standard benchmark for multiclass classification with three balanced classes. It is small, well-understood, and allows clear demonstration of multiclass Logistic Regression.
- **Target:** 0 = Setosa, 1 = Versicolor, 2 = Virginica
- **Features:** 4 numeric features (sepal length, sepal width, petal length, petal width)

## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer, load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)

import warnings
warnings.filterwarnings('ignore')

np.random.seed(42)

%matplotlib inline
sns.set_style('whitegrid')
```

## Part 1: Binary Classification (Breast Cancer)

### 1.1 Data Exploration

```python
cancer = load_breast_cancer()
df_cancer = pd.DataFrame(cancer.data, columns=cancer.feature_names)
df_cancer['target'] = cancer.target

print('Shape:', df_cancer.shape)
print()
print('Class distribution:')
print(df_cancer['target'].value_counts())
print('  (0 = Malignant, 1 = Benign)')
print()
print('Missing values:', df_cancer.isnull().sum().sum())
print()
df_cancer.head()
```

### 1.2 Preprocessing

```python
X_c = df_cancer.drop('target', axis=1)
y_c = df_cancer['target']

X_c_train, X_c_test, y_c_train, y_c_test = train_test_split(
    X_c, y_c, test_size=0.2, random_state=42, stratify=y_c
)

scaler_c = StandardScaler()
X_c_train_s = scaler_c.fit_transform(X_c_train)
X_c_test_s = scaler_c.transform(X_c_test)

print('Training set shape:', X_c_train.shape)
print('Test set shape:', X_c_test.shape)
```

### 1.3 Model Training

```python
logreg_binary = LogisticRegression(random_state=42, max_iter=10000)
logreg_binary.fit(X_c_train_s, y_c_train)

y_c_pred = logreg_binary.predict(X_c_test_s)
```

### 1.4 Evaluation

```python
print('=== Binary Classification Results ===')
print(f'Accuracy:  {accuracy_score(y_c_test, y_c_pred):.4f}')
print(f'Precision: {precision_score(y_c_test, y_c_pred):.4f}')
print(f'Recall:    {recall_score(y_c_test, y_c_pred):.4f}')
print(f'F1-score:  {f1_score(y_c_test, y_c_pred):.4f}')
print()
print('Classification Report:')
print(classification_report(y_c_test, y_c_pred, target_names=cancer.target_names))
```

```python
cm_binary = confusion_matrix(y_c_test, y_c_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm_binary, annot=True, fmt='d', cmap='Blues',
            xticklabels=cancer.target_names,
            yticklabels=cancer.target_names)
plt.title('Confusion Matrix - Binary Classification (Breast Cancer)', fontweight='bold')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### Discussion

- The model achieves high accuracy, precision, recall, and F1-score.
- The confusion matrix shows very few misclassifications.
- Standard scaling is important for Logistic Regression because it uses gradient descent during optimization; features with larger magnitudes can disproportionately influence the loss.

## Part 2: Multiclass Classification (Iris)

### 2.1 Data Exploration

```python
iris = load_iris()
df_iris = pd.DataFrame(iris.data, columns=iris.feature_names)
df_iris['target'] = iris.target

print('Shape:', df_iris.shape)
print()
print('Class distribution:')
print(df_iris['target'].value_counts())
print('  (0 = Setosa, 1 = Versicolor, 2 = Virginica)')
print()
print('Missing values:', df_iris.isnull().sum().sum())
print()
df_iris.head()
```

### 2.2 Preprocessing

```python
X_i = df_iris.drop('target', axis=1)
y_i = df_iris['target']

X_i_train, X_i_test, y_i_train, y_i_test = train_test_split(
    X_i, y_i, test_size=0.2, random_state=42, stratify=y_i
)

scaler_i = StandardScaler()
X_i_train_s = scaler_i.fit_transform(X_i_train)
X_i_test_s = scaler_i.transform(X_i_test)

print('Training set shape:', X_i_train.shape)
print('Test set shape:', X_i_test.shape)
```

### 2.3 Model Training

```python
logreg_multi = LogisticRegression(random_state=42, max_iter=10000, solver='lbfgs')
logreg_multi.fit(X_i_train_s, y_i_train)

y_i_pred = logreg_multi.predict(X_i_test_s)
```

### 2.4 Evaluation

```python
print('=== Multiclass Classification Results ===')
print(f'Accuracy:          {accuracy_score(y_i_test, y_i_pred):.4f}')
print(f'Precision (macro): {precision_score(y_i_test, y_i_pred, average="macro"):.4f}')
print(f'Recall (macro):    {recall_score(y_i_test, y_i_pred, average="macro"):.4f}')
print(f'F1-score (macro):  {f1_score(y_i_test, y_i_pred, average="macro"):.4f}')
print()
print('Classification Report:')
print(classification_report(y_i_test, y_i_pred, target_names=iris.target_names))
```

```python
cm_multi = confusion_matrix(y_i_test, y_i_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm_multi, annot=True, fmt='d', cmap='Greens',
            xticklabels=iris.target_names,
            yticklabels=iris.target_names)
plt.title('Confusion Matrix - Multiclass Classification (Iris)', fontweight='bold')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

### Discussion

- The multiclass Logistic Regression achieves perfect or near-perfect classification on the Iris dataset.
- The `multinomial` setting uses softmax regression (cross-entropy loss), extending binary logistic regression to multiple classes.
- Macro-averaged metrics treat all classes equally, which is appropriate for this balanced dataset.

## 3. Comparison: Binary vs. Multiclass

```python
comparison_data = {
    'Metric': ['Accuracy', 'Precision', 'Recall', 'F1-score'],
    'Binary (Cancer)': [
        accuracy_score(y_c_test, y_c_pred),
        precision_score(y_c_test, y_c_pred),
        recall_score(y_c_test, y_c_pred),
        f1_score(y_c_test, y_c_pred)
    ],
    'Multiclass (Iris)': [
        accuracy_score(y_i_test, y_i_pred),
        precision_score(y_i_test, y_i_pred, average='macro'),
        recall_score(y_i_test, y_i_pred, average='macro'),
        f1_score(y_i_test, y_i_pred, average='macro')
    ]
}

comp_df = pd.DataFrame(comparison_data).round(4)
print('=== Binary vs. Multiclass Performance Comparison ===')
comp_df
```

```python
comp_df.plot(x='Metric', kind='bar', figsize=(8, 5),
             color=['skyblue', 'lightgreen'], edgecolor='black')
plt.title('Binary vs. Multiclass Logistic Regression Performance', fontweight='bold')
plt.ylabel('Score')
plt.ylim(0, 1.05)
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.show()
```

### Discussion

**Challenges of multiclass classification:**
- Multiclass requires more complex decision boundaries since we must separate multiple classes simultaneously.
- The multinomial approach (softmax) estimates probabilities for each class and selects the highest.

**Interpretation of metrics:**
- For multiclass, metrics are averaged across classes. Macro-averaging gives equal weight to each class, while micro-averaging weights by class frequency.
- Both binary and multiclass models perform well on their respective datasets due to clear feature separability.

## 4. Conclusion

- Binary Logistic Regression achieved high performance on the Breast Cancer dataset with excellent precision and recall.
- Multiclass Logistic Regression (multinomial) performed perfectly on the Iris dataset.
- Feature scaling (StandardScaler) was applied because Logistic Regression uses iterative optimization and benefits from normalized features.
- The confusion matrices provided clear visualization of classification errors.
