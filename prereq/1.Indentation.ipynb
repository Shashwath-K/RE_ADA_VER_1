```python
#Numeric Expressions:
xx = 2      #Storing 2 in a varible xx
xx = xx + 2 #Incrementing the xx by 2 so now xx = 4
print(xx)

yy = 440 * 12 #multiply two values
print(yy)

zz = yy / 100 #Division operation
print(zz)

zz = 5281 // 100 #Purely integer division
print(zz)

jj = 23
kk = jj % 2 #Modulus operation
print(kk)

ii = 5
pp = 4
print(ii**pp) #Directly print ii to the power of pp
```

## Operator Precedence

| Operator                               | Description                                      |  
|----------------------------------------|--------------------------------------------------|
| ()                                     | Parentheses                                      |   
| **                                     | Exponentiation                                   |      
| +x, -x, ~x                             | Unary plus, unary minus, and bitwise NOT         |        
| *, /, //, %                             | Multiplication, division, floor division, and modulus |        
| +, -                                   | Addition and subtraction                         |        
| <<, >>                                  | Bitwise left and right shifts                    |       
| &                                      | Bitwise AND                                      |        
| ^                                      | Bitwise XOR                                      |        
| /                                      | Bitwise OR                                       |        
| ==, !=, >, >=, <, <=, is, is not, in, not in | Comparisons, identity, and membership operators |        
| not                                    | Logical NOT                                      |        
| and                                    | AND                                              |        
| or                                     | OR                                               |        


> Primarliy to narrow it down the Precedence rules will be as follows:
- PEMDAS
 - Parenthesis
 - Power(Exponentiation)
 - Multiplication
 - Division and Remainder
 - Addition Subtraction
 - Left to Right
