# 19. Remove Nth Node From End of List
Given the head of a linked list, remove the nth node from the end of the list and return its head.

``` bash
Example 1:


Input: head = [1,2,3,4,5], n = 2
Output: [1,2,3,5]

Example 2:

Input: head = [1], n = 1
Output: []

Example 3:

Input: head = [1,2], n = 1
Output: [1]
```

````python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val    
        self.next = next 
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
         first_list = list(head)
         rev_list = first_list[-n]
         second_list = []
         for i in range(len(head)):
             if first_list[i] == rev_list:
                 continue
             second_list.append(first_list[i])
         return second_list
        
if __name__ == "__main__":
    head = [1,2,3,4,5]
    head2 = [1,2]
    n1 = 1
    n = 2
    solution = Solution()
    print(solution.removeNthFromEnd(head2,n1))
````

````python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val    
        self.next = next 
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        l1 = ListNode(0)
        dummy = l1
        dummy.next = head
        first = dummy
        second = dummy
        for _ in range(n + 1):
            first = first.next

        while first is not None:
            first = first.next
            second = second.next

        second.next = second.next.next

        return dummy.next
    
if __name__ == "__main__":
    head = [1,2,3,4,5]
    head2 = [1,2]
    n1 = 1
    n = 2
    solution = Solution()
    print(solution.removeNthFromEnd(head,n))
````
