# 17. Letter Combinations of a Phone number

- Given a string containing digits from 2-9 inclusive, return all possible letter combinations that the number could represent. Return the answer in any order.

- A mapping of digits to letters (just like on the telephone buttons) is given below. Note that 1 does not map to any letters.



``` bash
> Example 1:

Input: digits = "23"
Output: ["ad","ae","af","bd","be","bf","cd","ce","cf"]

> Example 2:

Input: digits = "2"
Output: ["a","b","c"]
```

````python
from typing import List
class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        numPadDict = {
            2: ['a', 'b', 'c'],
            3: ['d', 'e', 'f'],
            4: ['g', 'h', 'i'],
            5: ['j', 'k', 'l'],
            6: ['m', 'n', 'o'],
            7: ['p', 'q', 'r', 's'],
            8: ['t', 'u', 'v'],
            9: ['w', 'x', 'y', 'z']
        }
        combinations = []
        lenDict = lambda d: 4 if d in (7, 9) else 3
        n = len(digits)
        for i in range(n):
            for j in range(lenDict(i)):
                if i == n - 1:
                    poss = numPadDict[int(digits[i])][j]+numPadDict[int(digits[i-1])][j]
                    combinations.append(poss)
                else:
                    poss = numPadDict[int(digits[i])][j]+numPadDict[int(digits[i+1])][j]
                    combinations.append(poss)
        return combinations

if __name__ == "__main__":
    digits = "2"
    solution = Solution()
    print(solution.letterCombinations(digits))
````

````python
class Solution:
    def letterCombinations(self, digits: str) -> List[str]:
        if not digits:
            return []

        digit_to_letters = {'2': 'abc','3': 'def','4': 'ghi','5': 'jkl','6': 'mno','7': 'pqrs','8': 'tuv','9': 'wxyz'}
        i=0
        result = list(digit_to_letters[digits[i]])
        char_length=0
        while i<len(digits)-1:
            if char_length==0:
                i+=1
                char_length=len(result)
            while char_length>0:
                root=result.pop(0)
                for next_char in digit_to_letters[digits[i]]:
                    result.append(root+next_char)
                char_length-=1
        return result

if __name__ == "__main__":
    digits = "42"
    solution = Solution()
    print(solution.letterCombinations(digits))
````
