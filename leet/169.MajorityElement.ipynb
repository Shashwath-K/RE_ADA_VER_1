# 169. Majority Element

Given an array nums of size n, return the majority element.

The majority element is the element that appears more than ⌊n / 2⌋ times. You may assume that the majority element always exists in the array.

 
``` bash
Example 1:

Input: nums = [3,2,3]
Output: 3
Example 2:

Input: nums = [2,2,1,1,1,2,2]
Output: 2
```

````python
from typing import List
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        n = len(nums)
        count_dict = {}
        res = 0
        sel_key = 0
        if n == 1:
            return nums[0]
        for i in range(n):
            if nums[i] in count_dict:
                count_dict[nums[i]] += 1
            else:
                count_dict[nums[i]] = 1
        
        for key,counts in count_dict.items():
            prob = counts/2
            count_dict[key] = prob
        for key,values in count_dict.items():
            if int(key*values) > res:
                res = key*values
                sel_key = key   
        return sel_key

if __name__ == "__main__":
    nums = [1,1,1,1,1,1,8]
    sol = Solution()
    print(sol.majorityElement(nums))
````

````python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        nums.sort()
        n = len(nums)
        return nums[n//2]
````
