# 168.ExcelSheet Column title

Given an integer columnNumber, return its corresponding column title as it appears in an Excel sheet.

For example:

A -> 1
B -> 2
C -> 3
...
Z -> 26
AA -> 27
AB -> 28 
...
 
``` bash
Example 1:

Input: columnNumber = 1
Output: "A"
Example 2:

Input: columnNumber = 28
Output: "AB"
Example 3:

Input: columnNumber = 701
Output: "ZY"
```

````python
class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        alpha_dict = {chr(i): i - 64 for i in range(65, 91)}
        for key,value in alpha_dict.items():
            if value == columnNumber:
                return key
            if columnNumber>26:
                res = ""
                while columnNumber > 0:
                    columnNumber -= 1
                    res = chr(columnNumber % 26 + 65) + res
                    columnNumber //= 26
                return res

if __name__ == "__main__":
    columnNumber = 58
    sol = Solution()
    print(sol.convertToTitle(columnNumber))
````
