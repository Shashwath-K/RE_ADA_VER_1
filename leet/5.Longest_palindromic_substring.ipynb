# Longest Palindromic Substrings

Given a string s, return the longest palindromic substring in s.

 

***Example 1:***
``` bash
Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.
```
***Example 2:***
``` bash 
Input: s = "cbbd"
Output: "bb"
``` 

Constraints:

1 <= s.length <= 1000
s consist of only digits and English letters.

````python
dev = 'strrr'
lst1 = []
if dev == reversed(dev):
    print('palindrome')
    lst1.append(dev)
else:
    print('not palindrome')

print(lst1)
````

````python
def func(str1):
    lst1 = []
    for i in range(len(str1)):
        for j in range(i+1, len(str1)+1):
            subStr = str1[i:j]
            if subStr == subStr[::-1]:
                lst1.append(subStr)
    longest = max(lst1, key=len)
    print("The longest palindromic substring is:", longest)

if __name__ == "__main__":
    str1 = input("Enter the string:")
    func(str1)
````

````python
# #Solution 1 : Using brute force
class Solution(object):
    def longestPalindrome(self, s):
        lst1 = []
        for i in range(len(s)):
            for j in range(i+1, len(s)+1):
                subStr = s[i:j]
                if subStr == subStr[::-1]:
                    lst1.append(subStr)
        longest = max(lst1, key=len)
        return longest
````

````python
# #solution 2 : Using the character expansion method
class Solution(object):
   def longestPalindrome(self,s):
    n = len(s)
    if n < 2:
        return s

    longest_palindrome = ""

    for i in range(n):
        l, r = i, i
        while l >= 0 and r < n and s[l] == s[r]:
            if (r - l + 1) > len(longest_palindrome):
                longest_palindrome = s[l:r + 1]
            l -= 1
            r += 1
        l, r = i, i + 1
        while l >= 0 and r < n and s[l] == s[r]:
            if (r - l + 1) > len(longest_palindrome):
                longest_palindrome = s[l:r + 1]
            l -= 1
            r += 1

    return longest_palindrome
````
