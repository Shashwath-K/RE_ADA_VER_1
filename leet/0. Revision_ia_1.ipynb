<h2>Data types and variables</h2>
<details>
<summary>Show more</summary>

**Variables** store data and are dynamically typed.
- it can be referenced and manipulated during program execution.
- A variable is essentially a name that is assigned to a value.
- In python the variables do not require an explicit declaration of a type.
- The type of variable is inferred based on the value assigned.

#### Rules for naming the variables:
- Variable names can only contain letters, digits and underscores(_).
- A variable name cannot start with a digit.
- Variable names are case-sensitive like `myVar` and `myvar` are different.
- Avoid using *Python Keywords* like `if, else, for` as variable names. 

#### Basic Data types:
| Type | Example|
|------|--------|
|int   | 10     |
|float   | 3.14     |
|str   | "hello" or 'hello'     |
|bool   | `True` / `False`    |
|list   | [1,2,3]     |
|tuple   | (1,2,3)     |
|dict   | {"name":"Alex"}     |

- To get the type of the variable use:
  ``` python
  x = 10
  print(type(x))
  ```
#### Object reference in python
Let us assign a variable x to value 5.
> x = 5
- When x = 5 is executed, Python creates an object to represent the value 5 and makes x reference this object.

Now, let's assign another variable y to the variable x.

> y = x

- Python encounters the first statement, it creates an object for the value 5 and makes x reference it. The second statement creates y and references the same object as x, not x itself. This is called a Shared Reference, where multiple variables reference the same object.

<h5>Arithmetic Operators</h5>

> Python Arithmetic operators are used to perform basic mathematical operations like addition, subtraction, multiplication and division.

| Operator | Example Expression | Result |
|----------|--------------------|--------|
| +        | a + b              | 15     |
| -        | a - b              | 5      |
| *        | a * b              | 50     |

<h5>Bitwise Operators</h5>

Bitwise Operators in Python are as follows:

<h5>Identical Operators</h5>

> In Python, is and is not are the identity operators both are used to check if two values are located on the same part of the memory. Two variables that are equal do not imply that they are identical. 

``` python 
is          # True if the operands are identical 
is not      # True if the operands are not identical 
```

````python
# Example for Identity operators
a = 10
b = 20
c = a

print(a is not b)
print(a is c)
````

<h5>Membership Operators</h5>
<details>
<summary>Show more</summary>

> In Python, in and not in are the membership operators that are used to test whether a value or variable is in a sequence.
``` python
in            # True if value is found in the sequence
not in        # True if value is not found in the sequence
```

````python
# Example for Membership Operators
x = 24
y = 20
list = [10, 20, 30, 40, 50]

if (x not in list):
    print("x is NOT present in given list")
else:
    print("x is present in given list")

if (y in list):
    print("y is present in given list")
else:
    print("y is NOT present in given list")
````

<h5>Ternary Operators</h5>
<details>
<summary>Show more</summary>

> in Python, Ternary operators also known as conditional expressions are operators that evaluate something based on a condition being true or false. It was added to Python in version 2.5. 

*It simply allows testing a condition in a single line replacing the multiline if-else, making the code compact.*

> Syntax :  [on_true] if [expression] else [on_false] 

````python
a, b = 10, 20
min = a if a < b else b

print(min)
````

<h2>Comment Lines</h2>
<details>
<summary>Show more</summary>

Comments are used to explain code and make it easier to understand. Python ignores comments when executing the program.


#### 1. Single-Line Comments

Single-line comments start with the `#` symbol.

```python
# This is a single-line comment
x = 10  # This comment explains the variable
```

#### 2. Multi-Line Comments

Python does not have a dedicated multi-line comment syntax. However, multi-line comments can be written using multiple # symbols.

``` python
# This is a multi-line comment
# written using multiple
# single-line comments
```

#### 3. Docstrings (Documentation Strings)

Docstrings are used to document functions, classes, or modules. They are written using triple quotes (''' or """).

``` py
def add(a, b):
    """
    This function adds two numbers
    and returns the result.
    """
    return a + b
```
