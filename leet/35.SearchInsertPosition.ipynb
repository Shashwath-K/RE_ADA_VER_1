# 35. Search Insert Position

Given a sorted array of distinct integers and a target value, return the index if the target is found. If not, return the index where it would be if it were inserted in order.

You must write an algorithm with O(log n) runtime complexity.

 
``` bash
Example 1:

Input: nums = [1,3,5,6], target = 5
Output: 2
Example 2:

Input: nums = [1,3,5,6], target = 2
Output: 1
Example 3:

Input: nums = [1,3,5,6], target = 7
Output: 4
```

````python
from typing import List
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        n = len(nums)
        if target not in nums:
            tgt = Solution.findTar(nums,target)
            return tgt
        else:
            for i in range(n):
                if nums[i]== target:
                    return i

    def findTar(nums,target):
        n = len(nums)
        tgt = target -1
        for i in range(n):
            if nums[i] == tgt:
                return i+1
            else:
                return 0

if __name__ == "__main__":
    nums = [1,3,5,6]
    target = 0
    sol = Solution()
    print(sol.searchInsert(nums, target))
````

````python
class Solution:
    def searchInsert(self, nums: List[int], target: int) -> int:
        n = len(nums)
        if target not in nums:
            tgt = Solution.findTar(nums,target)
            return tgt
        else:
            for i in range(n):
                if nums[i]== target:
                    return i

    def findTar(nums,target):
        n = len(nums)
        if target>nums[n-1]:
            return n
        for i in range(n):
            if nums[i] == target+1 and i>0 and nums[i]!=nums[i-1]+1:
                return i
            if nums[i] == target+1 and i>0:
                return i-1
            if nums[i] == target-1:
                return i+1
        return 0
````
