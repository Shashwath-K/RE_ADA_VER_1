### 1.Company resource allocation using if-else and looping statements
A company allocates employees into 3 shifts (Morning, Evening and Night) based on workload.

- You are given number of employees (N) and total tasks to be completed (T).

- Each employee can handle a maximum of 5 tasks per day (assume uniform
capacity).

- Allocate employees to shifts dynamically so that tasks are completed with minimum overtime.

- If at any point total capacity < total tasks, print “Additional Resources
Required

- If shifts overloaded (>10 employees per shift), trigger “Re-distribution
Recommended”.

```python
total_employees = int(input("Enter total employees :"))
total_tasks = int(input("Enter total tasks: "))
morning_shift = evening_shift = night_shift = 0
total_capacity = total_employees * 5
print("Output:")
if (total_capacity > total_tasks):
    for i in range(total_employees):
        if (morning_shift < 10):
            morning_shift+=1
            total_employees-=1
        elif(evening_shift < 10):
            evening_shift+= 1
            total_employees-=1
        elif(night_shift < 10):
            night_shift+= 1
            total_employees-=1
    if morning_shift > 10 or evening_shift > 10 or night_shift > 10:
        print("Re-distribution Recommended")
    print(f"Morning Shift: {morning_shift} employees\nEvening Shift: {evening_shift} employees\nNight Shift :{night_shift} employees")
else:
    print("Additional Resources required")
print(f"\nTotal Capacity : {total_capacity} tasks")
if(total_capacity >= total_tasks):
    print("Status : Tasks can be completed without any overtime")
```

### 2. Multi-Branch Loan Approval Simulator (Condition + Multi-Level Decision System)
A banking AI analyzes loan applications from multiple branches.
**For each applicant**
- If credit score ≥ 750 & annual income ≥ ₹8L, approve
- If 650–749 & income ≥ ₹10L, forward to “Manual Review”
- If credit score < 650, reject
- If rejected 3 consecutive cases, print "Audit Alert: Unusual Pattern Detected" and
continue


```python
applicants = int(input("Enter the number of applicants : "))
flag = reject_count = 0
approval_status = []
print("\nOutput :")
for i in range(applicants):
    credit_score = int(input("Credit Score : "))
    income = int(input("Income : "))
    if(credit_score >=750 and income >=800000):
        approval_status.append("Approved")
        reject_count = 0
    elif(credit_score in range(650,750) and income >=1000000):
        approval_status.append("Manual Review")
        reject_count = 0
    elif(credit_score < 650):
        approval_status.append("Rejected")
        reject_count+= 1
        if(reject_count >=3):
            flag = 1 
    else:
        approval_status.append("Rejected")
        reject_count+= 1
for i in range(len(approval_status)):
    print(f"Application {i} : {approval_status[i]}")  
if(flag == 1):
    print("\nAudit Alert : Unusual pattern detected")
```

### 3. AI-Based Data Center Energy Monitoring System
<p>A cloud computing company manages several AI-powered data centers that operate
continuously. Each data center tracks its CPU energy usage every hour, measured as a
percentage (% usage). The AI monitoring system must analyze daily usage patterns and
detect performance issues.</p>

**To ensure efficient operation:**
- Energy usage is recorded every hour for 24 hours (1 full day).
- A critical overload is considered if usage exceeds 95% in any hour.
- If average daily energy usage is greater than 80%, the system should recommend
a hardware upgrade.
- If no hourly usage exceeds 95%, it is considered stable operation.
- After processing all data centers, identify the most energy-efficient data center
(i.e., the one with the lowest average daily usage).

**Write a Python program to:**
1. Accept the number of data centers (N).
2. For each data center, accept 24 hourly usage readings (comma-separated
values).
3. Using loops and conditional statements, perform:
 - Calculate average usage.
 - Detect hourly critical overloads (if any hour > 95%).
 - Store and compare average usage across centers to find the most
efficient.
4. Display a summary report highlighting:
 - Average consumption per data center.
 - Any overload alerts.
 - The most energy-efficient data center.


```python
total_datacenters = int(input("Enter number of data centers: "))
total_hours = 24
data_center = []
average_usage = []
for i in range(total_datacenters):
    data_center.append([int(input()) for i in range(total_hours)])
print(f"-------------------------------------\n AI BASED ENERGY MONITORING REPORT\n-------------------------------------")
for i in range(total_datacenters):
    print(f"Data Center {i+1}")
    average_usage.append(sum(data_center[i])/total_hours)
    print(f"Average Usage : {round(average_usage[i],2)}%")
    for j in range(total_hours):
        if(average_usage[i] > 90 and data_center[i][j] > 95):
            print(f"Critical Overload detected at Hour {j+1} (Usage : {data_center[i][j]}%)")
            break
    if(average_usage[i] in range (80,90)):
        print(f"Hardware upgrade recommended")
    if(average_usage[i] < 80):
        print(f"No Critical overload detected")
min_data_usage = min(average_usage)
print(f"\n-------------------------------------\nMost Efficient Data Center : Data Center {average_usage.index(min(average_usage))+1}\n(Lowest Avg Energy Usage:{round(min_data_usage,2)}%)\n-------------------------------------")
print("Conclusion:")
for i in range(total_datacenters):
    if(average_usage[i] < 80):
        print(f"Data Center {i+1} performing normally")
    else:
        print(f"Data Center {i+1} requires performance optimization and load balancing.")
```

### 5. Hospital ICU Resource Optimizer (Complex Conditional + Loop)
An AI ICU monitoring system assigns nurses to critical patients based on severity score (1-10) and availability:

- If severity ≥ 8, assign 2 nurses
- If 5 ≤ severity < 8, assign 1 nurse
- If <5, monitor without nurse
- If total nurses required exceeds availability, recommend “Shift Reallocation”

```python
total_nurses_available = int(input("Total nurses available : "))
patients_severity = [int(input()) for i in range(total_nurses_available)]
n = len(patients_severity)
assigned_nurses = [] * n
print("Output\n")
for i in range(n):
    if(patients_severity[i] >= 8):
        assigned_nurses.append(2)
    elif(patients_severity[i] in range(5,8)):
        assigned_nurses.append(1)
    elif(patients_severity[i] < 5):
        assigned_nurses.append(0)
for i in range(n):
    print(f"Patient {i+1} -> {assigned_nurses[i]} nurse")
print(f"\nTotal assigned : {sum(assigned_nurses)} -> ", end="")
if(sum(assigned_nurses) > total_nurses_available):
    print("Shift Reallocation Recommended")
elif(sum(assigned_nurses) < total_nurses_available):
    print(f"{total_nurses_available-sum(assigned_nurses)} nurses available")
else:
    print("All nurses allocated, No Reallocation needed")
```
