# 67. Add Binary

Given two binary strings a and b, return their sum as a binary string.

 
``` bash
Example 1:

Input: a = "11", b = "1"
Output: "100"
Example 2:

Input: a = "1010", b = "1011"
Output: "10101"

```


````python
class Solution:
    def addBinary(self, a: str, b: str) -> str:
        sum = bin(int(a, 2) + int(b, 2))
        return sum[2:]

if __name__ == "__main__":
    a = "1010"
    b = "1011"
    sol = Solution()
    print(sol.addBinary(a,b))
````
