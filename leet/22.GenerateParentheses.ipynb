# 22. Generate Parentheses

Given n pairs of parentheses, write a function to generate all combinations of well-formed parentheses.

 ``` bash

Example 1:

Input: n = 3
Output: ["((()))","(()())","(())()","()(())","()()()"]

Example 2:

Input: n = 1
Output: ["()"]
```

````python
from typing import List
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        open_counter=close_counter = 0
        vars = [(open_counter, close_counter,'')]
        result = []
        while vars:
            open_counter, close_counter, current = vars.pop()
            if open_counter == n and close_counter == n:
                result.append(current)
            if open_counter < n:
                vars.append((open_counter + 1, close_counter, current + '('))
            if close_counter < open_counter:
                vars.append((open_counter, close_counter + 1, current + ')'))
        return result

if __name__ == "__main__":
    sol = Solution()
    print(sol.generateParenthesis(3)) 
````
