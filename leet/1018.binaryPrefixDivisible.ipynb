```python
# My brute force solution
class Solution(object):
    def prefixesDivBy5(self, nums):
        """
        :type nums: List[int]
        :rtype: List[bool]
        """
        n = len(nums)
        s = ""
        lst = []
        for num in nums:
            s+=str(num)
            lst.append(s)
        lst1 = []
        for i in range(len(lst)):
            lst1.append(int(lst[i],2))
        for j in range(len(lst1)):
            if lst1[j]% 5 == 0:
                lst1[j] = True
            else:
                lst1[j] = False
        return lst1

        
```
