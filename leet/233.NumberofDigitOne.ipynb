# 233. Number of Digit One


Given an integer n, count the total number of digit 1 appearing in all non-negative integers less than or equal to n.

 
``` bash
Example 1:

Input: n = 13
Output: 6

Example 2:

Input: n = 0
Output: 0
```

````python
class Solution:
    def countDigitOne(self, n: int) -> int:
        count = 0
        i = 1
        while i <= n:
            higher = n // (i * 10)
            current = (n // i) % 10
            lower = n % i

            if current < 1:
                count += higher * i
            elif current == 1:
                count += higher * i + lower + 1
            else:
                count += (higher + 1) * i

            i *= 10

        return count

if __name__ == "__main__":
    n = 13
    sol = Solution()
    print(sol.countDigitOne(n))
````

# Accepted Solution

````python
class Solution:
    def countDigitOne(self, n: int) -> int:
        return self._count_of_i(n ,1)

    def _count_of_i(self, n: int, i: int) -> int:
        count, x = 0, 1
        while x <= n:
            a, b = divmod(n, x * 10)
            count += a * x  # (i)
            k = i * x
            if b < k:
                b = 0  # (ii)
            else:
                b = min((b - k) + 1, x)  # (iii)
            count += b
            x *= 10

        return count
````
