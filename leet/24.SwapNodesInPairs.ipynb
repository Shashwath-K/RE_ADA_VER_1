# 24. Swap Nodes in Pairs

Given a linked list, swap every two adjacent nodes and return its head. You must solve the problem without modifying the values in the list's nodes (i.e., only nodes themselves may be changed.)

 
``` bash 
Example 1:

Input: head = [1,2,3,4]

Output: [2,1,4,3]

Input: head = []

Output: []

Example 3:

Input: head = [1]

Output: [1]

Example 4:

Input: head = [1,2,3]

Output: [2,1,3]
```


````python
from typing import Optional
# Definition for singly-linked list.
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
from typing import Optional
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        tempNode = ListNode(0, head)
        prev_ptr, curr_ptr = tempNode, head

        while curr_ptr and curr_ptr.next:
            next_ptr= curr_ptr.next.next
            sec_ptr = curr_ptr.next

            sec_ptr.next = curr_ptr
            curr_ptr.next = next_ptr
            prev_ptr.next = sec_ptr

            prev_ptr = curr_ptr
            curr_ptr = next_ptr
        return tempNode.next
    
if __name__ == "__main__":
    head = [1,1,2,3,3]
    sol = Solution()
    print(sol.swapPairs(head))
````
