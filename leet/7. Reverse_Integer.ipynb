## Reverse Integer

Given a signed 32-bit integer x, return x with its digits reversed. If reversing x causes the value to go outside the signed 32-bit integer range [-231, 231 - 1], then return 0.

Assume the environment does not allow you to store 64-bit integers (signed or unsigned).

 

*Example 1:*

Input: x = 123
Output: 321

*Example 2:*

Input: x = -123
Output: -321

*Example 3:*

Input: x = 120
Output: 21

```python
def reverse( x: int) -> int:
    print(x.bit_length())
    if x > 2**31-1 or x < -2**31:
        numStr = str(x)
        if x < 0 and x.bit_length() < 31:
            revStr = numStr[len(numStr):0:-1]
            revNum = int(revStr)
            return revNum*-1
        elif x > 0 and x.bit_length() < 31:
            revStr = numStr[len(numStr)::-1]
            revNum = int(revStr)
            return revNum
    return 0

if __name__ == "__main__":
    x = -2147483648
    print(reverse(x))
```

```python
def reverse( x: int) -> int:
    print(x.bit_length())
    numStr = str(x)
    if x < 0 and x.bit_length() in range (-2**31,2**31):
        revStr = numStr[len(numStr):0:-1]
        revNum = int(revStr)
        return revNum*-1
    elif x > 0 and x.bit_length() in range (-2**31,2**31):
        revStr = numStr[len(numStr)::-1]
        revNum = int(revStr)
        return revNum
    else:
        return 0

if __name__ == "__main__":
    x = 1534236469
    print(reverse(x))
```

## Accepted Solutions

```python
class Solution:
    def reverse(self, x: int) -> int:
        res = 0
        if x < 0:
            res = int(str(x)[1:][::-1]) * -1
        else:
            res = int(str(x)[::-1])
        
        if res > 2**31-1 or res < -2**31:
            return 0
        else:
            return res

            
```
