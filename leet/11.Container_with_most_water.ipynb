# Container With Most Water

- You are given an integer array height of length n. There are n vertical lines drawn such that the two endpoints of the ith line are (i, 0) an (i, height[i]).

- Find two lines that together with the x-axis form a container, such that the container contains the most water.

- Return the maximum amount of water a container can store.

- Notice that you may not slant the container.



```python
def calHeight(height):
    n = len(height)
    start_ptr= 0
    end_ptr = n - 1
    max_capacity = 0
    while start_ptr<end_ptr:
        if (end_ptr-start_ptr)*min(height[start_ptr],height[end_ptr]) > max_capacity:
            max_capacity = (end_ptr-start_ptr)*min(height[start_ptr],height[end_ptr])
        if height[start_ptr]<height[end_ptr]:
            start_ptr+=1
        else:
            end_ptr-=1
    return max_capacity
    

if __name__ == "__main__":
    height = [1,8,6,2,5,4,8,3,7]
    height1 = [1,2,1]
    print(calHeight(height))
```
