# 21. Merge Two Sorted Lists

You are given the heads of two sorted linked lists list1 and list2.

Merge the two lists into one sorted list. The list should be made by splicing together the nodes of the first two lists.

Return the head of the merged linked list.

``` bash
Example 1:
Input: list1 = [1,2,4], list2 = [1,3,4]
Output: [1,1,2,3,4,4]

Example 2:

Input: list1 = [], list2 = []
Output: []

Example 3:

Input: list1 = [], list2 = [0]
Output: [0]
```

````python
from typing import Optional
# Definition for singly-linked list.
class ListNode:
     def __init__(self, val=0, next=None):
         self.val = val
         self.next = next
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        list1.sort()
        list2.sort()
        list1.extend(list2)
        list1.sort()
        return list1

if __name__ == "__main__":
    list1 = [1,2,4]
    list2 = [1,3,4]
    sol = Solution()
    print(sol.mergeTwoLists(list1,list2))
````

````python
# Definition for singly-linked list.
# class ListNode:
#     def __init__(self, val=0, next=None):
#         self.val = val
#         self.next = next
class Solution:
    def mergeTwoLists(self, list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
        tempNode = ListNode(0, list1)
        prev_ptr, first_ptr, sec_ptr = tempNode, list1, list2

        while first_ptr and sec_ptr:
            if first_ptr.val > sec_ptr.val:
                prev_ptr.next = sec_ptr
                sec_ptr = sec_ptr.next
            else:
                prev_ptr.next = first_ptr
                first_ptr = first_ptr.next
            prev_ptr = prev_ptr.next

        if first_ptr:
            prev_ptr.next = first_ptr
        else:
            prev_ptr.next = sec_ptr
           
        return tempNode.next
````
