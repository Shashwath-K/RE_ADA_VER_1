### Accepted Solution

```python
class Solution(object):
    def findMedianSortedArrays(self, nums1, nums2):
        nums1.extend(nums2)
        res = nums1
        n = len(res)
        res.sort()
        if n % 2 != 0:
            median = float(res[n//2])
        else:
            median = float(res[(n//2)-1] + res[n//2])/2
        return median
```
