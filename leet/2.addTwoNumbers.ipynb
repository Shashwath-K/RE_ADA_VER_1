```python
class Solution(object):
    def addTwoNumbers(self, l1, l2, c=0):

        valt = l1.val + l2.val + c
        c = valt // 10
        ret = ListNode(valt % 10)

        if (l1.next != None or l2.next != None or c!=0):
            if l1.next == None:
                l1.next = ListNode(0)
            if l2.next == None:
                l2.next = ListNode(0)
            ret.next = self.addTwoNumbers(l1.next,l2.next,c)
        return ret
```
