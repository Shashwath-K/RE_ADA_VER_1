# Array Reverse

**Reverse an array arr[]. Reversing an array means rearranging the elements such that the first element becomes the last, the second element becomes second last and so on.**

*Examples:*

``` bash
Input: arr[] = [1, 4, 3, 2, 6, 5]  
Output:  [5, 6, 2, 3, 4, 1]
Explanation: The first element 1 moves to last position, the second element 4 moves to second-last and so on.

Input: arr[] = [4, 5, 1, 2]
Output: [2, 1, 5, 4]
Explanation: The first element 4 moves to last position, the second element 5 moves to second last and so on.
```

#### Method 1

````python
def reverse_array(arr):
    n = len(arr)
    rev_arr = [0]*n
    i = 0
    while n>=0 and i < len(arr):
        rev_arr[i] = arr [n-1]
        i += 1
        n -= 1
    return rev_arr

if __name__ == "__main__":
    arr = [1, 2, 3, 4, 5]
    print("The original array is:", arr)
    reversed_array = reverse_array(arr)
    print("The reversed array is:", reversed_array)
````

#### Method 2

**Using the in-bulit reverse method in Python**

````python
if __name__ == "__main__":
    arr = [1, 2, 3, 4, 5]
    print("The original array is:", arr)
    reversed_array = arr[::-1]
    print("The reversed array is:", reversed_array)
````
