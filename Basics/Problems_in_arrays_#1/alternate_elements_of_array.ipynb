# Alternate elements of an array

<p>Given an array arr[], the task is to print every alternate element of the array starting from the first element.</p>

#### The Following is an Iterative approach to generate the alternative numbers in the list

```python
if __name__ == "__main__":
    def alternate_print(arr):
       print("The alternate elements in the array are:")
       for i in range(0, len(arr),2):
            print(arr[i], end=' ')
        

    arr_input_size = int(input("Enter a number: "))
    arr = [] * arr_input_size
    for i in range(arr_input_size):
        arr.append(int(input(f'Enter {i+1} element:')))
    print("The Entered array:", arr)
    alternate_print(arr)
```

```python
function alternateArrayPrint(arr)
{
    console.log("The alternate elements in the array are: \n");
    for (let i = 0; i < arr.length; i += 2)
    {
        console.log(arr[i] + " ");
    }
}
let arr = [1, 2, 3, 4, 5, 6];
alternateArrayPrint(arr);
```

### The Recursive approach

<p>We can also print the alternate elements using recursion. We start from index = 0, that is the first element of the array and print its value. We then call the recursive function again with the (index + 2) as the current index.</p>

```python
if __name__ == "__main__":
    def recAlternateArrayPrint(arr, index):
        if index >=len(arr):
            return
        print(arr[index], end = " ")
        recAlternateArrayPrint(arr, index + 2)

    arr = [10, 20, 30, 40, 50, 60]
    print("The original array is: ", arr)
    print("The alternate elements in the array are: ")
    recAlternateArrayPrint(arr, 0)
```

```python
function recursiveAlternatePrint(arr, startIndex)
{
    if (startIndex >= arr.length)
    {
        return;
    }
    console.log(arr[startIndex] += " ");
    recursiveAlternatePrint(arr, startIndex + 2);
}
let arr1 = [10, 20, 30, 40, 50, 60];
console.log("The alternate elements in the array using recursion are: \n");
recursiveAlternatePrint(arr1, 0);
```
