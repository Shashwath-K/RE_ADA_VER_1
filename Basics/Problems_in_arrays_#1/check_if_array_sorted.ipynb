# Check if an Array is Sorted

<p>Given an array arr[], check if it is sorted in ascending order or not. Equal values are allowed in an array and two consecutive equal values are 
considered sorted.</p>

- Example:

``` raw
Input: arr[] = [10, 20, 30, 40, 50]
Output: true
Explanation: The given array is sorted.

Input: arr[] = [90, 80, 100, 70, 40, 30]
Output: false
Explanation: The given array is not sorted.
```

````python
def check_sorted(arr):
    if arr == sorted(arr):
        print("The array is sorted.")
    else:
        print("This is an unsorted array.", arr)
        new_array = sorted(arr)
        print("The sorted array is:", new_array)
sorted_arr = [10, 20, 30, 40, 50]
unsorted_arr = [30, 10, 50, 20, 40]
check_sorted(sorted_arr)
check_sorted(unsorted_arr)
````

````python
function check_sorted(arr)
{
    let flag = true;
    for (let i = 0; i < arr.length; i++)
    {
        if (arr[i] >= arr[i + 1])
        {
            flag = false;
            break;
        }
    }
    return flag;
}
function check_sorted_desc(arr)
{
    let res = check_sorted(arr);
    if (res == false)
    {
        console.log("Array is not sorted in ascending order");
    }
    else    
    {
        console.log("Array is sorted in ascending order");
    }
}
console.log(check_sorted_desc([1, 2, 3, 4, 5]));
````
