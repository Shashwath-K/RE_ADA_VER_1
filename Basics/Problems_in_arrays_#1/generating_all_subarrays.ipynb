# Generating All Subarrays

Given an array arr[], the task is to generate all the possible subarrays of the given array.

*Examples:* 

``` bash
Input: arr[] = [1, 2, 3]
Output: [ [1], [1, 2], [2], [1, 2, 3], [2, 3], [3] ]

Input: arr[] = [1, 2]
Output: [ [1], [1, 2], [2] ]
```

````python
def genrating_all_subarrays(arr):
    n = len(arr)
    subarrays = []
    for i in range(n):
        for j in range(i, n):
            subarrays.append(arr[i:j+1])
    return subarrays
if __name__ == "__main__":
    arr = [1, 2, 3]
    print("All subarrays are: ", genrating_all_subarrays(arr))
````
