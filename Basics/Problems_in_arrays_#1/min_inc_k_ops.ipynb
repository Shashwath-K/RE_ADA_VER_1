# Minimum increment by k operations to make all equal

<p>You are given an array of n-elements, you have to find the number of operations needed to make all elements of array equal. Where a single operation can increment an element by k. If it is not possible to make all elements equal print -1.</p>

*Example :*

``` bash
Input : arr[] = {4, 7, 19, 16},  k = 3
Output : 10

Input : arr[] = {4, 4, 4, 4}, k = 3
Output : 0

Input : arr[] = {4, 2, 6, 8}, k = 3
Output : -1
```



````python
def min_increment_k_ops(arr, k):
    n = len(arr)
    max_val = max(arr)
    res = 0
    for i in range(n):
        if (max_val - arr[i]) % k != 0:
            return -1
        else:
            res += (max_val - arr[i]) // k
    return res

    
if __name__ == "__main__":
    arr = [4, 7, 19, 16]
    k = 3
    print(f"Minimum number of operations for {k} to make equal to {max(arr)} is {min_increment_k_ops(arr, k)}")
````
