# Count 1's in a sorted binary array

> Given a binary array arr[] of size n, which is sorted in non-increasing order, count the number of 1's in it. 

**Examples:**

``` bash 
Input: arr[] = [1, 1, 0, 0, 0, 0, 0]
Output: 2
Explanation: Count of the 1's in the given array is 2.

Input: arr[] = [1, 1, 1, 1, 1, 1, 1]
Output: 7

Input: arr[] = [0, 0, 0, 0, 0, 0, 0]
Output: 0
```

````python
def countNumOfOnes(arr):
    n = len(arr)
    count = 0
    arr.sort(reverse=True)
    print("Sorted array: ", arr)
    for i in range(n):
        if arr[i] == 1:
            count +=1
    print(f"Number of 1's in the array of size {n} is : {count}")
if __name__ == "__main__":
    arr = [0,1,0,1,1,0,0,1,1,1]
    print("Original array: ", arr)
    countNumOfOnes(arr)
````
