### Binary Search

<p>
Binary search is a search algorithm which is one of the basic problems for searching a sorted array.
</p>

```python
def binary_search(arr, key):
    n = len(arr)
    low = 0
    high = n - 1

    while low <= high:
        mid = (low + high) // 2
        if arr [mid] == key:
            return mid
        elif arr[mid] < key:
            low = mid+1
        else:
            high = mid -1
    return -1

if __name__ == "__main__":
    arr = [22,35,43,55,67,72,81,90]
    print("Original array: ", arr)
    key = 55
    result = binary_search(arr, key)
    if result != -1:
        print(f"Element {key} found at index : {result}")
    else:
        print("Element not found")
```
