```python
def ternarySearch(arr, key):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid1 = low + (high - low) // 3
        mid2 = high - (high - low) // 3

        if arr[mid1] == key:
            return mid1
        if arr[mid2] == key:
            return mid2

        if key < arr[mid1]:
            high = mid1 - 1
        elif key > arr[mid2]:
            low = mid2 + 1
        else:
            low = mid1 + 1
            high = mid2 - 1
    return -1

if __name__ == "__main__":
    arr = [1, 2, 3, 4, 5, 6, 7]
    key = 6
    idx = ternarySearch(arr, key)
    if idx != -1:
        print(f"{key} found at {idx} position")
    else:
        print(f"Element {key} not found in array")
```
