
# Missing and Repeating in an Array

<p>Given an unsorted array arr[] of size n, containing elements from the range 1 to n, it is known that one number in this range is missing, and another number occurs twice in the array, find both the duplicate number and the missing number.</p>

***Examples:*** 

    Input: arr[] = [3, 1, 3]
    Output: [3, 2]
    Explanation: 3 is occurs twice and 2 is missing.

    Input: arr[] = [4, 3, 6, 2, 1, 1]
    Output: [1, 5] 
    Explanation: 1 is occurs twice and 5 is missing.

```python
def missingRepeating(arr):
    n = len(arr)
    seen = set()
    repeat = set()
    for i in range(n):
        if arr[i] in seen:
            repeat.add(arr[i])
        seen.add(arr[i])
    print(f"Repeating elements {repeat} from the set {seen}")
    count = 1
    for i in range(n+1):
        if count not in seen:
            return print("The missing number in sequence is:", count)
        count+=1
if __name__ == "__main__":
    arr = [3,4,2,6,7,1,3,4]
    print("The Original array:", arr)
    missingRepeating(arr)
```
