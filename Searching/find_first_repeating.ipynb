# Find the first repeating element in an array of integers

<p>Given an array of integers arr[], The task is to find the index of first repeating element in it i.e. the element that occurs more than once and whose index of the first occurrence is the smallest. </p>

**Examples:** 

``` bash 
Input: arr[] = {10, 5, 3, 4, 3, 5, 6}
Output: 5 
Explanation: 5 is the first element that repeats

Input: arr[] = {6, 10, 5, 4, 9, 120, 4, 6, 10}
Output: 6 
Explanation: 6 is the first element that repeats
```

````python
def findFirstRepeatingNumber(arr):
    seen = set()
    for i in range(len(arr)):
        if arr[i] in seen:
            print("First repeating number is:", arr[i])
            return arr[i]
        seen.add(arr[i])

if __name__ == "__main__":
    arr = [1,2,8,7,2,3,3,6]
    print("Original array:", arr)
    findFirstRepeatingNumber(arr)
````
