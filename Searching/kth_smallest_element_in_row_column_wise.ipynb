# Kth smallest element in a row-wise and column-wise sorted 2D array

<p>Given an n × n matrix mat[][] where each row and column is sorted in non-decreasing order, find the kth smallest element, where k lies in the range [1, n²].</p>

**Example:**

``` bash 
Input:  mat[][] = [[10, 20, 30, 40],    k = 3
                             [15, 25, 35, 45],
                             [24, 29, 37, 48],
                             [32, 33, 39, 50]]
Output: 20
Explanation: The sorted order is [10, 15, 20, ...]; the 3rd element is 20.

Input:  mat[][] = [[10, 20, 30, 40],   k = 7
                              [15, 25, 35, 45],
                              [24, 29, 37, 48],
                              [32, 33, 39, 50]]
Output: 30
Explanation: The sorted order is [10, 15, 20, 24, 25, 29, 30, ...]; the 7th element is 30.
```

````python
def sortAndFindKElement(arr, k):
    # flatten the 2D array as the length as of now the value of 'n' is assigned as number of rows instead of the total elements
    n = len(arr) * len(arr[0])
    # Flatten the 2D array
    flat_array = [item for sublist in arr for item in sublist]
    flat_array.sort()
    print("Flattened array(Sorted):", flat_array)
    print(flat_array[k-1])
if __name__ == "__main__":
    arr = [[ 10, 20, 30, 40 ],
           [ 15, 25, 35, 45 ],
           [ 24, 29, 37, 48 ],
           [ 32, 33, 39, 50 ]]
    print("Original 2D array: ",arr)
    sortAndFindKElement(arr, 7)
````
