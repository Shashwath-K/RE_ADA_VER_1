```python
# Print many statements at same time using the below command
from IPython. core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Import the dataset
import pandas as pd
import numpy as np

wine = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine/wine.data',header=None)

wine. columns = ['Class_label', 'Alcohol', 'Malic_acid', 'Ash',
'Alcalinity_of_ash', 'Magnesium', 'Total_phenols',
'Flavanoids', 'Nonflavanoid_phenols', 'Proanthocyanins',
'Color_intensity', 'Hue',
'OD280/OD315_of_diluted_wines', 'Proline']

# Data Exploration
wine.head()
wine. shape
wine.Class_label.nunique()
wine.groupby('Class_label').size()
wine. columns
means = np. round(pd.pivot_table(wine, values= ['Alcohol', 'Malic_acid', 'Ash', 'Alcalinity_of_ash','Magnesium', 'Total_phenols', 'Flavanoids', 'Nonflavanoid_phenols','Proanthocyanins', 'Color_intensity', 'Hue','OD280/OD315_of_diluted_wines', 'Proline'], index = 'Class_label', aggfunc = np.mean), 2)

means
```

```python
# Create Features and Labels
X = wine.drop('Class_label', axis=1)
Y = wine['Class_label']
X. head()
Y. head()
```

```python
# Standardize the data
from sklearn.preprocessing import scale
X_scaled = scale(X)
X_scaled = pd. DataFrame(X_scaled, columns =X. columns)
X_scaled.head()
X_scaled.describe()
```

```python
# Fit PCA model. The number of components will be same as number of variables.
from sklearn.decomposition import PCA
pca = PCA().fit(X_scaled)

pca
```

```python
str(pca.explained_variance_ratio_) # first component explains 36.2% of variance, second 19.2% and so on ....
```

```python
str(pca.explained_variance_ratio_.sum())
```

```python
pca = PCA(5).fit(X_scaled)
str(pca.explained_variance_ratio_.sum())
```

```python
# Kaiser Criteria (Eigen Values)
pca.explained_variance_
```

```python
# Create dataframe of components
pca_df = pd.DataFrame(pca.components_, columns= X.columns)
pca_df
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

import seaborn as sns

plt.figure(figsize=(10, 10))
sns.heatmap(pca_df, cmap='RdYlGn', annot=True)
plt.show()
```

```python
# Original features
X_scaled.shape
```

```python
X_scaled_trnsf = pca.fit_transform(X_scaled)
```

```python
# Transformed features. Instead of having 13, we have 5 now. This can now be plugged into any further modeling work.
X_scaled_trnsf = pd.DataFrame(X_scaled_trnsf, columns=('P1', 'P2', 'P3', 'P4', 'P5'))
X_scaled_trnsf.head()
X_scaled_trnsf.shape
```
