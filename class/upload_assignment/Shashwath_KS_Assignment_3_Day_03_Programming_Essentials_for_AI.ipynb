#### Python If Statement

#### Python Conditions and If statements
Python supports the usual logical conditions from mathematics:

- Equals: `a == b`
- Not Equals: `a != b`
- Less than: `a < b`
- Less than or equal to: `a <= b`
- Greater than: `a > b`
- Greater than or equal to: `a >= b`

These conditions can be used in several ways, most commonly in "if statements" and loops.

An `"if statement"` is written by using the if keyword.

````python
# Example of If statement

a = 33
b = 200
if b > a: # Check whether the value of b is greater than a
  print("b is greater than a") # prints only if the above condition is True
````

In this example we use two variables, `a` and `b`, which are used as part of the if statement to test whether b is greater than a. As `a` is `33`, and `b` is `200`, we know that `200` is greater than `33`, and so we print to screen that `"b is greater than a"`.

#### How If Statements Work
The if statement evaluates a condition (an expression that results in `True` or `False`). If the condition is true, the code block inside the if statement is executed. If the condition is false, the code block is skipped.

````python
# Checking if a number is positive:
number = 15
if number > 0: # condition to check the number is greater than 0
  print("The number is positive") # returns this block if the above statement is True.
````

#### Indentation

Python uses indentation (spaces or tabs at the beginning of a line) to determine the structure and scope of code blocks. Unlike many other programming languages that rely on curly braces to group statements, Python enforces consistent indentation as part of its syntax. This makes the code more readable and visually organized while clearly showing the relationship between different parts of the program.


````python
# If statement, without indentation (will raise an error):

a = 33
b = 200
if b > a: # The following line has no indentation so the if statement has no following block to execute
# There is no `pass` defined either, so this will return an error` 
print("b is greater than a") # you will get an error
````

#### Multiple Statements in If Block

In Python, an if block can contain several statements that are executed when the condition evaluates to true. All of these statements must be indented consistently at the same level to indicate they belong to the same block. Proper indentation ensures clarity and helps Python correctly interpret which lines are part of the conditional structure.


````python
# Multiple statements in an if block

age = 20
if age >= 18:
  '''Only enters this block of code if and only if the condition is True'''
  print("You are an adult")
  print("You can vote")
  print("You have full legal rights")
````

#### Using Variables in Conditions

In Python, boolean variables can be used directly in if statements without needing explicit comparison operators. If the variable evaluates to True, the block of code inside the condition will run. This simplifies the syntax and makes the code cleaner and easier to read while still being clear in its intent.


````python
# Using a boolean variable:

is_logged_in = True # Will be Set to True 
if is_logged_in: # Condtion is evaluated as True
  print("Welcome back!") # Block will be executed 
````

> Python can evaluate many types of values as `True` or `False` in an if statement.

- Zero (0), empty strings (""), None, and empty collections are treated as False. Everything else is treated as True.

- This includes positive numbers (5), negative numbers (-3), and any non-empty string (even "False" is treated as True because it's a non-empty string).

#### Python Elif Statement

#### The Elif Keyword

The `elif` keyword is Python's way of saying "if the previous conditions were not true, then try this condition".

The `elif` keyword allows you to check multiple expressions for True and execute a block of code as soon as one of the conditions evaluates to True.


> In this example a is equal to b, so the first condition is not true, but the elif condition is true, so we print to screen that "a and b are equal".

````python
a = 33
b = 33
if b > a: # Tests whether b is greater than a at first
  print("b is greater than a")
elif a == b: # If the initial condtion is not True then it passes on to the elif
  print("a and b are equal")
````

#### Multiple Elif Statements

You can have as many `elif` statements as you need. Python will check each condition in order and execute the first one that is true.

> In this example, the program checks each condition in order. Since score is 75, it prints "Grade: C" (the first condition that evaluates to true).

````python
# Multiple elif statements to check and classify the grade based on the score
score = 75
# Will evaluate the score sequentially if the initial condition is not True then it moves on to the next
if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
elif score >= 60:
  print("Grade: D")
````

#### How Elif Works

When you use `elif`, Python evaluates the conditions from top to bottom. As soon as it finds a condition that is true, it executes that block and skips all remaining conditions.

**Important:** Only the first true condition will be executed. Even if multiple conditions are true, Python stops after executing the first matching block.


````python
# The following program is to categorizing age groups:

age = 25

if age < 13:
  print("You are a child")
elif age < 20:
  print("You are a teenager")
elif age < 65:
  print("You are an adult")
elif age >= 65:
  print("You are a senior")
````

Use `elif` when you have multiple mutually exclusive conditions to check. This is more efficient than using multiple separate `if` statements because Python stops checking once it finds a true condition.

#### Python Else Statement

#### The Else Keyword

The `else` keyword catches anything which is not caught by the preceding conditions.

The `else` statement is executed when the `if` condition and any `elif` conditions evaluate to False.


````python
a = 200
b = 33
if b > a: # Checks this condition first
  print("b is greater than a")
elif a == b: # Checks this condtion if the first condition is not True
  print("a and b are equal")
else: # If all condtions above are not True then this block is executed.
  print("a is greater than b")
````

> In this example a is greater than b, so the first condition is not true, also the elif condition is not true, so we go to the else condition and print to screen that "a is greater than b".

#### Else Without Elif
You can also have an `else` without the `elif`

````python
# Using if-else statement
a = 200
b = 33
if b > a: # First condtion evaluation
  print("b is greater than a")
else: # Fallback for if condtion when evaluated False
  print("b is not greater than a")
````

### How Else Works

The `else` statement provides a default action when none of the previous conditions are true. Think of it as a "catch all" for any scenario not covered by your `if` and `elif` statements.

**Note:** The `else` statement must come last. You cannot have an `elif` after an `else`.


````python
# Checking even or odd numbers

number = 7

if number % 2 == 0: # Using modulus operator in if to evaluate the condtion
  print("The number is even")
else:
  print("The number is odd")
````

#### Python Logical Operators

Logical operators are used to combine conditional statements. Python has three logical operators:

- and - Returns True if both statements are true
- or - Returns True if one of the statements is true
- not - Reverses the result, returns False if the result is true.

#### The and Operator
The and keyword is a logical operator, and is used to combine conditional statements. Both conditions must be true for the entire expression to be true.

````python
# Test if a is greater than b, AND if c is greater than a:

a = 200
b = 33
c = 500
if a > b and c > a: # Two condtions evaluated in one-line using `and`
  print("Both conditions are True")
````

#### The or Operator
The or keyword is a logical operator, and is used to combine conditional statements. At least one condition must be true for the entire expression to be true.

````python
# Test if a is greater than b, OR if a is greater than c:

a = 200
b = 33
c = 500
if a > b or a > c: # Tests two conditions will return if either one of it is True
  print("At least one of the conditions is True")
````

#### The not Operator
The not keyword is a logical operator, and is used to reverse the result of the conditional statement.

````python
# Test if a is NOT greater than b:

a = 33
b = 200
if not a > b: # Checks the negation of the evaluated expression
  print("a is NOT greater than b")
````

#### Combining Multiple Operators

You can combine multiple logical operators in a single expression. Python evaluates `not` first, then `and`, then `or`.


````python
# Combining and, or, and not:

age = 25
is_student = False
has_discount_code = True
# Checks multiple condtions with mix of and, or 
if (age < 18 or age > 65) and not is_student or has_discount_code:
  print("Discount applies!")
````

#### Python Nested If

#### Nested If Statements

You can have `if` statements inside `if` statements. This is called nested `if` statements.


````python
# Example of nesting if within an If condtion
x = 41

if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
````

#### How Nested If Works

Each level of nesting creates a deeper level of decision making. The code evaluates from the outermost condition inward.


````python
# Checking multiple conditions with nesting:

age = 25
has_license = True
# Evaluating nested conditions
if age >= 18:
  if has_license: # expression evaluated if above is True
    print("You can drive")
  else:
    print("You need a license") # fallback code within nested condition
else:
  print("You are too young to drive") # Fallback for the main 
````

#### Multiple Levels of Nesting
You can nest as many levels deep as needed, but keep in mind that too many levels can make code harder to read.

````python
# Three levels of nesting

score = 85
attendance = 90
submitted = True

if score >= 60:  # first check: minimum passing score
  if attendance >= 80:  # second check: required attendance
    if submitted:  # third check: assignment submitted
      print("Pass with good standing")  # all conditions met
    else:
      print("Pass but missing assignment")  # no submission
  else:
    print("Pass but low attendance")  # attendance not sufficient
else:
  print("Fail")  # score too low
````

#### Nested If vs Logical Operators
Sometimes nested if statements can be simplified using logical operators like `and`. The choice depends on your logic.

````python
# This nested if:

temperature = 25
is_sunny = True

if temperature > 20:
  if is_sunny:
    print("Perfect beach weather!")

# Could also be written with and

temperature = 25
is_sunny = True

if temperature > 20 and is_sunny:
  print("Perfect beach weather!")
````

> Both approaches produce the same result. Use nested if statements when the inner logic is complex or depends on the outer condition. Use and when both conditions are simple and equally important.

````python
# Login validation with nested checks

username = "Emil"
password = "python123"
is_active = True

if username:  # check if username is provided
  if password:  # check if password is provided
    if is_active:  # check if account is active
      print("Login successful")  # all conditions satisfied
    else:
      print("Account is not active")  # user exists but inactive
  else:
    print("Password required")  # missing password
else:
  print("Username required")  # missing username
````

### Python Match

#### The match Statement

The `match` statement is used to perform different actions based on different conditions.

#### The Python Match Statement

Instead of writing many `if..else` statements, you can use the `match` statement.

The `match` statement selects one of many code blocks to be executed.

``` bash
Syntax:

match expression:
  case x:
    code block
  case y:
    code block
  case z:
    code block
```


**This is how it works:**

- The match expression is evaluated once.
- The value of the expression is compared with the values of each case.
- If there is a match, the associated block of code is executed.
The example below uses the weekday number to print the weekday name.

````python
day = 4

match day:  # check value of day using pattern matching
  case 1:
    print("Monday")  # day 1
  case 2:
    print("Tuesday")  # day 2
  case 3:
    print("Wednesday")  # day 3
  case 4:
    print("Thursday")  # day 4
  case 5:
    print("Friday")  # day 5
  case 6:
    print("Saturday")  # day 6
  case 7:
    print("Sunday")  # day 7
````

#### Default Value
Use the underscore character `_` as the last case value if you want a code block to execute when there are not other matches:

````python
day = 4

match day:  # match the value of day
  case 6:
    print("Today is Saturday")  # Saturday case
  case 7:
    print("Today is Sunday")  # Sunday case
  case _:
    print("Looking forward to the Weekend")  # default case if no match
````

#### Combine Values
Use the pipe character `|` as an or operator in the case evaluation to check for more than one value match in one case.

````python
day = 4

match day:  # check value of day using pattern matching
  case 1 | 2 | 3 | 4 | 5:  # multiple values for weekdays
    print("Today is a weekday")  # Monday to Friday
  case 6 | 7:  # weekend values
    print("I love weekends!")  # Saturday and Sunday
````

#### If Statements as Guards
You can add if statements in the case evaluation as an extra condition-check.

````python
month = 5
day = 4

match day:  # match based on day value
  case 1 | 2 | 3 | 4 | 5 if month == 4:  # weekday in April
    print("A weekday in April")
  case 1 | 2 | 3 | 4 | 5 if month == 5:  # weekday in May
    print("A weekday in May")
  case _:  # default case if no match
    print("No match")
````

#### Python While Loops

#### Python Loops

Python has two primitive loop commands:

* `while` loops
* `for` loops

#### The while Loop

With the `while` loop we can execute a set of statements as long as a condition is true.


````python
# Print i as long as i is less than 6:

i = 1
while i < 6: # Checks condition and exits the loop when i becomes 6
  print(i)
  i += 1
````

> The while loop requires relevant variables to be ready, in this example we need to define an indexing variable, i, which we set to 1.


#### The break Statement
With the break statement we can stop the loop even if the while condition is true.

````python
# Exit the loop when i is 3:

i = 1
while i < 6: # Loop is set to execure until i is less than 6
  print(i)
  if i == 3: # checks if i is 3
    break # exits the loop without continuing to iterate furhter
  i += 1
````

#### The continue Statement
With the continue statement we can stop the current iteration, and continue with the next.

````python
# Continue to the next iteration if i is 3:

i = 0
while i < 6:
  i += 1 # pre-increment
  if i == 3: # If i becomes 3, then the iteration is skipped
    continue # continue skips the current iteration
  print(i)
````

#### The else Statement
With the `else` statement we can run a block of code once when the condition no longer is true.

````python
i = 1

while i < 6:  # loop runs while condition is True
  print(i)  # print current value of i
  i += 1  # increment i by 1
else:
  print("i is no longer less than 6")  # runs when loop condition becomes False
````

### Python For Loops


A `for` loop is used for iterating over a sequence (that is either a list, a tuple, a dictionary, a set, or a string).

This is less like the `for` keyword in other programming languages, and works more like an iterator method as found in other object oriented programming languages.

With the `for` loop we can execute a set of statements, once for each item in a list, tuple, set etc.


````python
# Print each fruit in a fruit list:

fruits = ["apple", "banana", "cherry"]
for x in fruits: # prints the contents in the loop one-by-one
  print(x)
````

The `for` loop does not require an indexing variable to set beforehand.

#### Looping Through a String
Even strings are iterable objects, they contain a sequence of characters.

````python
# Loop through the letters in the word "banana":

for x in "banana":
  print(x) #prints the string character one after another
````

### The break Statement

With the `break` statement we can stop the loop before it has looped through all the items.


````python
# Exit the loop when x is "banana":

fruits = ["apple", "banana", "cherry"]
for x in fruits:
  print(x) # Fruits variable printed one after another
  if x == "banana": # but when the item is 'banana' it prints and then breaks the loop and exits
    break
````

````python
# Exit the loop when x is "banana", but this time the break comes before the print:

fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana": # Here the banana is not printed the loop exits 
    break
  print(x)
````

#### The continue Statement
With the continue statement we can stop the current iteration of the loop, and continue with the next:

````python
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana": # Skips the iteration of banana item in the list
    continue
  print(x)
````

#### The range() Function

To loop through a set of code a specified number of times, we can use the `range()` function.

The `range()` function returns a sequence of numbers, starting from 0 by default, and increments by 1 by default, and ends at a specified number.

> The range() function defaults to 0 as a starting value, however it is possible to specify the starting value by adding a parameter: range(2, 6), which means values from 2 to 6 (but not including 6):

````python
# Using the range() function:

for x in range(6): #prints from 0 as start to 5 end
  print(x, end='')
print("\n","-"*5)
# Using the start parameter:

for x in range(2, 6): # starts from 2 ends at 5
  print(x, end="")
````

> The range() function defaults to increment the sequence by 1, however it is possible to specify the increment value by adding a third parameter: range(2, 30, 3):

````python
# Increment the sequence with 3 (default is 1):

for x in range(2, 30, 3): #defines the start end and the number of the skips 
  print(x,"\t", end = "")
````

#### Else in For Loop
The else keyword in a for loop specifies a block of code to be executed when the loop is finished.

````python
for x in range(6):  # loop from 0 to 5
  print(x)  # print current number
else:
  print("Finally finished!")  # runs after loop completes normally
````

````python
# Break the loop when x is 3, and see what happens with the else block:

for x in range(6):
  if x == 3: break  # breaks the loop at x == 3 
  print(x)
else:
  print("Finally finished!") # else is not executed if it encounters break
````

#### Nested Loops
A nested loop is a loop inside a loop.

The `"inner loop"` will be executed one time for each iteration of the `"outer loop"`.

````python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:  # loop through adjectives
  for y in fruits:  # loop through fruits for each adjective
    print(x, y)  # print combination of adjective and fruit
````

#### Why Use Functions?

Imagine you need to convert temperatures from Fahrenheit to Celsius several times in your program. Without functions, you would have to write the same calculation code repeatedly.

#### Python Variable Scope

#### The Concept of Scope
A variable's reach is limited to the specific area where it was defined; this boundary is what we call **scope**. If you try to access a variable outside of its designated region, Python will not recognize it.

#### Local Scope
Whenever a variable is initialized within a function, it is confined to that function's **local scope**. 

* **Internal Access:** It can only be utilized by the code blocks located inside that specific function.
* **Isolation:** The rest of the program remains unaware of the variable's existence, ensuring that local data doesn't accidentally interfere with the wider script.

````python
# A variable created inside a function is available inside that function:

def myfunc():
  x = 300 # x is the scope defined inside the function, which is local to a function
  print(x)

myfunc()
````

#### Nested Functions

#### Function Within a Function
While a local variable is hidden from the main code, it remains accessible to any **nested functions** defined inside that primary function. 

#### Inner Accessibility
If you declare a variable in an outer function, any function sitting inside it inherits the right to use that variable. This creates a chain where the inner "child" function can see the outer "parent" function's data.

````python
# The local variable can be accessed from a function within the function
def myfunc():
  x = 300  # local variable inside outer function

  def myinnerfunc():
    print(x)  # inner function can access outer function's variable

  myinnerfunc()  # call inner function

myfunc()  # call outer function
````

#### Global Scope
A variable created in the main body of the Python code is a global variable and belongs to the global scope.

Global variables are available from within any scope, global and local.

````python
# A variable created outside of a function is global and can be used by anyone:
x = 300 # Global variable defined outside the function scope
def myfunc():
  print(x) # prints the global scope
myfunc()
print(x)
````

#### Naming Variables
If you operate with the same variable name inside and outside of a function, Python will treat them as two separate variables, one available in the global scope (outside the function) and one available in the local scope (inside the function).

````python
# The function will print the local x, and then the code will print the global x:

x = 300 # global x variable 

def myfunc():
  x = 200 # variable x local to the function 
  print(x)

myfunc() # calls the function and prints 200

print(x) # prints the global value
````

````python
# if you use the global keyword, the variable belongs to the global scope.

x = 300 # global x variable 

def myfunc():
  global x # Assigning the current variable x to global scope 
  x = 200
  print(x)

myfunc() # calls the function and prints 200

print(x) # prints the global value this time it's assigned as 200
````

#### Nonlocal Keyword
- The `nonlocal` keyword is used to work with variables inside nested functions.
- The `nonlocal` keyword makes the variable belong to the outer function.

````python
# If you use the nonlocal keyword, the variable will belong to the outer function:

def myfunc1():
  x = "Jane"  # variable in outer function

  def myfunc2():
    nonlocal x  # refers to variable in enclosing function (myfunc1)
    x = "hello"  # modifies outer variable

  myfunc2()  # call inner function
  return x  # return updated value

print(myfunc1())  # print result of function call
````

#### The LEGB Rule
Python follows the LEGB rule when looking up variable names, and searches for them in this order:

* Local - Inside the current function
* Enclosing - Inside enclosing functions (from inner to outer)
* Global - At the top level of the module
* Built-in - In Python's built-in namespace

````python
# Understanding the LEGB rule

x = "global" # Global variable

def outer():
  x = "enclosing" # Enclosing variable inside the function
  def inner():
    x = "local" # local variable of the inner function
    print("Inner:", x)
  inner()
  print("Outer:", x)

outer()
print("Global:", x)
````

#### Python Decorators

#### Decorator Basics
Decorators allow you to inject additional logic into an existing function without modifying its source code. They act as "wrappers" that enhance or change behavior dynamically.

#### The Structure
A decorator is essentially a function that accepts another function as an argument. It then returns a modified version of that function to be executed.



#### Implementation
To use one, define the decorator function first. You can then apply it to any target function by placing the `@decorator_name` syntax directly above the function definition.

````python
# A basic decorator that uppercases the return value of the decorated function.

def changecase(func):  # decorator function
  def myinner():  # wrapper function
    return func().upper()  # call original function and convert result to uppercase
  return myinner  # return wrapped function

@changecase  # apply decorator to function
def myfunction():
  return "Hello Sally"  # original function returns string

print(myfunction())  # call decorated function
````

* By placing `@changecase` directly above the function definition, the function myfunction is being `"decorated"` with the changecase function.
* The function `changecase` is the decorator.
* The function `myfunction` is the function that gets decorated.



#### Multiple Decorator Calls
A decorator can be called multiple times. Just place the decorator above the function you want to decorate.

````python
# Using the @changecase decorator on two functions:

def changecase(func):  # decorator function
  def myinner():  # wrapper function
    return func().upper()  # convert function output to uppercase
  return myinner  # return modified function

@changecase  # apply decorator to myfunction
def myfunction():
  return "Hello Sally"  # original string

@changecase  # apply decorator to otherfunction
def otherfunction():
  return "I am speed!"  # original string

print(myfunction()) 
print(otherfunction())  
````

#### Arguments in the Decorated Function
Functions that require arguments can also be decorated, just make sure you pass the arguments to the wrapper function.

````python
# Functions with arguments can also be decorated:

def changecase(func):
  def myinner(x):
    return func(x).upper() # changes the function and the argument to Uppercase
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John")) # passing the name as John
````

#### *args and **kwargs
Sometimes the decorator function has no control over the arguments passed from decorated function, to solve this problem, add `(*args, **kwargs)` to the wrapper function, this way the wrapper function can accept any number, and any type of arguments, and pass them to the decorated function.

````python
# Secure the function with *args and **kwargs arguments:

def changecase(func):  # decorator function
  def myinner(*args, **kwargs):  # wrapper that accepts any arguments
    return func(*args, **kwargs).upper()  # call original function and convert result to uppercase
  return myinner  # return wrapped function

@changecase  # apply decorator
def myfunction(*nam):
  return "Hello " + nam[1]  # return greeting string

print(myfunction("John", "Mary"))  
````

#### Decorators With Arguments

#### Passing Parameters
To create a decorator that accepts its own arguments, you must introduce an additional **wrapper level**. This essentially involves nesting a second function inside the decorator to handle the extra input.


#### The Extra Layer
The outermost function captures the decorator's arguments, the middle function captures the target function, and the innermost function executes the actual logic. This structure allows you to customize the decorator's behavior for different scenarios.

````python
# A decorator factory that takes an argument and transforms the casing based on the argument value.

def changecase(n):  # outer function that takes a parameter
  def changecase(func):  # decorator function
    def myinner():  # wrapper function
      if n == 1:  # check condition from outer function
        a = func().lower()  # convert result to lowercase
      else:
        a = func().upper()  # convert result to uppercase
      return a  # return modified result
    return myinner  # return wrapper
  return changecase  # return decorator

@changecase(1)  # apply decorator with argument
def myfunction():
  return "Hello Linus"  # original function

print(myfunction()) 
````

#### Multiple Decorators
- You can use multiple decorators on one function.
- This is done by placing the decorator calls on top of each other.
- Decorators are called in the reverse order, starting with the one closest to the function.

````python
# One decorator for upper case, and one for adding a greeting

def changecase(func): #Decorator to change the case of the function 
  def myinner():
    return func().upper()
  return myinner

def addgreeting(func): # Decorator to add the greeting before the function is called
  def myinner():
    return "Hello " + func() + " Have a good day!"
  return myinner

@changecase
@addgreeting
def myfunction():
  return "Tobias"

print(myfunction())
````

#### Handling Arguments in Decorated Functions

You can apply decorators to functions that take inputs; you simply need to ensure those arguments are passed into the **wrapper function**. By defining the inner wrapper to accept parameters, it can successfully relay them to the original function during execution.

````python
# Functions with arguments can also be decorated
def changecase(func):
    # Inner function to wrap the original logic and handle arguments
    def myinner(x):
        # Calls the original function and converts the result to uppercase
        return func(x).upper()
    return myinner

# Applying the decorator to modify the output of myfunction
@changecase
def myfunction(nam):
    return "Hello " + nam

# Executing the decorated function
print(myfunction("John"))
````

#### Flexible Argument Handling with `*args` and `**kwargs`

When a decorator needs to support functions with unpredictable signatures, you should use `*args` and `**kwargs` within the wrapper. This configuration allows the wrapper function to capture any number of positional and keyword arguments, ensuring they are seamlessly forwarded to the decorated function regardless of its specific requirements.

````python
# Secure the function with *args and **kwargs arguments

def changecase(func):
    # Inner function uses *args and **kwargs to accept any number of arguments
    def myinner(*args, **kwargs):
        # Passes arguments through to the original function and uppercases the result
        return func(*args, **kwargs).upper()
    return myinner

# Decorator applied to a function with one argument
@changecase
def myfunction(nam):
    return "Hello " + nam

# Prints the modified, uppercase string
print(myfunction("John"))
````

#### Decorator With Arguments
Decorators can accept their own arguments by adding another wrapper level.

````python
def changecase(n):
    # The outer function acts as a factory to pass arguments to the decorator
    def decorator(func):
        # The actual decorator that wraps the target function
        def myinner():
            # Conditional logic based on the argument 'n' passed to the decorator
            if n == 1:
                a = func().lower()
            else:
                a = func().upper()
            return a
        return myinner
    return decorator

# Applying the decorator with an argument to specify lowercase conversion
@changecase(1)
def myfunction():
    return "Hello Linus"

# Outputs the result based on the logic defined in the nested decorator
print(myfunction())
````

#### Multiple Decorators

* **Layered Application:** A single function can be wrapped by multiple decorators simultaneously.
* **Implementation:** To apply several decorators, stack the decorator calls directly on top of each other above the function definition.
* **Execution Order:** Decorators trigger in **reverse order** (bottom-to-top), meaning the one positioned closest to the function executes first.

````python
# One decorator for upper case, and one for adding a greeting

def changecase(func):
    # Wrapper function to uppercase the result
    def myinner():
        return func().upper()
    return myinner

def addgreeting(func):
    # Wrapper function to wrap the string with a greeting
    def myinner():
        return "Hello " + func() + " Have a good day!"
    return myinner

# Decorators are applied from the bottom up (closest to the function first)
@changecase
@addgreeting
def myfunction():
    return "Tobias"

# Executes addgreeting first, then changecase on the resulting string
print(myfunction())
````

#### Preserving Function Metadata
Functions in Python has metadata that can be accessed using the `__name__` and `__doc__` attributes.

````python
def myfunction():
  return "Have a great day!"

# Accessing the __name__ attribute to print the function's original name
print(myfunction.__name__)
````

````python
# Try returning the name from a decorated function and you will not get the same result

def changecase(func):
    # The wrapper function defined inside the decorator
    def myinner():
        return func().upper()
    return myinner

# The decorator replaces 'myfunction' with the 'myinner' closure
@changecase
def myfunction():
    return "Have a great day!"

# This will print 'myinner' instead of 'myfunction' 
# because the original function metadata is shadowed by the wrapper
print(myfunction.__name__)
````

> To fix this, Python has a built-in function called functools.wraps that can be used to preserve the original function's name and docstring.

````python
# Import functools.wraps to preserve the original function name and docstring.
import functools

def changecase(func):
    # Use @functools.wraps to copy the original function's metadata to the wrapper
    @functools.wraps(func)
    def myinner():
        # Transformation logic
        return func().upper()
    return myinner

# Applying the decorator
@changecase
def myfunction():
    return "Have a great day!"

# Now this correctly prints 'myfunction' instead of 'myinner'
print(myfunction.__name__)
````

#### Python Lambda Functions

* **Definition:** A lambda function is a compact, anonymous function that is defined without a specific name.
* **Capacity and Constraints:** It can accept an unlimited number of arguments, but it is restricted to a **single expression**.
* **Syntax Structure:** The standard format is written as `lambda arguments : expression`.
* **Execution Logic:** When the function is called, the expression is evaluated and the resulting value is automatically returned.

````python
# Add 10 to argument a, and return the result
x = lambda a : a + 10
print(x(5))
````

````python
# Lambda functions can take any number of arguments.

# Multiply argument a with argument b and return the result:
x = lambda a, b : a * b
print(x(5, 6))
````

````python
# Summarize argument a, b, and c and return the result.

x = lambda a, b, c : a + b + c
print(x(5, 6, 2))
````

#### Recursion
Recursion is when a function calls itself.

Recursion is a common mathematical and programming concept. It means that a function calls itself. This has the benefit of meaning that you can loop through data to reach a result.

The developer should be very careful with recursion as it can be quite easy to slip into writing a function which never terminates, or one that uses excess amounts of memory or processor power. However, when written correctly recursion can be a very efficient and mathematically-elegant approach to programming.

````python
# A simple recursive function that counts down from 5.

def countdown(n):
  if n <= 0:
    print("Done!")
  else:
    print(n)
    countdown(n - 1)

countdown(5)
````

#### Base Case and Recursive Case

* **The Base Case:** This is the essential condition that terminates the recursive process and prevents infinite loops.
* **The Recursive Case:** This component involves the function invoking itself, typically using a modified argument to move closer toward the base case.
* **Error Prevention:** In the absence of a base case, the function would execute indefinitely, ultimately resulting in a **stack overflow error**.

````python
# Identifying base case and recursive case.

def factorial(n):
  # Base case: provides a stopping condition to prevent infinite recursion
  if n == 0 or n == 1:
    return 1
  # Recursive case: the function calls itself with a reduced version of the problem
  else:
    return n * factorial(n - 1)

# Calculates 5 * 4 * 3 * 2 * 1
print(factorial(5))
````

#### Fibonacci Sequence
The Fibonacci sequence is a classic example where each number is the sum of the two preceding ones. The sequence starts with 0 and 1:

> 0, 1, 1, 2, 3, 5, 8, 13, ...

The sequence continues indefinitely, with each number being the sum of the two preceding ones.

We can use recursion to find a specific number in the sequence.

````python
# Find the 7th number in the Fibonacci sequence.

def fibonacci(n):
  # Base case: returns n if it is 0 or 1
  if n <= 1:
    return n
  # Recursive case: sums the two preceding numbers in the sequence
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

# Executes the function to find the 7th Fibonacci number (13)
print(fibonacci(7))
````

#### Recursion with Lists

* **Sequential Processing:** Recursion provides a method for navigating lists by operating on a single element during each function call.
* **Divide and Conquer:** Typically, the function processes the first item (the head) and then recursively calls itself with the remainder of the list (the tail) until the list is empty.

````python
# Calculate the sum of all elements in a list.

def sum_list(numbers):
  # Base case: if the list is empty, the sum is 0
  if len(numbers) == 0:
    return 0
  # Recursive case: add the first element to the sum of the remaining list
  else:
    return numbers[0] + sum_list(numbers[1:])

my_list = [1, 2, 3, 4, 5]
# Prints the total sum calculated through recursive slicing
print(sum_list(my_list))
````

````python
# Find the maximum value in a list:

def find_max(numbers):
  # Base case: if the list has only one element, that element is the maximum
  if len(numbers) == 1:
    return numbers[0]
  else:
    # Recursive case: find the maximum of the tail of the list
    max_of_rest = find_max(numbers[1:])
    # Compare the first element with the maximum found in the rest of the list
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest

my_list = [3, 7, 2, 9, 1]
# Prints the highest number found through recursive comparisons
print(find_max(my_list))
````

#### Python Generators
Generators are functions that can pause and resume their execution.

When a generator function is called, it returns a generator object, which is an iterator.

The code inside the function is not executed yet, it is only compiled. The function only executes when you iterate over the generator.

````python
# A simple generator function:

def my_generator():
  # yield suspends function execution and sends a value back to the caller
  yield 1
  yield 2
  yield 3

# Iterating over the generator object produced by the function call
for value in my_generator():
  # Prints each yielded value until the generator is exhausted
  print(value)
````

> Generators allow you to iterate over data without storing the entire dataset in memory. Instead of using `return`, generators use the `yield` keyword.

#### The yield Keyword
The `yield` keyword is what makes a function a generator.

When `yield` is encountered, the function's state is saved, and the value is returned. The next time the generator is called, it continues from where it left off.

````python
# Generator that yields numbers

def count_up_to(n):
  count = 1
  while count <= n:
    # yield pauses the function and returns the current count
    yield count
    # When resumed, the function continues from this point
    count += 1

# Creating a generator object and iterating through it
for num in count_up_to(5):
  # Prints numbers 1 through 5
  print(num)
````

> Unlike `return`, which terminates the function, `yield` pauses it and can be called multiple times.

#### Generators Saves Memory
Generators are memory-efficient because they generate values on-the-fly instead of storing everything in memory.

For large datasets, generators save memory.

````python
# Generator for large sequences:
def large_sequence(n):
  for i in range(n):
    yield i

# This doesn't create a million numbers in memory
gen = large_sequence(1000000)

# Fetch first value
print(next(gen))
# Fetch second value
print(next(gen))
# Fetch third value
print(next(gen))
````

#### Using next() with Generators
You can manually iterate through a generator using the `next()` function

````python
def simple_gen():
  # Yields individual strings one at a time
  yield "Emil"
  yield "Tobias"
  yield "Linus"

gen = simple_gen()

# Fetch first yield
print(next(gen))
# Fetch second yield
print(next(gen))
# Fetch third yield
print(next(gen))
````

#### Generator Expressions

* **Syntactic Similarity:** Generator expressions share a syntax nearly identical to list comprehensions.
* **Notation Difference:** To define a generator instead of a list, simply use **parentheses** `()` in place of square brackets `[]`.
* **Memory Efficiency:** Unlike comprehensions that build the entire collection in memory, these expressions create a generator object that yields items lazily (one at a time).

````python
# List comprehension vs generator expression.

# List comprehension - creates a list
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression - creates a generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
````

````python
# Using a generator expression with sum

# Calculate sum of squares without creating a list
total = sum(x * x for x in range(10))
print(total)
````

#### Fibonacci Sequence Generator
Generators can be used to create the Fibonacci sequence.

It can continue generating values indefinitely, without running out of memory.

````python
# Generate 100 Fibonacci numbers.

def fibonacci():
  a, b = 0, 1
  while True:
    yield a
    a, b = b, a + b

# Get first 100 Fibonacci numbers
gen = fibonacci()
for _ in range(100):
  print(next(gen))
````

#### Generator Methods
Generators have special methods for advanced control.

#### send() Method
The send() method allows you to send a value to the generator.

````python
def echo_generator():
  # Infinite loop to keep the generator alive
  while True:
    # yield pauses execution and can receive values sent via .send()
    received = yield
    print("Received:", received)

gen = echo_generator()

# Prime the generator to reach the first yield statement
next(gen) 

# Send data into the generator
gen.send("Hello")
# Send more data
gen.send("World")
````

#### close() Method
The `close()` method stops the generator.

````python
def my_gen():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    # This block executes when the generator is closed or exhausted
    print("Generator closed")

gen = my_gen()

# Fetch the first value
print(next(gen))

# Manually terminate the generator
gen.close()
````
