### Identity Operators
Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:

| Operator | Description                                              | Example     |
|----------|----------------------------------------------------------|-------------|
| is       | Returns True if both variables are the same object       | x is y      |
| is not   | Returns True if variables are not the same object        | x is not y  |

### Difference Between `is` and `==`
- `is` - Checks if both variables point to the same object in memory
- `==` - Checks if the values of both variables are equal

`````python
# The is operator returns True if both variables point to the same object
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z) # Returns True because z is the same object as x
print(x is y) # Returns False because y is a different object from x
print(x == y) # Returns True because the contents of x and y are the same
`````

### Bitwise Operators
Bitwise operators are used to compare (binary) numbers.

| Operator | Name                   | Description                                                                 | Example   |
|----------|------------------------|-----------------------------------------------------------------------------|-----------|
| &        | AND                    | Sets each bit to 1 if both bits are 1                                      | x & y     |
| \|       | OR                     | Sets each bit to 1 if one of two bits is 1                                 | x \| y    |
| ^        | XOR                    | Sets each bit to 1 if only one of two bits is 1                            | x ^ y     |
| ~        | NOT                    | Inverts all the bits                                                       | ~x        |
| <<       | Zero fill left shift   | Shifts left, pushing zeros from the right and dropping leftmost bits       | x << 2    |

`````python
# Bitwise operators examples
print (5 & 3) # Bitwise AND
print (5 | 3) # Bitwise OR
print (5 ^ 3) # Bitwise XOR
print (~5) # Bitwise NOT
`````

#### Python - List Comprehension

#### List Comprehension

List comprehension provides a shorter syntax for creating a new list based on the values of an existing list.

#### Example

For example, given a list of fruits, you may want to create a new list that contains only the fruits with the letter "a" in their names.

Without list comprehension, you would need to use a `for` loop along with a conditional statement.

`````python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [] # new empty list initialized

for x in fruits:
  if "a" in x:
    newlist.append(x) # append items to the list with value containing string 'a'

print(newlist)
`````

`````python
# Shorter and one-line method using list comprehension

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

newlist = [x for x in fruits if "a" in x]

print(newlist)
`````

#### The Syntax

```python
newlist = [expression for item in iterable if condition == True]
```

The result is a new list, while the original list remains unchanged.

#### Condition

The condition acts as a filter, allowing only the items that evaluate to `True` to be included in the new list.



`````python
newlist = [x for x in fruits if x != "apple"]
print(newlist)
`````

> The condition if x != "apple"  will return True for all elements other than "apple", making the new list contain all fruits except "apple".

#### Example

Without an `if` statement:

```python

```

#### Example

You can also use the `range()` function to create an iterable:



#### Expression

The expression represents the current item in the iteration, and it also defines the final value that will be included in the new list.

You can modify or manipulate this expression before it is added as an item in the resulting list.

`````python
# Converts all the list items in list 'fruits' and converts them to uppercase
newlist = [x.upper() for x in fruits]
print(newlist)
`````

`````python
# Set all values in the new list to 'hello':

newlist = ['hello' for x in fruits]
print(newlist)
`````

> The expression can also contain conditions, not like a filter, but as a way to manipulate the outcome:

`````python
# Assigns values to newlist excluding 'banana' and 'orange'
newlist = [x if x != "banana" else "orange" for x in fruits]
print(newlist)
`````

#### Python - Sort Lists

#### Sort List Alphanumerically

List objects have a `sort()` method that sorts the list alphanumerically in ascending order by default.

`````python
# Sorting the defined list 'thislist' of type string
print(f"The list before sorting : {thislist}")
thislist.sort()
print(f"The list after sorting : {thislist}")
`````

#### Sort Descending
To sort descending, use the keyword argument `reverse = True`

`````python
# Sorting the list in the descending order
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort(reverse = True)
print(thislist)

# For numerical
numList = [100, 50, 65, 82, 23]
numList.sort(reverse = True)
print(numList)
`````

#### Customize Sort Function
You can also customize your own function by using the keyword argument `key = function`.

The function will return a number that will be used to sort the list (the lowest number first).

`````python
# Script to call a function and sort the elements in the list
def myfunc(n):
  return abs(n - 50)

thislist = [100, 50, 65, 82, 23]
thislist.sort(key = myfunc)
print(thislist)
`````

#### Case Insensitive Sort

By default, the `sort()` method is case sensitive, which means that all uppercase letters are sorted before lowercase letters.

`````python
# the regular sort with case-sensitve approach
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort()
print(thislist)
`````

`````python
# Performing a case-insensitive sort of the list

thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
`````

`````python
# Reverse order of the list without case-sensitiveness
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.reverse()
print(thislist)
`````

#### Python : Update Tuples

Tuples are immutable, which means their elements cannot be modified, added, or removed after creation.

However, there are some alternative approaches to work around this limitation.

#### Change Tuple Values

Once a tuple has been created, its values cannot be altered. This property is known as immutability.

A common workaround is to convert the tuple into a list, make the necessary changes to the list, and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
fruit_list[1] = "kiwi"
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
print(fruit_tuple)  # Updated tuple
`````

#### Add Items

Since tuples are immutable, they do not support an `append()` method. However, there are alternative ways to include new elements.

1. **Convert to a list:** Similar to modifying a tuple, you can transform it into a list, add the desired item(s), and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
# Add a new item to the list
fruit_list.append("orange")
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
`````

2. Add tuple to a tuple. You are allowed to add tuples to tuples, so if you want to add one item, (or many), create a new tuple with the item(s), and add it to the existing tuple.

`````python
# Create a single-item tuple
extra_fruit = ("orange",)
# Add the new tuple to the existing tuple
fruit_tuple += extra_fruit
print(fruit_tuple)
`````

#### Remove Items

**Note:** You cannot directly remove items from a tuple.

Since tuples are immutable, their elements cannot be deleted. However, you can use the same workaround as used for modifying and adding items: convert the tuple into a list, perform the removal, and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
# Remove "apple" from the list
fruit_list.remove("apple")
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
print(fruit_tuple)
`````

> But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking".

`````python
fruits = ("apple", "banana", "cherry")
# Unpacking tuple values into variables
green, yellow, red = fruits
print(green)
print(yellow)
print(red)
`````

#### Using Asterisk (*)

If the number of variables is fewer than the number of values in a tuple, you can use an asterisk (`*`) before a variable name. The remaining values will then be collected into that variable as a list.

`````python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
# Unpack first two values, rest go into a list
green, yellow, *red = fruits
print(green)
print(yellow)
print(red)
`````

> If the asterisk is added to another variable name than the last, Python will assign values to the variable until the number of values left matches the number of variables left.

`````python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")
# Unpack first and last item, middle items go into a list
green, *tropic, red = fruits
print(green)
print(tropic)
print(red)
`````

#### Python - Join Sets

#### Join Sets

In Python, there are multiple ways to combine two or more sets.

The `union()` and `update()` methods are used to merge all elements from the sets.

The `intersection()` method returns only the common elements between sets.

The `difference()` method returns elements that are present in the first set but not in the other set(s).

The `symmetric_difference()` method returns all elements except those that appear in both sets.

#### Union

The `union()` method creates and returns a new set containing all elements from both sets.

`````python
set_a = {"a", "b", "c"}
set_b = {1, 2, 3}
# Combine both sets into a new set
combined_set = set_a.union(set_b)
print(combined_set)
`````

> You can use the | operator instead of the union() method, and you will get the same result.

#### Intersection
**Keep ONLY the duplicates**

The `intersection()` method will return a new set, that only contains the items that are present in both sets.

`````python
# Join set1 and set2, but keep only the duplicates

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# To keep the common elements from two sets
set3 = set1.intersection(set2)
print(set3)
`````

> You can use the & operator instead of the intersection() method, and you will get the same result.

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# Use `&` to join two sets:
set3 = set1 & set2
print(set3)
`````

> **Note:** The `&` operator can only be used to find the intersection between sets. Unlike the `intersection()` method, it cannot be directly applied with other data types.

The `intersection_update()` method also retains only the common elements between sets, but instead of returning a new set, it updates the original set in place.

`````python
# Keep the items that exist in both set1, and set2:

set_a = {"apple", "banana", "cherry"}
set_b = {"google", "microsoft", "apple"}

# Keep only items that exist in both sets using intersection_update
set_a.intersection_update(set_b)

print(set_a)
`````

> The values True and 1 are considered the same value. The same goes for False and 0.

`````python
# Join sets that contains the values True, False, 1, and 0, and see what is considered as duplicates:
set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}
set3 = set1.intersection(set2)
print(set3)
`````

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
# Keep all items from set1 that are not in set2:
set3 = set1.difference(set2)
print(set3)
`````

> You can use the - operator instead of the difference() method, and you will get the same result.

`````python
# Use - to join two sets.
set3 = set1 - set2
print(set3)
`````

> The `difference_update()` method will keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
# Use the difference_update() method to keep only the items from the first set that are not present in the other set.
set1.difference_update(set2)
print(set1)
`````

#### Symmetric Differences
The `symmetric_difference()` method will keep only the elements that are NOT present in both sets.

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
# Keep the items that are not present in both sets.
set3 = set1.symmetric_difference(set2)
print(set3)
`````

> You can use the `^` operator instead of the `symmetric_difference()` method, and you will get the same result.

`````python
# Use ^ to join two sets.
set3 = set1 ^ set2
print(set3)
`````

The `symmetric_difference_update()` method will also keep all but the duplicates, but it will change the original set instead of returning a new set.

`````python
# symmetric_difference_update() method to keep the items that are not present in both sets

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.symmetric_difference_update(set2)

print(set1)
`````

#### Python - Loop Dictionaries

#### Loop Through a Dictionary

You can iterate through a dictionary using a `for` loop.

By default, looping through a dictionary returns its keys. However, there are methods available to access values or key-value pairs as well.

`````python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
`````

#### Python - Copy Dictionaries

#### Copy a Dictionary

You cannot copy a dictionary by simply assigning it using `dict2 = dict1`, because this only creates a reference to the original dictionary. Any changes made to `dict1` will also affect `dict2`.

To properly copy a dictionary, you can use built-in methods such as the `copy()` method.

`````python
# Make a copy of a dictionary with the copy() method.

thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
# using copy() method 
mydict = thisdict.copy()
print(mydict)
`````

> Another way to make a copy is to use the built-in function dict().

`````python
# Make a copy of a dictionary with the dict() function:
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
# Using a dict() method to create a new copy of the existing dictionary
mydict = dict(thisdict)
print(mydict)
`````

#### Nested Dictionaries

A dictionary can hold other dictionaries as values. This structure is known as a **nested dictionary**.

`````python
# Create a dictionary that contain three dictionaries

# A dictionary called myfamily containing nested dictionaries
# Each key (child1, child2, child3) represents a child

myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}
`````

> Or, if you want to add three dictionaries into a new dictionary.

`````python
# Create three dictionaries, then create one dictionary that will contain the other three dictionaries.

child1 = {
  "name" : "Emil",
  "year" : 2004
}
child2 = {
  "name" : "Tobias",
  "year" : 2007
}
child3 = {
  "name" : "Linus",
  "year" : 2011
}

myfamily = {
  "child1" : child1,
  "child2" : child2,
  "child3" : child3
}
`````

#### Access Items in Nested Dictionaries
To access items from a nested dictionary, you use the name of the dictionaries, starting with the outer dictionary.

`````python
# Print the name of child 2
print(myfamily["child2"]["name"])
`````

### Loop Through Nested Dictionaries

You can iterate through a nested dictionary using the `items()` method.

This allows you to access both the keys and their corresponding values while looping through the dictionary.

`````python
# Loop through the keys and values of all nested dictionaries.

for x, obj in myfamily.items():
  print(x)

  for y in obj:
    print(y + ':', obj[y])
`````

### Python Dictionary Methods

Python provides several built-in methods for working with dictionaries.

| Method | Description |

| `items()` | Returns a view of key-value pairs as tuples |
| `keys()` | Returns a view of all keys in the dictionary |
| `pop()` | Removes the element with the specified key |

| `setdefault()` | Returns the value of a key; if not present, inserts it with a specified value |

| `values()` | Returns a view of all values in the dictionary |
