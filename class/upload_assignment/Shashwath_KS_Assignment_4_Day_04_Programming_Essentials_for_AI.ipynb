#### Python range
The built-in `range()` function returns an immutable sequence of numbers, commonly used for looping a specific number of times.

This set of numbers has its own data type called range.

#### Creating ranges
The `range()` function can be called with 1, 2, or 3 arguments, using this syntax:

``` python
range(start, stop, step)
```

#### Call range() With One Argument
If the range function is called with only one argument, the argument represents the stop value.


The `start` argument is optional, and if not provided, it defaults to 0.

range(10) returns a sequence of each number from 0 to 9. (The start argument, 0 is inclusive, and the stop argument, 10 is exclusive).

````python
x = range(10) #Calling range with a single argument i.e start
````

````python
x = range(3, 10) # Calling range with two arguments i.e start and end
````

````python
x = range(3,10,2) # Calling range with three arguments i.e start, stop and step
````

#### Using Ranges
Ranges are often used in `for` loops to iterate over a sequence of numbers.

````python
for i in range(10): # called a range of 0 to iterate upto 10 times ending at 9.
    print(i) #returns the result
````

#### Using List to Display Ranges
The range object is a data type that represents an immutable sequence of numbers, and it is not directly displayable.

Therefore, ranges are often converted to lists for display.

````python
# Convert different ranges to lists:
print(list(range(5))) # prints list of range 0 to 4
print(list(range(1,6))) # prints list of range 1 to 5
print(list(range(5,20,5))) # prints list of range 5 to 20 with 3 steps skips
````

#### Slicing Ranges
Like other sequences, ranges can be sliced to extract a subsequence.

````python
r = range(10) # Taing a range from 0 to 9 (10 numbers)
print(r[2]) # Retriving the item in index position 2
# The following returns a Rnage object
print(r[:3]) # Slicing the list to print upto index position 3
````

#### Membership Testing
Ranges support membership testing with the in operator.

````python
# Test if the numbers 6 and 7 are present in a range:

r = range(0, 10, 2) # Defining the range of start from 0 to end 10 with 2 skips steps
print(6 in r) # Returns True if item in variable r
print(7 in r) # Returns False if item in variable r is absent
````

#### Length
Ranges support the len() function to get the number of elements in the range.


````python
r = range(0, 10, 2) # Defining the range of start from 0 to end 10 with 2 skips steps
print(len(r)) # Returns the length of the list
````

#### Python Iterators
An iterator is an object that contains a countable number of values.

An iterator is an object that can be iterated upon, meaning that you can traverse through all the values.

Technically, in Python, an iterator is an object which implements the iterator protocol, which consist of the methods `__iter__()` and `__next__()`.

#### Iterator vs Iterable
Lists, tuples, dictionaries, and sets are all iterable objects. They are iterable containers which you can get an iterator from.

All these objects have a iter() method which is used to get an iterator

````python
# Return an iterator from a tuple, and print each value:

mytuple = ("apple", "banana", "cherry") # Create a tuple with 3 items (iterable object)
myit = iter(mytuple) # iter() method used for iterating through the iterables.
# Prints the Items in order 
print(next(myit)) # apple
print(next(myit)) # banana
print(next(myit)) # cherry
````

````python
#Similarly using a string as an iterable
mystr = "banana" 
myit = iter(mystr)
# String treats each letters as an item it can be iterated as 
print(next(myit)) # Outputs: b
print(next(myit)) # Outputs: a
print(next(myit)) # Outputs: n
print(next(myit)) # Outputs: a
print(next(myit)) # Outputs: n
print(next(myit)) # Outputs: a
````

#### Looping Through an Iterator
We can also use a for loop to iterate through an iterable object.

> The for loop actually creates an iterator object and executes the `next()` method for each loop.

````python
# Iterate the values of a tuple:

mytuple = ("apple", "banana", "cherry") # tuple as an iterable
# Using for loop as an iterator
for x in mytuple:
  print(x)
````

#### Create an Iterator
To create an object/class as an iterator you have to implement the methods __iter__() and __next__() to your object.

As you will learn in the Python Classes/Objects chapter, all classes have a function called __init__(), which allows you to do some initializing when the object is being created.

The __iter__() method acts similar, you can do operations (initializing etc.), but must always return the iterator object itself.

The __next__() method also allows you to do operations, and must return the next item in the sequence.

````python
class MyNumbers:
  def __iter__(self):
    self.a = 1          # Initialize starting value
    return self         # Return the iterator object

  def __next__(self):
    x = self.a          # Store current value
    self.a += 1         # Increment value
    return x            # Return next number

myclass = MyNumbers()   # Create object of MyNumbers
myiter = iter(myclass)  # Get iterator from object

print(next(myiter))     # Print 1
print(next(myiter))     # Print 2
print(next(myiter))     # Print 3
print(next(myiter))     # Print 4
print(next(myiter))     # Print 5
````

#### StopIteration
The example above would continue forever if you had enough next() statements, or if it was used in a for loop.

To prevent the iteration from going on forever, we can use the `StopIteration` statement.

In the __next__() method, we can add a terminating condition to raise an error if the iteration is done a specified number of times.

````python
# Import the datetime module and display the current date:

import datetime

x = datetime.datetime.now()
print(x)
````

````python
import re

txt = "The rain in Spain"
x = re.search("\s", txt)

print("The first white-space character is located in position:", x.start())
````

### The split() Function

````python
# Split at each white-space character:

import re

txt = "The rain in Spain"
x = re.split("\s", txt)
print(x)
````

### Python Try Except

- The try block lets you test a block of code for errors.
- The except block lets you handle the error.
- The else block lets you execute code when there is no error.
- The finally block lets you execute code, regardless of the result of the try- and except blocks.

### Exception Handling
When an error occurs, or exception as we call it, Python will normally stop and generate an error message.

These exceptions can be handled using the try statement

````python
# The try block will generate an exception, because x is not defined:

try:
  print(x)
except:
  print("An exception occurred")
  
````

### Many Exceptions
You can define as many exception blocks as you want, e.g. if you want to execute a special block of code for a special kind of error.

````python
# Print one message if the try block raises a NameError and another for other errors:

try:
  print(x)
except NameError:
  print("Variable x is not defined")
except:
  print("Something else went wrong")
````

````python
# In this example, the try block does not generate any error:

try:
  print("Hello")
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
````

````python
# The finally block, if specified, will be executed regardless if the try block raises an error or not.
try:
  print(x)
except:
  print("Something went wrong")
finally:
  print("The 'try except' is finished")
````

````python
# Try to open and write to a file that is not writable:

try:
  f = open("demofile.txt")
  try:
    f.write("Lorum Ipsum")
  except:
    print("Something went wrong when writing to the file")
  finally:
    f.close()
except:
  print("Something went wrong when opening the file")
````

### Raise an exception
As a Python developer you can choose to throw an exception if a condition occurs.

To throw (or raise) an exception, use the raise keyword.

- The raise keyword is used to raise an exception.

````python
# Raise an error and stop the program if x is lower than 0:

x = -1

if x < 0:
  raise Exception("Sorry, no numbers below zero")
````

````python
# Raise a TypeError if x is not an integer:

x = "hello"

if not type(x) is int:
  raise TypeError("Only integers are allowed")
````

#### Python Inheritance

#### Add the `__init__()` Function

So far, we have created a child class that inherits the properties and methods from its parent.

Now, we want to add the `__init__()` function to the child class (instead of using the `pass` keyword).



#### Note

The `__init__()` function is called **automatically** every time the class is used to create a new object.

````python
# Parent class definition: Person
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname  # Initialize first name
        self.lastname = lname    # Initialize last name

    def printname(self):
        print(self.firstname, self.lastname)  # Print full name


# Child class definition: Student, which inherits from Person
class Student(Person):
    def __init__(self, fname, lname, year):
        super().__init__(fname, lname)  # Call the parent constructor to initialize name
        self.graduationyear = year     # Add new property for graduation year

    def welcome(self):
        print(f"Welcome {self.firstname} {self.lastname} to the class of {self.graduationyear}")  # Custom method for Student


# Create Person object
person1 = Person("John", "Doe")
person1.printname()  # Call method from Person class

# Create Student object
student1 = Student("Mike", "Olsen", 2022)
student1.printname()  # Call inherited method from Person class
student1.welcome()    # Call method from Student class
````

#### Python Polymorphism

- **Definition**: The word "polymorphism" means "many forms". In programming, it refers to methods/functions/operators with the same name that can be executed on many objects or classes.

---

#### Function Polymorphism

- A function like `len()` can be used on different types of objects such as strings, tuples, and dictionaries, returning different results based on the object type.

---

#### Class Polymorphism

- Polymorphism can be applied to class methods, allowing multiple classes to have methods with the same name but different behaviors.

---

#### Inheritance Class Polymorphism

- Polymorphism can also be used with child classes inheriting from a parent class. Child classes can override inherited methods, allowing different behaviors while still following the same method name.

---


#### Additional Details to Help Understand Python Polymorphism

---

#### Key Concepts in Polymorphism

1. **Method Overloading vs. Method Overriding**  
   - **Method Overloading**: In some programming languages (like Java), you can have multiple methods with the same name but different parameters. Python doesn’t support method overloading in the traditional sense, but you can achieve similar behavior by using default arguments or variable-length argument lists (`*args` or `**kwargs`).
   - **Method Overriding**: This happens when a child class defines a method with the same name as one in the parent class. The child class method **overrides** the parent class method, which means the child’s version of the method will be called, not the parent’s.

---

#### Benefits of Polymorphism

- **Flexibility**: Polymorphism allows you to write code that can work with objects of different classes in a generic way. This reduces the complexity of the code and makes it more flexible and reusable.
  
- **Maintainability**: By using polymorphism, you can modify or extend your code without changing the core structure, leading to cleaner and more maintainable code.

- **Simplifies Code**: You don't need to write multiple `if-else` or `switch` statements to handle different object types. Polymorphism lets you call the same method on different objects, regardless of their class.

---

#### Polymorphism in Practice

- **Real-world analogy**: Consider a `Shape` class with methods like `area()` and `perimeter()`. Different subclasses like `Circle`, `Rectangle`, and `Triangle` will all have their own implementations of these methods, but the method name stays the same. So, when you call `shape.area()`, Python knows the exact behavior based on the object type (circle, rectangle, etc.), even though you are calling the same method name.

- **Dynamic Dispatch**: Python’s polymorphism works through dynamic dispatch, meaning the method call is resolved at runtime. This is one of the reasons why polymorphism in Python is so flexible—methods are bound to objects at runtime, not at compile-time.

---

#### Practical Example: Using Polymorphism in Collections

Polymorphism is particularly useful when you want to deal with a collection of objects, where the objects can belong to different classes but implement the same methods. For example, you could store a variety of different vehicles in a list, and call the same `move()` method on each, even though `move()` behaves differently for each class (Car, Boat, Plane).


````python
# Polymorphism in Python

# Parent class Vehicle
class Vehicle:
    def __init__(self, brand, model):
        self.brand = brand  # Brand of the vehicle
        self.model = model  # Model of the vehicle

    def move(self):
        print("Move!")  # Default move method for Vehicle


# Child class Car, inherits from Vehicle
class Car(Vehicle):
    def move(self):
        print("Drive!")  # Overridden move method for Car


# Child class Boat, inherits from Vehicle
class Boat(Vehicle):
    def move(self):
        print("Sail!")  # Overridden move method for Boat


# Child class Plane, inherits from Vehicle
class Plane(Vehicle):
    def move(self):
        print("Fly!")  # Overridden move method for Plane


# Create objects of each class
car1 = Car("Ford", "Mustang")
boat1 = Boat("Ibiza", "Touring 20")
plane1 = Plane("Boeing", "747")

# List of objects to iterate through
vehicles = [car1, boat1, plane1]

# Call move() method for each object in the list
for vehicle in vehicles:
    print(f"{vehicle.brand} {vehicle.model} - ", end="")
    vehicle.move()  # Polymorphism in action
````

#### Python Encapsulation

**Encapsulation** is about protecting data inside a class. It involves keeping data (properties) and methods together while controlling how the data can be accessed from outside the class. This helps in preventing accidental changes to data and hides internal details of the class's workings.

---

#### Private Properties

In Python, properties can be made **private** by using a double underscore `__` prefix.

- **Example**: You can create a private class property such as `__age` to restrict access from outside the class.

---

#### Accessing Private Property Value

To access private properties, you can create a **getter method**.

- **Example**: Use a method like `get_age()` to access the value of a private property like `__age`.

---

#### Modifying Private Property Value

To modify a private property, you can create a **setter method**. The setter method can also validate the value before modifying the property.

- **Example**: Use `set_age()` to change the value of a private property like `__age`, ensuring it's a valid positive number.

---

#### Why Use Encapsulation?

- **Data Protection**: Prevents accidental modification of data.
- **Validation**: Allows you to validate data before it is set.
- **Flexibility**: Internal implementation can change without affecting external code.
- **Control**: You maintain control over how data is accessed and modified.

---

#### Protected Properties

Python uses a **single underscore `_`** for **protected** properties. It’s a convention indicating that the property should not be accessed directly, but Python doesn't enforce this restriction.

- **Example**: A property like `_salary` is intended for internal use, but can still be accessed outside the class.

---

#### Private Methods

You can also make methods **private** using the double underscore `__` prefix.

- **Example**: Create a private method like `__validate()` that can only be used within the class itself.

---

#### Name Mangling

Python uses **name mangling** for private properties and methods. This means that private attributes (with double underscores) are internally renamed by adding `_ClassName` in front of the name.

- **Example**: A private property like `__age` becomes `_Person__age`. While you can technically access it, it's not recommended because it defeats the purpose of encapsulation.

---

Encapsulation helps keep your code secure, organized, and easy to maintain by controlling how data is accessed and manipulated.

````python
# Encapsulation in Python

# Class with Encapsulation to protect data (private properties)
class Person:
    def __init__(self, name, age):
        self.name = name  # Public property
        self.__age = age  # Private property (Encapsulation)

    # Getter method to access private property
    def get_age(self):
        return self.__age

    # Setter method to modify private property with validation
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Age must be positive")


# Create an instance of Person
p1 = Person("Tobias", 25)

# Access public property
print(p1.name)

# Try accessing private property directly (will cause an error)
# print(p1.__age)  # Uncommenting this will cause an AttributeError

# Correct way to access private property using getter method
print(p1.get_age())

# Modify private property using setter method
p1.set_age(26)
print(p1.get_age())  # After modification

# Try setting invalid age (negative number)
p1.set_age(-5)

# Class with Encapsulation for student grade validation
class Student:
    def __init__(self, name):
        self.name = name
        self.__grade = 0  # Private property for grade

    # Getter for grade
    def get_grade(self):
        return self.__grade

    # Setter for grade with validation
    def set_grade(self, grade):
        if 0 <= grade <= 100:
            self.__grade = grade
        else:
            print("Grade must be between 0 and 100")

    # Method to check student's status based on grade
    def get_status(self):
        if self.__grade >= 60:
            return "Passed"
        else:
            return "Failed"


# Create a student object and set grade with validation
student1 = Student("Emil")
student1.set_grade(85)  # Valid grade
print(student1.get_grade())
print(student1.get_status())  # Check student's status

# Try setting an invalid grade
student1.set_grade(150)  # Invalid grade, will print an error message

# Class with a private method for calculation
class Calculator:
    def __init__(self):
        self.result = 0

    # Private method (not meant to be accessed outside the class)
    def __validate(self, num):
        if not isinstance(num, (int, float)):
            return False
        return True

    # Public method to add number
    def add(self, num):
        if self.__validate(num):
            self.result += num
        else:
            print("Invalid number")


# Create Calculator object
calc = Calculator()
calc.add(10)  # Valid number
calc.add(5)   # Valid number
print(calc.result)  # Output the result

# Try calling the private method directly (will cause an error)
# calc.__validate(5)  # Uncommenting this will cause an AttributeError
````

#### Python Inner Classes

**Inner classes** are classes defined inside another class. The inner class can access the properties and methods of the outer class. They help in organizing code and grouping related classes that are only used within the context of the outer class.

---

#### Creating an Inner Class

You can create an inner class by simply defining a class inside another class.

- **Example**: 
  ```python
  class Outer:
      def __init__(self):
          self.name = "Outer Class"
  
      class Inner:
          def __init__(self):
              self.name = "Inner Class"
  
          def display(self):
              print("This is the inner class")
    ```

---

#### Accessing the Inner Class from the Outside

To access the inner class, you need to create an instance of the outer class first, then create an object of the inner class.

* **Example**:

  ```python
  outer = Outer()
  inner = outer.Inner()
  inner.display()
  ```

---

#### Accessing the Outer Class from the Inner Class

Inner classes do not automatically have access to the outer class instance. You can pass the outer class instance as a parameter to the inner class to allow access.

* **Example**:

  ```python
  class Outer:
      def __init__(self):
          self.name = "Emil"

      class Inner:
          def __init__(self, outer):
              self.outer = outer

          def display(self):
              print(f"Outer class name: {self.outer.name}")

  outer = Outer()
  inner = outer.Inner(outer)
  inner.display()
  ```

---

#### Practical Example: Using Inner Classes for Helpers

Inner classes are ideal for representing helper objects that are only used within the outer class.

* **Example**:

  ```python
  class Car:
      def __init__(self, brand, model):
          self.brand = brand
          self.model = model
          self.engine = self.Engine()

      class Engine:
          def __init__(self):
              self.status = "Off"

          def start(self):
              self.status = "Running"
              print("Engine started")

          def stop(self):
              self.status = "Off"
              print("Engine stopped")

      def drive(self):
          if self.engine.status == "Running":
              print(f"Driving the {self.brand} {self.model}")
          else:
              print("Start the engine first!")

  car = Car("Toyota", "Corolla")
  car.drive()
  car.engine.start()
  car.drive()
  ```

---

#### Multiple Inner Classes

A class can contain multiple inner classes. Each inner class can perform different tasks while still being encapsulated within the outer class.

* **Example**:

  ```python
  class Computer:
      def __init__(self):
          self.cpu = self.CPU()
          self.ram = self.RAM()

      class CPU:
          def process(self):
              print("Processing data...")

      class RAM:
          def store(self):
              print("Storing data...")

  computer = Computer()
  computer.cpu.process()
  computer.ram.store()
  ```

---


Inner classes provide a way to organize your code and encapsulate related functionality, especially when the inner class is not needed outside the scope of the outer class.



````python
# Python Inner Classes

# Outer class with an inner class
class Outer:
    def __init__(self):
        self.name = "Outer Class"  # Outer class property

    # Inner class defined inside the Outer class
    class Inner:
        def __init__(self):
            self.name = "Inner Class"  # Inner class property

        def display(self):
            print("This is the inner class")  # Display message from the inner class


# Creating an object of Outer class
outer = Outer()
print(outer.name)  # Access outer class property

# Accessing Inner class from the outside
inner = outer.Inner()  # Create an instance of Inner class
inner.display()  # Call method of Inner class


# Accessing Outer class from Inner class
class Outer:
    def __init__(self):
        self.name = "Emil"

    class Inner:
        def __init__(self, outer):
            self.outer = outer  # Passing the outer class instance

        def display(self):
            print(f"Outer class name: {self.outer.name}")  # Accessing outer class property


# Creating an object of Outer class
outer = Outer()

# Creating an object of Inner class and passing the outer class instance
inner = outer.Inner(outer)
inner.display()  # Access property of Outer class inside Inner class


# Practical example of using an inner class for encapsulation (Car and Engine)
class Car:
    def __init__(self, brand, model):
        self.brand = brand  # Car's brand
        self.model = model  # Car's model
        self.engine = self.Engine()  # Initialize the inner Engine class

    # Inner class for Engine
    class Engine:
        def __init__(self):
            self.status = "Off"  # Engine status

        def start(self):
            self.status = "Running"  # Start engine
            print("Engine started")

        def stop(self):
            self.status = "Off"  # Stop engine
            print("Engine stopped")

    # Method to drive the car
    def drive(self):
        if self.engine.status == "Running":
            print(f"Driving the {self.brand} {self.model}")
        else:
            print("Start the engine first!")


# Create a Car object and drive it
car = Car("Toyota", "Corolla")
car.drive()  # Attempt to drive without starting the engine
car.engine.start()  # Start the engine using inner class method
car.drive()  # Now the car can be driven


# Example of multiple inner classes
class Computer:
    def __init__(self):
        self.cpu = self.CPU()  # Initialize the inner CPU class
        self.ram = self.RAM()  # Initialize the inner RAM class

    # Inner class CPU
    class CPU:
        def process(self):
            print("Processing data...")  # CPU processing data

    # Inner class RAM
    class RAM:
        def store(self):
            print("Storing data...")  # RAM storing data


# Create a Computer object and use the inner classes
computer = Computer()
computer.cpu.process()  # Call method of inner CPU class
computer.ram.store()  # Call method of inner RAM class
````
