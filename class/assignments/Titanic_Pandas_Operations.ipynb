# Titanic Dataset: 20 Pandas Operations

This notebook demonstrates how to import the Titanic dataset from seaborn, export it as a CSV file, read it back as a pandas DataFrame, and perform at least 15 different pandas operations.

## 1. Import Required Libraries
Import pandas and seaborn libraries.

```python
import pandas as pd
import seaborn as sns
```

## 2. Load Titanic Dataset from Seaborn
Use seaborn's `load_dataset` function to load the Titanic dataset.

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

## 3. Export Dataset to CSV File
Export the loaded Titanic DataFrame to a CSV file using pandas' `to_csv` method.

```python
titanic.to_csv('titanic_seaborn.csv', index=False)
print('CSV file exported as titanic_seaborn.csv')
```

## 4. Read CSV File as Pandas DataFrame
Read the exported CSV file back into a pandas DataFrame using `read_csv`.

```python
df = pd.read_csv('titanic_seaborn.csv')
df.sample(10)
```

## 11. Value Counts for 'sex' Column
Use `value_counts()` on the 'sex' column to see the distribution of genders.

## 12. Groupby 'class' and Calculate Mean Age
Group the DataFrame by 'class' and calculate the mean age for each class.

```python
df.groupby('sex')['survived'].mean()
```

## 13. Sort by Fare Descending
Sort the DataFrame by the 'fare' column in descending order.

```python
df.sort_values('fare', ascending=False).head()
```

## 14. Filter Rows Where Age > 30
Filter and display rows where the 'age' column is greater than 30.

```python
df[df['age'] > 30].head()
```

## 15. Create a New Column: 'family_size'
Create a new column 'family_size' as the sum of 'sibsp' and 'parch' plus 1.

```python
df['family_size'] = df['sibsp'] + df['parch'] + 1
df[['sibsp', 'parch', 'family_size']].head()
```

## 16. Drop Rows with Missing Age
Drop rows where the 'age' column has missing values using `dropna`.

```python
df_no_missing_age = df.dropna(subset=['age'])
df_no_missing_age.shape
```

## 17. Pivot Table: Survival Rate by Sex and Class
Create a pivot table showing the mean survival rate by 'sex' and 'class'.

```python
import numpy as np
np.round(pd.pivot_table(df, values='survived', index='sex', columns='class', aggfunc='mean')*100, 1)
```

## 18. Select Specific Columns
Select and display only the 'age', 'fare', and 'survived' columns.

## 19. Set 'Passenger Class' as Index
Set the 'class' column as the DataFrame index using `set_index`.

```python
df_class_index = df.set_index('class')
df_class_index.head()
```

## 20. Reset Index
Reset the DataFrame index to default using `reset_index`.

```python
df_reset = df_class_index.reset_index()
df_reset.head()
```

## 21. Using `loc` for Label-based Selection
Select rows and columns by labels using `loc`.

```python
# Select first 3 rows and 'age', 'fare' columns using loc
# loc uses labels, so we use row indices as labels here
print(df.loc[10:15, ['age', 'fare']])
```

## 22. Using `iloc` for Position-based Selection
Select rows and columns by integer position using `iloc`.

```python
# Select first 3 rows and first 2 columns using iloc
print(df.iloc[2:8, :])
```
