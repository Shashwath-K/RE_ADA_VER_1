A company allocates employees into 3 shifts (Morning, Evening and Night) based on workload.

- You are given number of employees (N) and total tasks to be completed (T).

- Each employee can handle a maximum of 5 tasks per day (assume uniform
capacity).

- Allocate employees to shifts dynamically so that tasks are completed with minimum overtime.

- If at any point total capacity < total tasks, print “Additional Resources
Required

- If shifts overloaded (>10 employees per shift), trigger “Re-distribution
Recommended”.

```python
total_employees = int(input("Enter total employees :"))
total_tasks = int(input("Enter total tasks: "))
morning_shift = evening_shift = night_shift = 0
total_capacity = total_employees * 5
print("Output:")
if (total_capacity > total_tasks):
    for i in range(total_employees):
        if (morning_shift < 10):
            morning_shift+=1
            total_employees-=1
        elif(evening_shift < 10):
            evening_shift+= 1
            total_employees-=1
        elif(night_shift < 10):
            night_shift+= 1
            total_employees-=1
    print(f"Morning Shift: {morning_shift} employees\nEvening Shift: {evening_shift} employees\nNight Shift :{night_shift} employees")
else:
    print("Additional Resources required")
print(f"\nTotal Capacity : {total_capacity} tasks")
if(total_capacity >= total_tasks):
    print("Status : Tasks can be completed without any overtime")
```

```python
applicants = int(input("Enter the number of applicants :"))
applicants_list = dict()
for i in range(applicants):
    credit_score = int(input("Credit Score: "))
    income = int(input("Income : "))
    applicants_list[i] = {"Credit_Score":credit_score,"Income":income}
application_status = [] * len(applicants_list)
flag = 0
reject_count = 0
for i in range(len(applicants_list)):
    if(applicants_list[i]["Credit_Score"]>=750 and applicants_list[i]["Income"]>=800000):
        application_status.append("Approved")
        reject_count = 0
    elif(applicants_list[i]["Credit_Score"] in range(650,750) and applicants_list[i]["Income"]>=1000000):
        application_status.append("Manual Review")
        reject_count = 0
    elif(applicants_list[i]["Credit_Score"]<650):
        application_status.append("Rejected")
        reject_count+=1
        if(reject_count >=3):
            flag = 1
print("Output")
for i in range(len(application_status)):
    print(f"Application {i+1} : {application_status[i]}")
if(flag == 1):
    print("\nAudit Alert : Unusual pattern detected")
 
```

```python
total_datacenters = int(input("Enter number of data centers: "))
total_hours = 24
data_center = []
average_usage = []
for i in range(total_datacenters):
    data_center.append([int(input()) for i in range(total_hours)])
print(f"-------------------------------------\n AI BASED ENERGY MONITORING REPORT\n-------------------------------------")
for i in range(total_datacenters):
    print(f"Data Center {i+1}")
    average_usage.append(sum(data_center[i])/total_hours)
    print(f"Average Usage : {round(average_usage[i],2)}%")
    for j in range(total_hours):
        if(average_usage[i] > 90 and data_center[i][j] > 95):
            print(f"Critical Overload detected at Hour {j+1} (Usage : {data_center[i][j]}%)")
            break
    if(average_usage[i] in range (80,90)):
        print(f"Hardware upgrade recommended")
    if(average_usage[i] < 80):
        print(f"No Critical overload detected")
min_data_usage = min(average_usage)
print(f"\n-------------------------------------\nMost Efficient Data Center : Data Center {average_usage.index(min(average_usage))+1}\n(Lowest Avg Energy Usage:{round(min_data_usage,2)}%)")
print("-------------------------------------")

print("Conclusion:")
for i in range(total_datacenters):
    if(average_usage[i] < 80):
        print(f"Data Center {i+1} performing normally")
    else:
        print(f"Data Center {i+1} requires performance optimization and load balancing.")
```

```python
total_nurses_available = int(input("Total nurses available : "))
patients_severity = [int(input()) for i in range(total_nurses_available)]
n = len(patients_severity)
assigned_nurses = [] * n
print("Output\n")
for i in range(n):
    if(patients_severity[i] >= 8):
        assigned_nurses.append(2)
    elif(patients_severity[i] in range(5,8)):
        assigned_nurses.append(1)
    elif(patients_severity[i] < 5):
        assigned_nurses.append(0)
for i in range(n):
    print(f"Patient {i+1} -> {assigned_nurses[i]} nurse")
print(f"\nTotal assigned : {sum(assigned_nurses)} -> ", end="")
if(sum(assigned_nurses) > total_nurses_available):
    print("Shift Reallocation Recommended")
elif(sum(assigned_nurses) < total_nurses_available):
    print(f"{total_nurses_available-sum(assigned_nurses)} nurses available")
else:
    print("All nurses allocated, No Reallocation needed")
```
