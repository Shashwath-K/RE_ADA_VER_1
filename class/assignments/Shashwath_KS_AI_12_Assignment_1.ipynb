### Global Variables

A global variable is defined outside of any function or class, making it accessible throughout the entire script or module.

`````python
# Create a variable outside of a function, and use it inside the function

global_var = "I am a global variable"

def sample_func():
    print(global_var) # Accessing the global variable inside the function

sample_func() # Calling the function to see the output of the global variable
`````

`````python
# Create a variable inside a function, with the same name as the global variable
x = "awesome"

def myfunc():
  x = "fantastic"
  print("Python is " + x)

myfunc()

print("Python is " + x)
`````

### The `global` Keyword
Using the global keyword inside a function tells Python to treat the variable as if it were defined in the top-level scope of your script. This allows you to create or modify a variable that persists even after the function finishes running.



`````python
# If you use the global keyword, the variable belongs to the global scope
def myfunc():
  global g
  g = "Awesome"

myfunc()

print("Python is " + g)
`````

`````python
# To change the value of a global variable inside a function, refer to the variable by using the global keyword:

x = "awesome" # This is a global variable 

def myfunc():
  global x # Explicitly declaring that we are using the global variable x
  x = "fantastic" # This will change the value of the global variable x to "fantastic"

myfunc()

print("Python is " + x)
`````

### Python Casting
Python Casting refers to the process of converting a variable from one data type to another using built-in functions. It is commonly performed using functions like int(), float(), str(), and bool() to ensure compatibility between different data types. Casting can be implicit (automatic by the interpreter) or explicit (manually specified by the programmer). Explicit casting is often used to avoid type errors during operations. This mechanism is essential for data manipulation, input handling, and ensuring correct program behavior.

- int() - constructs an integer number from an integer literal, a float literal (by removing all decimals), or a string literal (providing the string represents a whole number)
- float() - constructs a float number from an integer literal, a float literal or a string literal (providing the string represents a float or an integer)
- str() - constructs a string from a wide variety of data types, including strings, integer literals and float literals

`````python
# Conversion to int
x = int(5.35)
y = int("2")
z = int(5)
print(f"{x}\n{y}\n{z}")
print(f"{type(x)}\n{type(y)}\n{type(z)}")

# Conversion to float
a = float(5)
b = float("3.14")
print(f"\n\n{a}\n{b}")
print(f"{type(a)}\n{type(b)}")

# Conversion to string
s1 = str(5)
s2 = str(3.14)
print(f"\n\n{s1}\n{s2}")
print(f"{type(s1)}\n{type(s2)}")

# Conversion to boolean
e = bool(0)
f = bool(1)
print(f"\n\n{e}\n{f}")
print(f"{type(e)}\n{type(f)}")
`````

### String concatenation
String Concatenation in Python is the process of merging two or more strings into a single sequence using the `+` operator. Because Python strings are **immutable**, this operation doesn't modify the originals; instead, it creates a brand-new string object in memory by copying the characters from the operands. To prevent a `TypeError`, all involved variables must be of the `str` data type, often requiring explicit type casting for numbers. While intuitive for simple tasks, joining a large volume of strings is more efficiently handled by the `.join()` method to optimize memory allocation.

`````python
first_name = "Ravi"
last_name = "Kumar"
c = first_name + last_name # Concatenating the first name and last name without space in between
print(c)

# with whitespaces
first_name = "Ravi"
last_name = "Kumar"
c = first_name + " " + last_name # Concatenating the first name and last name with a space in between
print(c)
`````

### Format Strings

In Python, string formatting provides a technical workaround for the language's strict type safety, which prevents the direct concatenation of strings and numeric types. <br/>
Modern Python primarily utilizes f-strings (formatted string literals), where a prefix f and curly braces {} allow expressions to be evaluated and cast into the string at runtime. <br/>
Alternatively, the .format() method uses placeholders to inject variables into a template, handling the necessary type conversion internally. <br/>
These methods are preferred over manual casting because they improve code readability and allow for precise control over decimal precision and padding.

### F-Strings

#### Python : Update Tuples

Tuples are immutable, which means their elements cannot be modified, added, or removed after creation.

#### Change Tuple Values

Once a tuple has been created, its values cannot be altered. This property is known as immutability.

A common workaround is to convert the tuple into a list, make the necessary changes to the list, and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
fruit_list[1] = "kiwi"
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
print(fruit_tuple)  # Updated tuple
`````

#### Add Items

Since tuples are immutable, they do not support an `append()` method. However, there are alternative ways to include new elements.

1. **Convert to a list:** Similar to modifying a tuple, you can transform it into a list, add the desired item(s), and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
# Add a new item to the list
fruit_list.append("orange")
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
`````

2. Add tuple to a tuple. You are allowed to add tuples to tuples, so if you want to add one item, (or many), create a new tuple with the item(s), and add it to the existing tuple.

Since tuples are immutable, their elements cannot be deleted. However, you can use the same workaround as used for modifying and adding items: convert the tuple into a list, perform the removal, and then convert it back into a tuple.

`````python
fruit_tuple = ("apple", "banana", "cherry")
# Convert tuple to list to modify it
fruit_list = list(fruit_tuple)
# Remove "apple" from the list
fruit_list.remove("apple")
# Convert list back to tuple
fruit_tuple = tuple(fruit_list)
print(fruit_tuple)
`````

#### Using Asterisk (*)

`````python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
# Unpack first two values, rest go into a list
green, yellow, *red = fruits
print(green)
print(yellow)
print(red)
`````

> If the asterisk is added to another variable name than the last, Python will assign values to the variable until the number of values left matches the number of variables left.

`````python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")
# Unpack first and last item, middle items go into a list
green, *tropic, red = fruits
print(green)
print(tropic)
print(red)
`````

#### Intersection
**Keep ONLY the duplicates**

The `intersection()` method will return a new set, that only contains the items that are present in both sets.

`````python
# Join set1 and set2, but keep only the duplicates

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# To keep the common elements from two sets
set3 = set1.intersection(set2)
print(set3)
`````

> You can use the & operator instead of the intersection() method, and you will get the same result.

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

# Use `&` to join two sets:
set3 = set1 & set2
print(set3)
`````

`````python
# Keep the items that exist in both set1, and set2:

set_a = {"apple", "banana", "cherry"}
set_b = {"google", "microsoft", "apple"}

# Keep only items that exist in both sets using intersection_update
set_a.intersection_update(set_b)

print(set_a)
`````

`````python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
# Use the difference_update() method to keep only the items from the first set that are not present in the other set.
set1.difference_update(set2)
print(set1)
`````

#### How If Statements Work

#### Python Elif Statement

#### The Elif Keyword

The `elif` keyword is Python's way of saying "if the previous conditions were not true, then try this condition".

The `elif` keyword allows you to check multiple expressions for True and execute a block of code as soon as one of the conditions evaluates to True.


> In this example a is equal to b, so the first condition is not true, but the elif condition is true, so we print to screen that "a and b are equal".

`````python
a = 33
b = 33
if b > a: # Tests whether b is greater than a at first
  print("b is greater than a")
elif a == b: # If the initial condtion is not True then it passes on to the elif
  print("a and b are equal")
`````

#### Multiple Elif Statements

You can have as many `elif` statements as you need. Python will check each condition in order and execute the first one that is true.

`````python
# Multiple elif statements to check and classify the grade based on the score
score = 75
# Will evaluate the score sequentially if the initial condition is not True then it moves on to the next
if score >= 90:
  print("Grade: A")
elif score >= 80:
  print("Grade: B")
elif score >= 70:
  print("Grade: C")
elif score >= 60:
  print("Grade: D")
`````

#### How Elif Works

When you use `elif`, Python evaluates the conditions from top to bottom. As soon as it finds a condition that is true, it executes that block and skips all remaining conditions.

**Important:** Only the first true condition will be executed. Even if multiple conditions are true, Python stops after executing the first matching block.


`````python
# The following program is to categorizing age groups:

age = 25

if age < 13:
  print("You are a child")
elif age < 20:
  print("You are a teenager")
elif age < 65:
  print("You are an adult")
elif age >= 65:
  print("You are a senior")
`````

#### Python Logical Operators

Logical operators are used to combine conditional statements. Python has three logical operators:

- and - Returns True if both statements are true
- or - Returns True if one of the statements is true
- not - Reverses the result, returns False if the result is true.

#### The and Operator
The and keyword is a logical operator, and is used to combine conditional statements. Both conditions must be true for the entire expression to be true.

`````python
# Test if a is greater than b, AND if c is greater than a:

a = 200
b = 33
c = 500
if a > b and c > a: # Two condtions evaluated in one-line using `and`
  print("Both conditions are True")
`````

#### The or Operator
The or keyword is a logical operator, and is used to combine conditional statements. At least one condition must be true for the entire expression to be true.

`````python
# Test if a is greater than b, OR if a is greater than c:

a = 200
b = 33
c = 500
if a > b or a > c: # Tests two conditions will return if either one of it is True
  print("At least one of the conditions is True")
`````

#### The not Operator
The not keyword is a logical operator, and is used to reverse the result of the conditional statement.

`````python
# Test if a is NOT greater than b:

a = 33
b = 200
if not a > b: # Checks the negation of the evaluated expression
  print("a is NOT greater than b")
`````

#### Combining Multiple Operators

You can combine multiple logical operators in a single expression. Python evaluates `not` first, then `and`, then `or`.


`````python
# Combining and, or, and not:

age = 25
is_student = False
has_discount_code = True
# Checks multiple condtions with mix of and, or 
if (age < 18 or age > 65) and not is_student or has_discount_code:
  print("Discount applies!")
`````

### Python Match

#### The match Statement

The `match` statement is used to perform different actions based on different conditions.

#### The Python Match Statement

Instead of writing many `if..else` statements, you can use the `match` statement.

The `match` statement selects one of many code blocks to be executed.

``` bash
Syntax:

match expression:
  case x:
    code block
  case y:
    code block
  case z:
    code block
```


**This is how it works:**

- The match expression is evaluated once.
- The value of the expression is compared with the values of each case.
- If there is a match, the associated block of code is executed.
The example below uses the weekday number to print the weekday name.

`````python
day = 4

match day:  # check value of day using pattern matching
  case 1:
    print("Monday")  # day 1
  case 2:
    print("Tuesday")  # day 2
  case 3:
    print("Wednesday")  # day 3
  case 4:
    print("Thursday")  # day 4
  case 5:
    print("Friday")  # day 5
  case 6:
    print("Saturday")  # day 6
  case 7:
    print("Sunday")  # day 7
`````

#### Default Value
Use the underscore character `_` as the last case value if you want a code block to execute when there are not other matches:

`````python
day = 4

match day:  # match the value of day
  case 6:
    print("Today is Saturday")  # Saturday case
  case 7:
    print("Today is Sunday")  # Sunday case
  case _:
    print("Looking forward to the Weekend")  # default case if no match
`````

#### Combine Values
Use the pipe character `|` as an or operator in the case evaluation to check for more than one value match in one case.

`````python
day = 4

match day:  # check value of day using pattern matching
  case 1 | 2 | 3 | 4 | 5:  # multiple values for weekdays
    print("Today is a weekday")  # Monday to Friday
  case 6 | 7:  # weekend values
    print("I love weekends!")  # Saturday and Sunday
`````

#### If Statements as Guards
You can add if statements in the case evaluation as an extra condition-check.

`````python
month = 5
day = 4

match day:  # match based on day value
  case 1 | 2 | 3 | 4 | 5 if month == 4:  # weekday in April
    print("A weekday in April")
  case 1 | 2 | 3 | 4 | 5 if month == 5:  # weekday in May
    print("A weekday in May")
  case _:  # default case if no match
    print("No match")
`````

#### Python Variable Scope

#### The Concept of Scope
A variable's reach is limited to the specific area where it was defined; this boundary is what we call **scope**. If you try to access a variable outside of its designated region, Python will not recognize it.

#### Local Scope
Whenever a variable is initialized within a function, it is confined to that function's **local scope**. 

* **Internal Access:** It can only be utilized by the code blocks located inside that specific function.
* **Isolation:** The rest of the program remains unaware of the variable's existence, ensuring that local data doesn't accidentally interfere with the wider script.

`````python
# A variable created inside a function is available inside that function:

def myfunc():
  x = 300 # x is the scope defined inside the function, which is local to a function
  print(x)

myfunc()
`````

#### Nested Functions

#### Function Within a Function
While a local variable is hidden from the main code, it remains accessible to any **nested functions** defined inside that primary function. 

#### Inner Accessibility
If you declare a variable in an outer function, any function sitting inside it inherits the right to use that variable. This creates a chain where the inner "child" function can see the outer "parent" function's data.

`````python
# The local variable can be accessed from a function within the function
def myfunc():
  x = 300  # local variable inside outer function

  def myinnerfunc():
    print(x)  # inner function can access outer function's variable

  myinnerfunc()  # call inner function

myfunc()  # call outer function
`````

#### Global Scope
A variable created in the main body of the Python code is a global variable and belongs to the global scope.

Global variables are available from within any scope, global and local.

`````python
# A variable created outside of a function is global and can be used by anyone:
x = 300 # Global variable defined outside the function scope
def myfunc():
  print(x) # prints the global scope
myfunc()
print(x)
`````

#### Naming Variables
If you operate with the same variable name inside and outside of a function, Python will treat them as two separate variables, one available in the global scope (outside the function) and one available in the local scope (inside the function).

`````python
# The function will print the local x, and then the code will print the global x:

x = 300 # global x variable 

def myfunc():
  x = 200 # variable x local to the function 
  print(x)

myfunc() # calls the function and prints 200

print(x) # prints the global value
`````

`````python
# if you use the global keyword, the variable belongs to the global scope.

x = 300 # global x variable 

def myfunc():
  global x # Assigning the current variable x to global scope 
  x = 200
  print(x)

myfunc() # calls the function and prints 200

print(x) # prints the global value this time it's assigned as 200
`````

#### Nonlocal Keyword
- The `nonlocal` keyword is used to work with variables inside nested functions.
- The `nonlocal` keyword makes the variable belong to the outer function.

`````python
# If you use the nonlocal keyword, the variable will belong to the outer function:

def myfunc1():
  x = "Jane"  # variable in outer function

  def myfunc2():
    nonlocal x  # refers to variable in enclosing function (myfunc1)
    x = "hello"  # modifies outer variable

  myfunc2()  # call inner function
  return x  # return updated value

print(myfunc1())  # print result of function call
`````

#### The LEGB Rule
Python follows the LEGB rule when looking up variable names, and searches for them in this order:

* Local - Inside the current function
* Enclosing - Inside enclosing functions (from inner to outer)
* Global - At the top level of the module
* Built-in - In Python's built-in namespace

`````python
# Understanding the LEGB rule

x = "global" # Global variable

def outer():
  x = "enclosing" # Enclosing variable inside the function
  def inner():
    x = "local" # local variable of the inner function
    print("Inner:", x)
  inner()
  print("Outer:", x)

outer()
print("Global:", x)
`````

#### Python Decorators

#### Decorator Basics
Decorators allow you to inject additional logic into an existing function without modifying its source code. They act as "wrappers" that enhance or change behavior dynamically.

#### The Structure
A decorator is essentially a function that accepts another function as an argument. It then returns a modified version of that function to be executed.



#### Implementation
To use one, define the decorator function first. You can then apply it to any target function by placing the `@decorator_name` syntax directly above the function definition.

`````python
# A basic decorator that uppercases the return value of the decorated function.

def changecase(func):  # decorator function
  def myinner():  # wrapper function
    return func().upper()  # call original function and convert result to uppercase
  return myinner  # return wrapped function

@changecase  # apply decorator to function
def myfunction():
  return "Hello Sally"  # original function returns string

print(myfunction())  # call decorated function
`````

* By placing `@changecase` directly above the function definition, the function myfunction is being `"decorated"` with the changecase function.
* The function `changecase` is the decorator.
* The function `myfunction` is the function that gets decorated.



#### Multiple Decorators
- You can use multiple decorators on one function.
- This is done by placing the decorator calls on top of each other.
- Decorators are called in the reverse order, starting with the one closest to the function.

`````python
# One decorator for upper case, and one for adding a greeting

def changecase(func): #Decorator to change the case of the function 
  def myinner():
    return func().upper()
  return myinner

def addgreeting(func): # Decorator to add the greeting before the function is called
  def myinner():
    return "Hello " + func() + " Have a good day!"
  return myinner

@changecase
@addgreeting
def myfunction():
  return "Tobias"

print(myfunction())
`````

#### Multiple Decorator Calls
A decorator can be called multiple times. Just place the decorator above the function you want to decorate.

`````python
# Using the @changecase decorator on two functions:

def changecase(func):  # decorator function
  def myinner():  # wrapper function
    return func().upper()  # convert function output to uppercase
  return myinner  # return modified function

@changecase  # apply decorator to myfunction
def myfunction():
  return "Hello Sally"  # original string

@changecase  # apply decorator to otherfunction
def otherfunction():
  return "I am speed!"  # original string

print(myfunction()) 
print(otherfunction())  
`````

#### Arguments in the Decorated Function
Functions that require arguments can also be decorated, just make sure you pass the arguments to the wrapper function.

`````python
# Functions with arguments can also be decorated:

def changecase(func):
  def myinner(x):
    return func(x).upper() # changes the function and the argument to Uppercase
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John")) # passing the name as John
`````

#### *args and **kwargs
Sometimes the decorator function has no control over the arguments passed from decorated function, to solve this problem, add `(*args, **kwargs)` to the wrapper function, this way the wrapper function can accept any number, and any type of arguments, and pass them to the decorated function.

`````python
# Secure the function with *args and **kwargs arguments:

def changecase(func):  # decorator function
  def myinner(*args, **kwargs):  # wrapper that accepts any arguments
    return func(*args, **kwargs).upper()  # call original function and convert result to uppercase
  return myinner  # return wrapped function

@changecase  # apply decorator
def myfunction(*nam):
  return "Hello " + nam[1]  # return greeting string

print(myfunction("John", "Mary"))  
`````

#### Decorators With Arguments

#### Passing Parameters
To create a decorator that accepts its own arguments, you must introduce an additional **wrapper level**. This essentially involves nesting a second function inside the decorator to handle the extra input.


#### The Extra Layer
The outermost function captures the decorator's arguments, the middle function captures the target function, and the innermost function executes the actual logic. This structure allows you to customize the decorator's behavior for different scenarios.

`````python
# A decorator factory that takes an argument and transforms the casing based on the argument value.

def changecase(n):  # outer function that takes a parameter
  def changecase(func):  # decorator function
    def myinner():  # wrapper function
      if n == 1:  # check condition from outer function
        a = func().lower()  # convert result to lowercase
      else:
        a = func().upper()  # convert result to uppercase
      return a  # return modified result
    return myinner  # return wrapper
  return changecase  # return decorator

@changecase(1)  # apply decorator with argument
def myfunction():
  return "Hello Linus"  # original function

print(myfunction()) 
`````

#### Arguments in the Decorated Function
Functions that require arguments can also be decorated, just make sure you pass the arguments to the wrapper function.

`````python
def my_gen():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    # This block executes when the generator is closed or exhausted
    print("Generator closed")

gen = my_gen()

# Fetch the first value
print(next(gen))

# Manually terminate the generator
gen.close()
`````

### Python Try Except

- The try block lets you test a block of code for errors.
- The except block lets you handle the error.
- The else block lets you execute code when there is no error.
- The finally block lets you execute code, regardless of the result of the try- and except blocks.

### Exception Handling
When an error occurs, or exception as we call it, Python will normally stop and generate an error message.

These exceptions can be handled using the try statement

`````python
# The try block will generate an exception, because x is not defined:

try:
  print(x)
except:
  print("An exception occurred")
  
`````

### Many Exceptions
You can define as many exception blocks as you want, e.g. if you want to execute a special block of code for a special kind of error.

`````python
# Print one message if the try block raises a NameError and another for other errors:

try:
  print(x)
except NameError:
  print("Variable x is not defined")
except:
  print("Something else went wrong")
`````

`````python
# In this example, the try block does not generate any error:

try:
  print("Hello")
except:
  print("Something went wrong")
else:
  print("Nothing went wrong")
`````

`````python
# The finally block, if specified, will be executed regardless if the try block raises an error or not.
try:
  print(x)
except:
  print("Something went wrong")
finally:
  print("The 'try except' is finished")
`````

`````python
# Try to open and write to a file that is not writable:

try:
  f = open("demofile.txt")
  try:
    f.write("Lorum Ipsum")
  except:
    print("Something went wrong when writing to the file")
  finally:
    f.close()
except:
  print("Something went wrong when opening the file")
`````

### Raise an exception

To throw (or raise) an exception, use the raise keyword.

- The raise keyword is used to raise an exception.
- You can define what kind of error to raise, and the text to print to the user.

`````python
# Raise an error and stop the program if x is lower than 0:

x = -1

if x < 0:
  raise Exception("Sorry, no numbers below zero")
`````

`````python
# Raise a TypeError if x is not an integer:

x = "hello"

if not type(x) is int:
  raise TypeError("Only integers are allowed")
`````

### String Formatting

F-String was introduced in Python 3.6, and is now the preferred way of formatting strings.

Before Python 3.6 we had to use the `format()` method.

### F-Strings
F-string allows you to format selected parts of a string.

To specify a string as an f-string, simply put an f in front of the string literal, like this:

`````python
text_message = f"The price is 50 dollars"  # creating a formatted string
print(text_message)
`````

In Python formatting, a modifier (placed after a colon `:`) acts as a technical directive for the interpreter on how to present a value's data. For instance, `.2f` is a type specifier that forces a float into a fixed-point representation with exactly two decimal places, handling rounding and padding automatically during the string conversion.

`````python
# Display the price with 2 decimals
amount = 59  # price value
formatted_text = f"The price is {amount:.2f} dollars"  # format to 2 decimal places
print(formatted_text)

# Display the value 95 with 2 decimals
amount = 95  # price value
formatted_text = f"The price is {amount:.2f} dollars"  # format to 2 decimal places
print(formatted_text)
`````

### Perform Operations in F-Strings
- You can perform Python operations inside the placeholders.
- You can do math operations.

`````python
# Perform a math operation in the placeholder, and return the result:

txt = f"The price is {50 * 59} dollars"
print(txt)
`````

`````python
# Add taxes before displaying the price:

price = 59
tax = 0.25
txt = f"The price is {price + (price * tax)} dollars"
print(txt)
`````

`````python
item_price = 49  # given price

result_text = f"It is very {'Expensive' if item_price > 50 else 'Cheap'}"  # condition inside f-string

print(result_text)
`````

### Execute Functions in F-Strings
You can execute functions inside the placeholder.


`````python

# Use the string method upper()to convert a value into upper case letters:
snack = "mangoes"  # fruit name
message = f"I enjoy {snack.upper()}"  # convert to uppercase
print(message)
`````
