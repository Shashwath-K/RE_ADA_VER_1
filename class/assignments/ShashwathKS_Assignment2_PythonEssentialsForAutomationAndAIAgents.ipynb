## 11.Automated Hostel Attendance Checker

The hostel warden wants a small system to quickly check if a student meets the required
attendance to appear for end-semester exams. The program should accept total classes
conducted, classes attended, and medical leave count. If attendance percentage (considering
medical leave) is ≥ 75%, print “Eligible”; otherwise print “Not Eligible.”

<details>
<summary>Show More </summary>

**Test Cases:**
``` bash
Input: 120, 85, 5 → Eligible
Input: 100, 60, 0 → Not Eligible
```

````python
# Function to calculate attendance of student
def calculateAttendance(total_classes, classes_attended, medical_leave):
    attendance_percentage = ((classes_attended + medical_leave)/total_classes)*100
    if attendance_percentage >= 75.0:
        print("Student is eligible")
    else:
        print("Student is Not eligible")

if __name__ == "__main__":
    try:
        # Taking input from the user for total classes, classes attended and medical leave
        total_classes = int(input("Enter the total number of classes : "))
        classes_attended = int(input("Enter the total classes attended : ")) 
        medical_leave = input("Medical leave taken ?")  # Yes or No
        total_medical_leave = 0 
        if medical_leave.lower() == "yes":
            total_medical_leave = int(input("Enter the number of leave taken : "))
        elif medical_leave.lower() == "no":
            total_medical_leave = 0
        else:
            print("Invalid input for medical leave. Please enter 'Yes' or 'No'.")
        # Function call to claculate the attendance
        calculateAttendance(total_classes, classes_attended, total_medical_leave)
    except Exception as e:
        print("Error : ", e)
        
````

# 12. Cafeteria Purchase Validator
A cafeteria is implementing an automated billing system. Write a Python program where a student
enters the menu item, quantity, and available wallet balance. Using conditionals and loops, ensure
the student can only complete the purchase if sufficient balance exists. The system should
continue allowing purchases until the wallet balance becomes insufficient.

<details>
<summary> Show more</summary>

**Test Cases:**

``` bash
Input: Item=Tea(10), Qty=3, Wallet=50 → Remaining = 20
Input: Item=Sandwich(40), Qty=2, Wallet=50 → Insufficient balance
```

````python
def cafeteriaBill(item_price,qty, wallet):
    calculate_total = item_price * qty
    update_wallet = wallet - calculate_total
    if (update_wallet < 0):
        return "Insufficient balance"
    return (calculate_total, update_wallet)

if __name__ == "__main__":
    items = [['Tea', 10],['Sandwich', 40],['Donut', 70],['Coffee', 15]]
    print(f"item.No\t Item Name\t Price")
    item_price = 0
    for i in range(len(items)):
        print(f'{i+1}.\t {items[i][0]}\t\t {items[i][1]}')
    try:
        select_item = int(input("Enter the item number: "))
        quantity = int(input("Enter the Quantity: "))
        wallet_balance = float(input("Enter wallet balance: "))
        item_price = items[select_item-1][1]
        result = cafeteriaBill(item_price, quantity, wallet_balance)
        print("Total Bill: ", result[0])
        print("Remaining Balance: ", result[1])
    except Exception as e:
        print("Invalid", e)
````
