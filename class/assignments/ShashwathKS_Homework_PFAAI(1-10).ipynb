# 1.Arithmetic Chatbot for University Helpdesk
<p>A university helpdesk is building a simple chatbot that helps students perform basic arithmetic
operations when calculators are not allowed during online tests. The chatbot must ask the user for
two numbers and an operation such as add, subtract, multiply, or divide. Based on the operation
entered, the bot should compute the result and display it neatly to the user. Ensure that division
handles floating-point values and avoids division by zero</p>

```python
def chooseOperation(choice):
  x = float(input("Enter 1st number : "))
  y = float(input("Enter 2nd number : "))
  if(choice == "1"):
      return x + y
  elif(choice == "2"):
      return x - y
  elif(choice == "3"):
      return x * y
  elif(choice == "4"):
      try :
          return x / y
      except :
          print("Cannot divide number by 0")
  elif(choice == "5"):
      return x % y

choice = 0
while(choice != "0"):
  print(f"Arithmetic Chatbot for University Helpdesk\n{"-"*20}\n1.Addition\n2.Subtraction\n3.Multiplication\n4.Division\n5.Modulus\n0.Exit\n{"-"*20}")
  choice = input("Enter your choice : ")
  if choice == "0":
      break
  print(f"{chooseOperation(choice)}\n")

```

# 2.Distance Calculator for Campus Security Robot
<p>Your college is deploying a campus security robot that moves between coordinate points. You
must create a Python program that calculates the Euclidean distance between two given 2D points
to help the robot determine how far it must travel. The program should take coordinates as input
and compute the distance using the standard formula. It should work for multiple test inputs.</p>

```python
import math
def distanceCovered(x,y):
    res = math.sqrt((x[1]-x[0])**2 + (y[1]-y[0])**2)
    return res
print(f"{"*"*50}\nDistance calculator for campus security robot\n{"*"*50}\n")
x = [0]*2
y = [0]*2
for i in range(2):
    x[i] = int(input(f"Enter x{[i+1]} :"))
    y[i] = int(input(f"Enter y{[i+1]} :"))
print(f"Distance covered: {distanceCovered(x,y)}")
```

# 3.Student Scholarship Eligibility Checker
A scholarship committee wants a small Python tool to check whether a student is eligible for a
merit scholarship. The committee checks a student’s total marks, attendance percentage, and
participation score. Write a program that accepts these three values and prints whether the
student is eligible. Marks ≥ 85, attendance ≥ 75%, and participation ≥ 50 qualify a student.

```python
def eligibilityCheck(student_details):
    for i in range(len(student_details)):
        if student_details[i]["Marks"]>=85 and student_details[i]["Attendance"]>=75 and student_details[i]["Participation"]>=50:
            return "Eligible"
        else:
            return "Not Eligible"
print(f"{"*"*50}\nStudent Scholarship eligibility Checker\n{"*"*50}\n")
student_details = {}
num = int(input("Enter the Number of candidates: "))
for i in range(num):
    marks = int(input(f'Enter marks of student {i+1}'))
    attendance = int(input(f'Enter attendance of student {i+1} in %'))
    participation = int(input(f'Enter participation of student {i+1}'))
    student_details[i] = {"Marks":marks,"Attendance":attendance,"Participation":participation}
print(f"Stuent is {eligibilityCheck(student_details)} for scholarship")
```

# 4.Hostel Mess Bill Splitter
In a college hostel, students often split their monthly mess bills. Write a Python program that
asks for the total bill amount and the number of students sharing it. The system should validate
inputs, ensure students > 0, and compute how much each student must pay.

```python
def billSplitter(n,t):
    res = t // n
    return res
if __name__ == "__main__":
    print(f"{"*"*50}\nHostel Mess bill splitter\n{"*"*50}\n")
    num_students = int(input("Enter number of students: "))
    total_amount = int(input("Enter total amount :")) 
    print(f"{billSplitter(num_students,total_amount)} per student")
```

# 5.Course Registration Token Generator
During course registration, every student receives a unique ID token generated by combining
their name initials, department code, and last four digits of their phone number. Write a Python
program to automatically generate this token in a formatted manner. The program must convert
all tokens to upper case

```python
def codeGen(name, dept, phone):
    initials = name[(name.rindex(" "))+1]
    ph_tkn = phone[(len(phone)-4):]
    return (f"{name[0]}{initials}_{dept}_{ph_tkn}")
if __name__ == "__main__":
    print(f"{"*"*50}\nCourse Registration Token generator\n{"*"*50}\n")
    name = input("Enter name : ")
    dept = input("Enter Department : ")
    phn  = input("Enter phone number : ")
    print(f"Generated code : {codeGen(name,dept,phn)}")
```

# 6.Team Selection Based on Performance Score
A coding club wants to select top performers for a hackathon. Write a Python script that takes
the student name and performance score (0–100). Based on the score, categorize them as
“Elite”, “Proficient”, “Average”, or “Needs Improvement.”

```python
def candidateSelection(candidates):
    if candidates["Score"] > 90:
        return "Elite"
    elif candidates["Score"] in range(75,90):
        return "Proficient"
    elif candidates["Score"] in range(50,75):
        return "Average"
    else:
        return "Needs Improvement"

if __name__ == "__main__":
    print(f"{"*"*50}\nTeam Selection Based on Performance Score\n{"*"*50}\n")
    num = int(input("Enter the number of candidates: "))
    candidates = {}
    for i in range(num):
        name = input("Enter name : ")
        perform_score = int(input("Enter performance score : "))
        candidates[i] = {"Name": name, "Score": perform_score}
        print(f"{candidates[i]["Name"]} : {candidateSelection(candidates[i])}")
```

# 7.ATM Mini Simulator for Banking Lab
A campus fintech lab is simulating an ATM system. Write a Python program that takes the user's
balance and withdrawal amount and checks if the withdrawal is allowed. It must ensure the
amount is not more than the balance and is not negative. Display the updated balance if the
transaction succeeds

```python
def analyzeTransaction(bal, withdrawl):
    if withdrawl<=bal:
        print(f"Transaction successful -> New Balance : {bal-withdrawl}")
    else:
        print("Insufficient funds")
if __name__ == "__main__":
    print(f"{"*"*50}\nATM Mini Simulator for Banking Lab\n{"*"*50}\n")
    balance = int(input("Enter Balance : "))
    withdrawl = int(input("Enter the amount to be withdrawed : "))
    analyzeTransaction(balance,withdrawl)
```

# 8.Library Fine Calculator
The library charges a fine of ₹5 per day for late returns. Write a program that asks the user for
the number of days a book is overdue and displays the total fine. If days ≤ 0, no fine should be
charged.

```python
def calculateFine(num):
    if num == 0:
        return "No fine"
    return num*5
if __name__ == "__main__":
    print(f"{"*"*50}\nLibrary Fine Calculator\n{"*"*50}\n")
    num = int(input("Enter the number of days the book is overdue : "))
    print(f"Overdue days : {num} -> Penalty : {calculateFine(num)}")
```

# 9.Basic BMI Calculator for Health Club
The college gym needs a BMI calculator to assess student health. Input weight (kg) and height
(m) and compute BMI = weight / height². Categorize as Underweight (<18.5), Normal (18.5–
24.9), Overweight (25–29.9), Obese (30+)

```python
def calculateBMI(weight, height):
    bmi_res = round((int(weight)/float(height)**2),2)
    if bmi_res < 18.5:
        print(f"BMI = {bmi_res} (Underweight)")
    elif 18.5 <= bmi_res <= 24.9:
        print(f"BMI = {bmi_res} (Normal)")
    elif 25.0 <= bmi_res <= 29.9:
        print(f"BMI = {bmi_res} (Overweight)")
    elif bmi_res > 30:
        print(f"BMI = {bmi_res} (Obese)")
    else:
        print("Invalid")
if __name__ == "__main__":
    print(f"{"*"*50}\nBasic BMI Calculator for Health Club\n{"*"*50}\n")
    weight = input("Enter Weight : ")
    height = input("Enter Height : ")
    calculateBMI(weight, height)
```

# 10.Simple ID Verification System
Develop a Python program that accepts a student name and college ID. It must verify that ID is
exactly 10 characters long and contains both letters and digits. Display “Valid ID” only if all
conditions are met.

```python
import re
def verifyStudentId(student_id):
    re_pattern = re.compile(r'^[A-Z]{2}[0-9]{4}[A-Z][0-9]{3}$')
    if len(student_id) == 10 and re_pattern.match(student_id):
        print(f"ID : {student_id} is Valid")
    else:
        print(f"ID : {student_id} is Invalid")
if __name__ == "__main__":
    print(f"{"*"*50}\nSimple ID Verification System\n{"*"*50}\n")
    name = input("Enter name : ")
    student_id = input("Enter ID")
    verifyStudentId(student_id)
```
