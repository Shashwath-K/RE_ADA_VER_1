```python
# Create a class named Playlist
class Playlist:
  # The constructor method to initialize a playlist object.
  def __init__(self, name):
    # 'name' is an instance attribute to store the playlist's name.
    self.name = name
    # 'songs' is an instance attribute, an empty list to store song titles.
    self.songs = []

  # A method to add a song to the playlist.
  def add_song(self, song):
    # Use the append() method to add the song to the end of the 'songs' list.
    self.songs.append(song)
    # Print a confirmation message indicating which song was added.
    print(f"Added: {song}")

  # A method to remove a song from the playlist.
  def remove_song(self, song):
    # Check if the specified song exists in the 'songs' list.
    if song in self.songs:
      # If it exists, remove the first occurrence of the song.
      self.songs.remove(song)
      # Print a confirmation message indicating which song was removed.
      print(f"Removed: {song}")

  # A method to display all songs in the playlist.
  def show_songs(self):
    # Print the name of the playlist.
    print(f"Playlist '{self.name}':")
    # Loop through each 'song' in the 'songs' list.
    for song in self.songs:
      # Print each song on a new line, prefixed with a dash.
      print(f"- {song}")

# Create an instance of the Playlist class named 'my_playlist' with the name "Favorites".
my_playlist = Playlist("Favorites")
# Call the add_song method to add "Bohemian Rhapsody" to the playlist.
my_playlist.add_song("Bohemian Rhapsody")
# Call the add_song method again to add "Stairway to Heaven".
my_playlist.add_song("Stairway to Heaven")
# Call the show_songs method to display the final list of songs.
my_playlist.show_songs()
```

```python
# Create a class named Person
class Person:
  # The constructor method initializes the object.
  def __init__(self, name, age):
    # 'name' is a public attribute.
    self.name = name
    # '__age' is a private attribute.
    self.__age = age

  # A "getter" method to provide controlled access to the private '__age' attribute.
  def get_age(self):
    return self.__age

  # A "setter" method to provide controlled modification of the private '__age' attribute.
  def set_age(self, age):
    # Include validation logic: only update if the new age is a positive number.
    if age > 0:
      # If valid, update the private '__age' attribute.
      self.__age = age
    else:
      # If invalid, print an error message.
      print("Age must be positive")

# Create an instance of the Person class.
p1 = Person("Tobias", 25)
# Use the getter method to access and print the initial age.
print(p1.get_age())

# Use the setter method to update the age to a new value.
p1.set_age(26)
# Use the getter method again to access and print the updated age.
print(p1.get_age())
```

```python
# Create a class named Student
class Student:
  # The constructor method initializes the student's name and a private grade.
  def __init__(self, name):
    # 'name' is a public attribute.
    self.name = name
    # '__grade' is a private attribute, initialized to 0.
    self.__grade = 0

  # A "setter" method to update the student's grade.
  def set_grade(self, grade):
    # It includes validation to ensure the grade is within a valid range.
    if 0 <= grade <= 100:
      # If the grade is valid, update the private '__grade' attribute.
      self.__grade = grade
    else:
      # If the grade is invalid, print an error message.
      print("Grade must be between 0 and 100")

  # A "getter" method to retrieve the student's grade.
  def get_grade(self):
    # It provides controlled access to the private '__grade' attribute.
    return self.__grade

  # A method to determine the student's academic status based on their grade.
  def get_status(self):
    # The logic is based on the value of the private '__grade' attribute.
    if self.__grade >= 60:
      return "Passed"
    else:
      return "Failed"

# Create an instance of the Student class.
student = Student("Emil")
# Use the setter method to assign a grade to the student.
student.set_grade(85)
# Use the getter method to retrieve and print the student's grade.
print(student.get_grade())
# Call the get_status method to determine and print the student's status.
print(student.get_status())
```

```python
# Create a class named Calculator
class Calculator:
  # The constructor method initializes the calculator's state.
  def __init__(self):
    # 'result' is a public attribute to store the current calculation result.
    self.result = 0

  # A double underscore prefix makes this a "private" method.
  # It's intended for internal use within the class and not accessible directly from outside.
  def __validate(self, num):
    # Check if the input 'num' is either an integer or a float.
    if not isinstance(num, (int, float)):
      return False
    return True

  # A public method to add a number to the current result.
  def add(self, num):
    # Use the internal '__validate' method to check the input number.
    if self.__validate(num):
      # If the number is valid, add it to the result.
      self.result += num
    else:
      # If the number is invalid, print an error message.
      print("Invalid number")

# Create an instance of the Calculator class.
calc = Calculator()
# Call the public 'add' method to add 10 to the result.
calc.add(10)
# Call the 'add' method again to add 5.
calc.add(5)
# Print the final value of the 'result' attribute.
print(calc.result)

# Attempting to call the private '__validate' method directly from outside the class
# would raise an AttributeError due to name mangling.
# calc.__validate(5)
```

```python
# Create an outer class named Car
class Car:
  # The constructor for the Car class.
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
    # Create an instance of the inner Engine class and assign it as an attribute to the Car instance.
    self.engine = self.Engine()

  # Define a nested or inner class named Engine.
  # This logically groups the Engine class within the Car class, as it's a related component.
  class Engine:
    # The constructor for the inner Engine class.
    def __init__(self):
      # Initialize the engine's status to "Off".
      self.status = "Off"

    # A method to change the engine's status to "Running".
    def start(self):
      self.status = "Running"
      print("Engine started")

    # A method to change the engine's status to "Off".
    def stop(self):
      self.status = "Off"
      print("Engine stopped")

  # A method in the outer Car class.
  def drive(self):
    # Check the status of the nested engine object.
    if self.engine.status == "Running":
      # If the engine is running, print a driving message.
      print(f"Driving the {self.brand} {self.model}")
    else:
      # If the engine is off, prompt to start it.
      print("Start the engine first!")

# Create an instance of the outer Car class.
car = Car("Toyota", "Corolla")
# Try to drive the car; the engine is initially off.
car.drive()
# Access the 'engine' object within the 'car' instance and call its 'start' method.
car.engine.start()
# Try to drive the car again; now the engine is running.
car.drive()
```

```python
# Create an outer class named Computer.
class Computer:
  # The constructor method for the Computer class.
  def __init__(self):
    # Create an instance of the inner CPU class and assign it to the 'cpu' attribute.
    self.cpu = self.CPU()
    # Create an instance of the inner RAM class and assign it to the 'ram' attribute.
    self.ram = self.RAM()

  # Define a nested class for the CPU.
  class CPU:
    # A method within the CPU class to simulate data processing.
    def process(self):
      print("Processing data...")

  # Define another nested class for the RAM.
  class RAM:
    # A method within the RAM class to simulate data storage.
    def store(self):
      print("Storing data...")

# Create an instance of the Computer class. This also creates the inner CPU and RAM objects.
computer = Computer()
# Access the 'cpu' object within the 'computer' instance and call its 'process' method.
computer.cpu.process()
# Access the 'ram' object within the 'computer' instance and call its 'store' method.
computer.ram.store()
```
