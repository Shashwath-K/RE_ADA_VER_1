# Titanic Dataset: 30 Matplotlib Visualizations

This notebook demonstrates 30 different charts and visualizations using matplotlib (and seaborn for some advanced plots) with the seaborn Titanic dataset.

## 1. Import Required Libraries
Import seaborn, matplotlib.pyplot, pandas, and numpy.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set style for seaborn
sns.set(style="whitegrid")
%matplotlib inline
```

## 2. Load the Seaborn Titanic Dataset
Load the Titanic dataset using seaborn's `load_dataset` function.

```python
titanic = sns.load_dataset('titanic')
titanic.head()
```

## 3. Basic Data Exploration
Display the first few rows and summary statistics of the dataset.

```python
# Display first 5 rows
titanic.head()
```

```python
# Display summary statistics
titanic.describe(include='all')
```

## 4. Bar Plot: Survival Count
Create a bar plot showing the count of survivors and non-survivors.

```python
plt.figure(figsize=(8,5))
titanic['survived'].value_counts().plot(kind='bar', color=['salmon', 'skyblue'])
plt.title('Survival Count')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.xticks([0,1], ['No', 'Yes'], rotation=0)
plt.show()
```

## 5. Bar Plot: Survival by Sex
Create a bar plot showing survival counts grouped by sex.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='sex', hue='survived', palette=['salmon', 'skyblue'])
plt.title('Survival by Sex')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## 6. Bar Plot: Survival by Passenger Class
Create a bar plot showing survival counts grouped by passenger class.

```python
plt.figure(figsize=(6,4))
sns.countplot(data=titanic, x='pclass', hue='survived', palette=['salmon', 'skyblue'])
plt.title('Survival by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.show()
```

## 7. Histogram: Age Distribution
Plot a histogram of the age distribution of passengers.

```python
plt.figure(figsize=(7,4))
plt.hist(titanic['age'].dropna(), bins=30, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

## 8. Histogram: Fare Distribution
Plot a histogram of the fare distribution.

```python
plt.figure(figsize=(7,4))
plt.hist(titanic['fare'].dropna(), bins=30, color='orange', edgecolor='black')
plt.title('Fare Distribution')
plt.xlabel('Fare')
plt.ylabel('Count')
plt.show()
```

## 9. Box Plot: Age by Survival
Simple box plot of age grouped by survival status.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='survived', y='age', data=titanic)
plt.title('Age by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.show()
```

## 10. Box Plot: Fare by Passenger Class
Simple box plot of fare grouped by passenger class.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='pclass', y='fare', data=titanic)
plt.title('Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

## 11. Bar Plot: Count by Embarked Port
Simple bar plot of passengers by port of embarkation.

```python
plt.figure(figsize=(6,4))
titanic['embarked'].value_counts().plot(kind='bar', color='lightgreen')
plt.title('Count by Embarked Port')
plt.xlabel('Embarked')
plt.ylabel('Count')
plt.show()
```

## 12. Pie Chart: Survival Rate
Simple pie chart showing the proportion of survivors and non-survivors.

```python
plt.figure(figsize=(5,5))
titanic['survived'].value_counts().plot.pie(labels=['No', 'Yes'], autopct='%1.1f%%', colors=['salmon', 'skyblue'])
plt.title('Survival Rate')
plt.ylabel('')
plt.show()
```

## 13. Pie Chart: Passenger Class Distribution
Pie chart showing the distribution of passenger classes.

```python
plt.figure(figsize=(5,5))
titanic['pclass'].value_counts().plot.pie(labels=['3rd', '1st', '2nd'], autopct='%1.1f%%', colors=['#ffcc99','#99ccff','#99ff99'])
plt.title('Passenger Class Distribution')
plt.ylabel('')
plt.show()
```

## 14. Scatter Plot: Age vs Fare
Simple scatter plot of age versus fare.

```python
# Create a scatter plot of fare vs age
plt.figure(figsize=(10, 6))
sns.scatterplot(data=titanic, x='age', y='fare')

# Adding titles and labels
plt.title('Titanic: Fare vs Age Scatter Plot')
plt.xlabel('Age')
plt.ylabel('Fare')

# Display the plot
plt.show()
```
