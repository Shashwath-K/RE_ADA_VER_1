## Artificial Neurons and Networks


- Artificial Neural Networks (ANN) are among the oldest and most intensely studied Machine Learning approaches
- Inspired by biological neural networks, they mimic the learning process of animals and humans.
- However, the model is very coarse, and several improvements to the basic approach have even abandoned the biological analogy

**Advantage**
- Highly flexible & good performance 

**Disadvantage**
- Black-box models not easy to interpret

# FeedForward Neural Networks

- Forward Propagation: In a feedforward network, data flows from the input layer, through the hidden layers, to the output layer.
- No Feedback Loops: Unlike recurrent neural networks (RNNs), FNNsdon't have feedback connections; information moves in one direction, from input to output.
- Activation Functions: Each neuron applies an activation function (such as ReLU, Sigmoid, or Tanh) to introduce non-linearity and help the network model complex patterns
