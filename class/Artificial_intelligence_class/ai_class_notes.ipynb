# Data Analytics



### How to select the right chart
- Choosing the right chart requires identifying the core message (comparison, trend, relationship, or composition) and understanding the data structure.

- Use bar charts for comparisons, line charts for trends over time, and scatter plots for relationships.

- Keep charts simple, avoid clutter, and ensure the visualization directly answers the primary question.


1. Want to compare values?
- Charts are perfect for comparing one or many value sets, and they can easily show the low and high values in the data sets.
- Use Bar/Column charts for comparing items, Horizontal Bar for ranking data, and Grouped Bar for comparing qategories.
  - Use these charts to show comparisons:
    -  Column/bar
    -  Circular area
    -  Line
    -  Scatter plot
    -  Bullet

2. Want to show the composition of something?
- To show how individual parts make up the whole of something (such as the device used for mobile visitors to your website, or total sales broken down by sales rep)
- Pie or Donut charts work for simple parts-of-a-whole, but Stacked Bar charts are better for comparing compositions across categories
- Use these charts to show composition:
  - Pie, Stacked bar, Stacked column, Area

3. Want to understand the distribution of your data?
- Distribution charts help you to understand outliers, the normal tendency, and the range of information in your values.
- Scatter plots or Bubble charts are best for showing correlations between variables.
- Use these charts to show distribution:
  - Scatter plot
  - Line
  - Column
  - Bar

4. Interested in analyzing trends in your data set?
- If you want more information about how a data set performed during a specific period, there are specific chart types that do this extremely well.
- Line charts are ideal for displaying data changes qver time, while Area charts can show trends but should be used carefully.
- Use these charts to analyze trends:
   - Line, Dual-axis line, Column

5. Want to better understand the relationships among value sets?
- Relationship charts are designed to show how one variable relates to one or many different variables. You could show how something positively affects (or has no effect, or negatively affects) another variable.
 - Use these charts to show relationships:
   - Scatter plot
   - Bubble
   - Line

#### Z-scores or standard distribution
Z-score is a numerical measure of how far an individual score is away from the mean score, within a normal distribution.

This is a standardized version of the normal distribution where the mean is set to zero and the standard deviation to 1. With this you can calculate the probability of a number being above (or below) a certain value. Z score (+ve, 0, -ve)

Where:
1. z is the z-score you're trying to find.
2. x is the specific data point you're interested in.
3. u (mu) is the mean of the entire data set.
4. o (sigma) is the standard deviation of the data set.

Z-scores are calculated using a formula:
> z = (x - u) / o,

### Probability Distribution

##### Introduction to Probability

Decoding Uncertainty: The Language of Chance
- **Key Concept:** Probability is the numerical measure of the likelihood that an event will occur.
  - **The Scale:** * 0: Impossible (The event cannot happela).
  - 0.5: Even Chance (Like a fair coin toss).
  - 1: Certain (It must happen).
  - **The Golden Rule:** The sum of all probabilities in a sample space always equals 1

##### Theory vs. Reality
- The Law of Large Numbers states that as the number of trials increases, the Experimental Probability converges to the Theoretical Probability.
- Theoretical Probability: Based on reasoning/math.

- Number of favorable outcomes / Total number of possible outcomes

- Experimental Probability: Based on actual trials or historical data.

#### The power of complement
- Title: Working Backward: P(Not A)
- Concept: Sometimes it is easier to calculate the probability of an event not happening.
  - Formula: P(A) = 1 - P(A')
- Real-World Example: In Cybersecurity (your area of mterest), if the probability of a system being breached is 0.001, the probability of the system remaining secure is 1 - 0.001 = 0.999.

### Why Probability Matters in Your Career
- **Machine Learning (NLP/Vision):** Softmax layers in Neural Networks output probabilities for classification (e.g., "Is this a face or a car?").
- **FinTech:** Credit scoring models use logistic regression to output the probability of a user defaulting on a loan.
- **Digital Twins (ASTRON-DT):** Predicting the probability of an astronaut experiencing a health anomaly based on biometric sensor data.

### What is Probability distribution?
A probability distribution is a function that describes how the values of a random variable are distributed. It gives the probability that a random variable takes on a particular value (or range of values).



## Types of Distributions
1. Discrete Distributions
2. Continuous Distributions

### Discrete Distributions
A discrete probability distribution describes the probabilities of the outcomes of a discrete random variable-a variable that can take only finite or countably infinite values.

**Examples:**
- Number of heads in 5-coin tosses
- Number of students absent in a class
- Number of cars passing a signal in 10 minutes

#### Bernoulli distribution

The Bernoulli distribution is a discrete probability distribution that models a random variable with two possible outcomes, typically
labeled as "success" (1) and "failure" (0).
- The probability of success is denoted by p, and the probability of failure is 1-p.

**Examples:**
- Flipping a coin (Heads/Tails).
- A single question on a multiple-choice test (correct/incorrect).
- Whether a customer will make a purchase (yes/no).
- Whether a machine part will fail during a specific time period.

#### Binomial Distribution

- A binomial distribution is a discrete probability distribution that describes the probability of obtaining a certain number of successes in a fixed number of independent trials, where each trial has only two possible outcomes (success or failure).
- Note: It can be interpreted as Number of successes in n independent Bernoulli trials.

- It's defined by two parameters: n, the number of trials, and p, the probability of success in a single trial.
- Mean = np, Variance = np(1-p)
- Example: Probability of getting 3 heads in 5 coin tosses.

#### Poisson distribution

- The Poisson distribution is a discrete probability distribution that expresses the probability of a given number of events occurring in a fixed interval of time.
- Number of events in a fixed time/space interval, for rare events.

#### Normal Distribution

A normal distribution, also known as a Gaussian distribution or bell curve, is a probability distribution that is symmetric about the mean, with most values clustering around the mean and tapering off towards the extremes. It's a fundamental concept in statistics, used to model various phenomena in nature and social sciences.

- The distribution is symmetrical around the mean, meaning the left and right sides are mirror images of each other.
> Ex: Natural measurements like height, IQ scores, test marks, etc.
