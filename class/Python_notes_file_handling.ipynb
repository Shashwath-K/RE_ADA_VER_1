#### File handling modes
| Mode | Description              | File must exist? | Existing content | Pointer position on open                          |
|------|--------------------------|------------------|------------------|--------------------------------------------------|
| `r`  | Read only                | Yes              | Intact           | Beginning                                        |
| `w`  | Write only               | No               | Overwritten      | Beginning                                        |
| `a`  | Append only              | No               | Intact           | End                                              |
| `x`  | Exclusive creation       | No               | N/A              | Beginning                                        |
| `r+` | Read and write           | Yes              | Intact           | Beginning                                        |
| `w+` | Write and read           | No               | Overwritten      | Beginning                                        |
| `a+` | Append and read          | No               | Intact           | End (for writing), Beginning (for reading)       |


#### Read types in file handling
| Method        | Functionality                                                                 | Return Type        | Memory Efficiency                                                                 |
|---------------|--------------------------------------------------------------------------------|--------------------|-----------------------------------------------------------------------------------|
| `read()`      | Reads the entire file (or a specified number of characters) at once.           | Single string      | Less efficient for large files as it loads all content into memory.               |
| `readline()`  | Reads a single line from the file, up to the newline character (`\n`).          | String             | More efficient for large files as it reads line by line.                          |
| `readlines()` | Reads all lines from the file and returns them as a list of strings.            | List of strings    | Less efficient for large files as it loads all lines into memory.                 |

#### Q1. Write a Python program to count the number of words and lines.

````python
try: 
    file = open("race.txt","r")
    fr = file.read()
    print(f"Number of words : {len(fr.split(' '))}")
    count = 0
    for i in fr.split('\n'):
        count+=1
    print(f"Number of lines : {count}")
    file.close()
except Exception as e:
    print("Error:", e)
````

````python

count = 0
try: 
    file = open("race.txt","r")
    fr = file.read()
    for i in fr:
        if i.isupper():
            count += 1
    print(f"Number of uppercase letters : {count}")
    file.close()
except Exception as e:
    print("Error:", e)
````

#### Q2. Write a Python program to count words from each line from a file

````python
# Write a Python program to display all the lines from a file
count = 0
try: 
    file = open("race.txt","r")
    fr = file.read()
    for i in fr.split('\n'):
        print(f"{count+1}. {i}")
        count += 1
    print(f"Number of Lines : {count}")
    file.close()
except Exception as e:
    print("Error:", e)
````

#### Q6. Create a Python function make_copy() that reads a text file 'input.txt' and writes its contents to a new file 'output.txt', capitalizing the first letter of each word.

``` text
For example, if 'input.txt' contains the following content:
"In the world of programming, there are no limits to what you can achieve. Aim high!"

The 'output.txt' should contain:
"In The World Of Programming, There Are No Limits To What You Can Achieve. Aim
High!"
```

````python
try:
    file = open('input.txt',"r")
    file_op = open('output.txt',"w")
    file_read = file.read()
    file_op.write(file_read.title())
    file.close()
    file_op.close()

except Exception as e:
    print("Error : ",e )
````

````python
## Gradio app
import gradio as gr
import os

def file_copy_gradio(input_file):
    try:
        # Read uploaded file
        with open(input_file.name, "r", encoding="utf-8") as f:
            content = f.read()

        # Modify content
        modified_content = content.title()

        # Write to output file
        output_path = "output.txt"
        with open(output_path, "w", encoding="utf-8") as f:
            f.write(modified_content)

        return output_path

    except Exception as e:
        return f"Error: {e}"


iface = gr.Interface(
    fn=file_copy_gradio,
    inputs=gr.File(label="Upload a text file"),
    outputs=gr.File(label="Download processed file"),
    title="Dynamic File Processing App",
    description="Upload a text file and download the title-cased version."
)

if __name__ == "__main__":
    iface.launch()

````
