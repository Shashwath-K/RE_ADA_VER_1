```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv")
print("CSV Data loaded successfully")
print(df.head())
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing values per column:")
print(missing_values)
```

```python
# Imputation techniques (can be only proceeded if the missing values do not significantly impact the dataset)

# drop any rows with the missing values
df = df.dropna()

# Drop columns with more than 30% of missing values
threshold = 0.2 * len(df)
df = df.dropna(axis=1, thresh = threshold)
print(df.isnull().sum())
```

```python
# mean median and mode level imputation.
# MEAN
# Definition: The mean is the average of a set of numbers, calculated by summing all values and dividing by the number of values.
# Formula: mean = (sum of all values) / (number of values)
# Mean Imputation: Replacing missing values in a dataset with the mean of the available (non-missing) values.

# MEDIAN
# Definition: The median is the middle value in a sorted dataset. If the number of values is even, it is the average of the two middle values.
# Formula:
#   If odd: median = middle value after sorting
#   If even: median = (middle value 1 + middle value 2) / 2
# Median Imputation: Replacing missing values with the median of the available values, useful when data has outliers.

# MODE
# Definition: The mode is the value that appears most frequently in a dataset.
# Formula: No specific formula; mode is the value with the highest frequency count.
# Mode Imputation: Replacing missing values with the most frequently occurring value, commonly used for categorical data.
```

```python
# Fill missing values with mean
df['addiction_level'].fillna(df['addiction_level'].mean(), inplace=True)

# Fill missing values with median
df['addiction_level'].fillna(df['addiction_level'].median(), inplace=True)
```

```python
# Forward fill and Backward fill

# Forward fill
df.fillna(method="ffill", inplace=True)

# Backward fill
df.fillna(method="bfill", inplace=True)
```

```python
from sklearn.impute import KNNImputer

imputer = KNNImputer(n_neighbors= 3)
df_imputed = imputer.fit_transform(df)
```

### 1. Difference between Parametric and Non-Parametric Models

A parametric model assumes a fixed form and a fixed number of parameters, regardless of the size of the dataset. For example, linear regression always learns a fixed number of coefficients. These models are simpler, faster, and require less data, but they may underfit if the true relationship is complex.

A non-parametric model does not assume a fixed structure. Its complexity can grow with the data. For example, decision trees can become deeper and more complex as more data is provided. These models are flexible and can capture complex patterns, but they are more prone to overfitting and require more computation.

In summary:
- Parametric: fixed complexity
- Non-parametric: flexible complexity

### 2. Role of an Activation Function in a Neural Network

An activation function introduces non-linearity into a neural network. Without it, even multiple layers would behave like a single linear function, limiting the model’s ability to learn complex patterns.

Activation functions such as ReLU, Sigmoid, and Tanh allow the network to:
- Capture non-linear relationships
- Learn complex features
- Approximate a wide range of functions

They are essential for making neural networks powerful beyond simple linear models.

### 3. Cold Start Problem in Recommendation Systems

The cold start problem occurs when a recommendation system has insufficient data about a new user or a new item.

- New user: no past interactions to base recommendations on
- New item: no ratings or engagement data

Collaborative filtering struggles in this scenario because it depends on historical interactions.

Common solutions include:
- Content-based filtering using user or item features
- Asking users for initial preferences
- Hybrid recommendation systems

---

### 4. Difference between Bagging and Boosting

Bagging and boosting are ensemble learning techniques used to improve model performance.

Bagging (Bootstrap Aggregating):
- Models are trained independently in parallel
- Uses random sampling with replacement
- Reduces variance

Example: Random Forest

Boosting:
- Models are trained sequentially
- Each model focuses on correcting errors made by previous models
- Reduces bias and can also reduce variance

Example: AdaBoost, Gradient Boosting

Key difference:
- Bagging: independent models
- Boosting: dependent models

---

### 5. What is Overfitting? Ways to Prevent It

Overfitting occurs when a model learns both the underlying patterns and the noise in the training data. As a result, it performs well on training data but poorly on unseen data.

Ways to prevent overfitting:
1. Use regularization (L1 or L2)
2. Apply cross-validation
3. Simplify the model (e.g., pruning in trees)
4. Increase training data
5. Use dropout in neural networks

---

### 6. Imbalanced Dataset: Why Accuracy is a Bad Metric

In a dataset with 90% class A and 10% class B, a model can predict only class A and achieve 90% accuracy while completely ignoring class B. This makes accuracy misleading.

Better evaluation metrics include:
- Precision: proportion of correct positive predictions
- Recall: proportion of actual positives correctly identified
- F1-score: harmonic mean of precision and recall
- ROC-AUC: measures classification performance across thresholds

### 7. Difference between Validation Set and Test Set

Validation set:
- Used during model training
- Helps tune hyperparameters and select models

Test set:
- Used after training is complete
- Provides an unbiased evaluation of final model performance

Key idea:
- Validation set is for model selection
- Test set is for final evaluation

---

### 8. Handling Missing Values in a Dataset

Two common approaches:

1. Removal:
   - Drop rows or columns with missing values
   - Suitable when missing data is minimal

2. Imputation:
   - Replace missing values using:
     - Mean or median for numerical data
     - Mode for categorical data
     - Advanced methods like KNN or model-based imputation

The choice depends on the amount and importance of missing data.

---

### 9. Difference between Supervised and Unsupervised Learning

Supervised learning:
- Uses labeled data (input-output pairs)
- Learns a mapping from inputs to outputs

Example: classification and regression

Unsupervised learning:
- Uses unlabeled data
- Identifies hidden patterns or structures

Example: clustering (e.g., K-means)

Key difference:
- Supervised learning uses known outputs
- Unsupervised learning does not

---

### 10. Training Set vs Test Set

Training set:
- Used to train the model
- The model learns patterns from this data

Test set:
- Used to evaluate model performance on unseen data

Why they must be kept separate:
If the model is exposed to test data during training, it may memorize it, leading to overly optimistic performance. Keeping them separate ensures the model can generalize to new data.
