```python
import matplotlib.pyplot as plt
import numpy as np

x1 = np.array([0, 1, 2, 3])
y1 = np.array([3, 8, 1, 10])
x2 = np.array([0, 1, 2, 3])
y2 = np.array([6, 2, 7, 11])

plt.plot(x1, y1, color="red", label="Line 1")
plt.plot(x2, y2, color="blue", label="Line 2")

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.xlabel("Average Pulse", fontdict=font1)
plt.ylabel("Calorie Burger", fontdict=font2)
plt.legend()
plt.title("Example", loc ="left")
plt.grid(axis="x", linestyle="dashed", color="green")
plt.show()
```
