```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self, length, width, height):
        self.length = length
        self.breadth = width 
        self.height = height
        return self.length*self.breadth*self.height  
rect = Rectangle(2,3)
print(rect.area(5,2,2))
```

```python
def parse_student_records(raw_data):
    subjects = ['Math', "Science", "English", "Social"]
    student_list = []

    for record in raw_data:
        try:
            name, scores_part = record.split('|')
        except ValueError:
            continue

        scores_dict = {sub: 0 for sub in subjects} # Default set to 0 for all the subjects

        # Split the subject entries
        entries = scores_part.split(",")

        for entry in entries:
            if ":" not in entry:
                continue

            subject, value = entry.split(":")
            subject = subject.strip()
            value = value.strip()

            # Handle invalid or non-numerical values
            if value.lower() == "Invalid":
                score = 0
            else:
                try:
                    score = int(value)
                except ValueError:
                    score = 0
            
            # Validate Range
            if score <0 or score >100:
                score = 0
            
            # Assign if subject is Valid
            if subject in subjects:
                scores_dict[subject] = score
            
            student_list.append({
                "name": name.strip(),
                "scores": scores_dict
            })
        return student_list
```

```python
def parser_student_list(raw_data):
    subjects = ["Math", "Social", "Science", "English"]
    student_list = []

    for record in raw_data:
        try:
            name, score_split = record.split("|")
        except ValueError:
            continue

        student_dict = {sub:0 for sub in subjects}

        entries = score_split.split(",")

        for entry in entries:
            if ":" not in entry:
                continue
            subject, value = entry.split(":")
            subject = subject.strip()
            value = value.strip()

            if value.lower() == "Invalid":
                score = 0
            else:
                try:
                    score = int(value)
                except ValueError:
                    score = 0

            if score < 0 or score > 100:
                score = 0
            
            if subject in subjects:
                student_dict[subject] = score
            
        student_list.append({
            "name": name,
            "scores": student_dict,
        })

    return student_list

def calculate_student_grades(student_records):
    for student in student_records:
        scores = student.get("scores", {})

        # Calculate average
        total = sum(scores.values())
        count = len(scores)
        average = total / count if count> 0 else 0

        # Assign grade (you can adjust scale if needed)
        if average >= 90:
            grade = "A"
        elif average >= 75:
            grade = "B"
        elif average >= 60:
            grade = "C"
        elif average >= 50:
            grade = "D"
        else:
            grade = "F"

        status = "Pass" if average>=50 else "Fail"

        student["average"] = round(average,2)
        student["grade"] = grade
        student["status"] = status
    return student_records


if __name__ == "__main__":
    '''
    raw_student_data = [
    "Rahul|Math:85,Science:90,English:78,Social:88",
    "Priya|Math:92,Science:88,English:95,Social:90",
    "Amit|Math:65,Science:70,English:invalid,Social:68",
    "Sneha|Math:78,Science:82,English:85,Social:80",
    "Vikram|Math:45,Science:50,English:55,Social:48"
]
'''
raw_student_data = ["Amit|Math:65,Science:70,English:invalid,Social:68"]
student = parser_student_list(raw_student_data)
print(student)
graded = calculate_student_grades(student)
print(graded)
```

```python
import numpy as np

def class_statistics(student_records):
    # Extract averages
    averages = np.array([student["average"] for student in student_records])
    
    # Basic statistics
    mean = np.mean(averages)
    median = np.median(averages)
    std_dev = np.std(averages)
    
    # Highest and lowest
    max_index = np.argmax(averages)
    min_index = np.argmin(averages)
    
    highest = {
        "name": student_records[max_index]["name"],
        "average": averages[max_index]
    }
    
    lowest = {
        "name": student_records[min_index]["name"],
        "average": averages[min_index]
    }
    
    # Above average count
    above_avg_count = np.sum(averages > mean)
    
    # Pass percentage (average ≥ 50)
    pass_count = np.sum(averages >= 50)
    pass_percentage = (pass_count / len(averages)) * 100
    
    return {
        "mean": round(mean, 2),
        "median": round(median, 2),
        "std_dev": round(std_dev, 2),
        "highest": highest,
        "lowest": lowest,
        "above_average_count": int(above_avg_count),
        "pass_percentage": round(pass_percentage, 2),
        "averages_array": averages
    }
```

```python
stats = class_statistics(graded)
print(stats)
```
