```python
# Define a function that accepts a variable number of positional arguments using *numbers
# The *numbers parameter collects all passed arguments into a tuple named 'numbers'
def my_function(*numbers):
  # If no arguments are passed, return None
  if len(numbers) == 0:
    return None
  # Initialize max_num with the first element of the tuple
  max_num = numbers[0]
  # Loop through each number in the numbers tuple
  for num in numbers:
    # Update max_num if a larger number is found
    if num > max_num:
      max_num = num
  # Return the largest number found
  return max_num

# Call the function with several arguments and print the result (should print 9)
print(my_function(3, 7, 2, 9, 1))
```

Useful when you want your function to accept any number of named arguments

Python follows the LEGB rule when looking up variable names, and searches for them in this order:

Local - Inside the current function
Enclosing - Inside enclosing functions (from inner to outer)
Global - At the top level of the module
Built-in - In Python's built-in namespace

Decorator functions in Python are necessary because they provide a powerful, elegant way to extend or alter the functionality of existing functions or methods without modifying their actual source code.

```python
# Define a decorator function 'changecase' that takes a function 'func' as input
def changecase(func):
  # Define an inner wrapper function
  def myinner():
    # Call the original function 'func', then convert its return value to uppercase
    return func().upper()
  # Return the wrapper function
  return myinner

# Use the decorator '@changecase' to decorate 'myfunction'
@changecase
def myfunction():
  # Original function simply returns the string "Hello Sally"
  return "Hello Sally"

# When we call 'myfunction', it is actually the wrapped version provided by 'changecase'
# It returns the uppercase version of the original string
print(myfunction())
```

```python
# Define a decorator function 'changecase' which takes a function 'func' as input
def changecase(func):
  # Define a nested wrapper function 'myinner' that wraps 'func'
  def myinner():
    # Call the original function 'func', then convert its output string to uppercase
    return func().upper()
  # Return the wrapper function 'myinner' to replace the original function
  return myinner

# Use the decorator '@changecase' to decorate 'myfunction'
@changecase
def myfunction():
  # Original function returns the string "Hello Sally"
  return "Hello Sally"

# Use the decorator '@changecase' to decorate 'otherfunction'
@changecase
def otherfunction():
  # Original function returns the string "I am speed!"
  return "I am speed!"

# Call the decorated 'myfunction' which will return uppercase version of its string
print(myfunction())  # Outputs: HELLO SALLY

# Call the decorated 'otherfunction' which will return uppercase version of its string
print(otherfunction())  # Outputs: I AM SPEED!
```

```python
# Define a decorator function named 'changecase' that takes a function 'func' as an argument
def changecase(func):
  # Define an inner function 'myinner' that takes one argument 'x'
  def myinner(x):
    # Call the original function 'func' with argument 'x'
    # Then convert the returned string to uppercase
    return func(x).upper()
  # Return the inner wrapper function 'myinner'
  return myinner

# Use the decorator '@changecase' to decorate 'myfunction'
@changecase
def myfunction(nam):
  # Original function returns a greeting string concatenated with the input 'nam'
  return "Hello " + nam

# Call the decorated function 'myfunction' with argument "John"
# It actually calls 'myinner', which calls the original function and converts output to uppercase
print(myfunction("John"))
```

```python
# Define a decorator function 'changecase' that takes a function 'func' as input
def changecase(func):
  # Define a wrapper function 'myinner' that accepts any number of positional and keyword arguments
  def myinner(*args, **kwargs):
    # Call the original function 'func' with all received arguments and keyword arguments
    # Then convert its return value (assumed to be a string) to uppercase
    return func(*args, **kwargs).upper()
  # Return the wrapper function
  return myinner

# Apply the 'changecase' decorator to 'myfunction'
@changecase
def myfunction(nam):
  # Original function returns a greeting concatenated with the argument 'nam'
  return "Hello " + nam

# Call the decorated 'myfunction' with argument "John"
# This calls 'myinner', which calls the original function and then converts output to uppercase
print(myfunction("John"))
```

```python
# Define a decorator factory 'changecase' that accepts an argument 'n'
def changecase(n):
  # This is the actual decorator function that takes the function to be decorated as input
  def changecase(func):
    # Define the wrapper function 'myinner' that will replace the original function
    def myinner():
      # Check the value of 'n' passed to the outer decorator
      if n == 1:
        # If n == 1, call the original function and convert its output to lowercase
        a = func().lower()
      else:
        # Otherwise, call the original function and convert its output to uppercase
        a = func().upper()
      # Return the modified output
      return a
    # Return the wrapper function
    return myinner
  # Return the decorator function
  return changecase

# Apply the parameterized decorator with argument 1 to 'myfunction'
@changecase(-1)
def myfunction():
  # The original function returns the string "Hello Linus"
  return "Hello Linus"

# Call the decorated function 'myfunction', which will apply the decorator logic to output "hello linus"
print(myfunction())
```

```python
# Define a decorator function 'changecase' that takes a function 'func' as input
def changecase(func):
  # Define the wrapper function 'myinner' that replaces the original function
  def myinner():
    # Call the original function and convert its returned string to uppercase
    return func().upper()
  # Return the wrapper function as the decorated version of 'func'
  return myinner

# Apply the 'changecase' decorator to 'myfunction'
@changecase
def myfunction():
  # Original function returns the string "Have a great day!"
  return "Have a great day!"

# Print the __name__ attribute of the decorated function
print(myfunction.__name__)  # Outputs: 'myinner' instead of 'myfunction'
```

```python
import functools

# Define the decorator 'changecase' that wraps a function
def changecase(func):
  # Use functools.wraps to preserve the original function's metadata (name, docstring)
  @functools.wraps(func)
  def myinner():
    # Call the original function and convert output to uppercase
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Have a great day!"

# Print the function's name
print(myfunction.__name__)  # Outputs: myfunction (not 'myinner')
```

```python
# Define a recursive function 'countdown' that takes one parameter 'n'
def countdown(n):
  # Base case: check if n is less than or equal to 0
  if n <= 0:
    # When the base case is reached, print "Done!" and stop recursion
    print("Done!")
  else:
    # Recursive case: print the current value of 'n'
    print(n)
    # Call the function again with a decremented value (n - 1)
    countdown(n - 1)

# Call the function starting with n = 5
countdown(5)
```

```python
# Define a recursive function 'factorial' that calculates the factorial of n
def factorial(n):
  # Base case: if n is 0 or 1, the factorial is 1
  # This is the stopping condition for recursion
  if n == 0 or n == 1:
    return 1
  # Recursive case: for any other value of n
  else:
    # Return n multiplied by the factorial of (n - 1)
    # This breaks the problem into smaller subproblems
    return n * factorial(n - 1)

# Call the function with n = 5 and print the result
print(factorial(5))
```

```python
# Define a recursive function 'fibonacci' that calculates the nth Fibonacci number
def fibonacci(n):
  # Base case: if n is 0 or 1, return n itself
  # This is the stopping condition for recursion
  if n <= 1:
    return n
  # Recursive case: for any value of n greater than 1
  else:
    # Return the sum of the two previous Fibonacci numbers
    # fibonacci(n-1) + fibonacci(n-2)
    return fibonacci(n - 1) + fibonacci(n - 2)

# Call the function with n = 7 and print the result
print(fibonacci(7))
```

```python
# Define a recursive function 'sum_list' that calculates the sum of all numbers in a list
def sum_list(numbers):
  # Base case: if the list is empty (length is 0), return 0
  # This is the stopping condition for recursion
  if len(numbers) == 0:
    return 0
  # Recursive case: for non-empty lists
  else:
    # Return the first element (numbers[0]) plus the sum of the remaining list
    # numbers[1:] creates a slice containing all elements except the first one
    return numbers[0] + sum_list(numbers[1:])

# Create a list of numbers to sum
my_list = [1, 2, 3, 4, 5]

# Call the function with the list and print the result
print(sum_list(my_list))
```

```python
# Define a recursive function 'find_max' that finds the maximum number in a list
def find_max(numbers):
  # Base case: if the list contains only one element, return that element
  # This is the stopping condition for recursion
  if len(numbers) == 1:
    return numbers[0]
  # Recursive case: for lists with more than one element
  else:
    # Call find_max recursively on the rest of the list (all elements except the first)
    max_of_rest = find_max(numbers[1:])
    # Compare the first element with the maximum of the rest
    # Return the first element if it's greater, otherwise return the max of the rest
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest

# Create a list of numbers
my_list = [3, 7, 2, 9, 1]

# Call the function with the list and print the result
print(find_max(my_list))
```

```python
# Define a generator function 'my_generator'
# A generator is a special function that uses 'yield' instead of 'return'
def my_generator():
  # yield 1: Suspend execution and return 1 to the caller
  # The function state is saved for the next call
  yield 1
  # yield 2: Resume from where it paused and return 2
  yield 2
  # yield 3: Resume again and return 3
  yield 3
  # When the function ends, the generator stops

# Use a for loop to iterate through the generator
# Each iteration calls the generator and retrieves the next yielded value
for value in my_generator():
  # Print each value returned by the generator
  print(value)
```

```python
# Define a generator function 'count_up_to' that yields numbers from 1 to n
def count_up_to(n):
  # Initialize a counter variable starting at 1
  count = 1
  # Loop while count is less than or equal to n
  while count <= n:
    # Yield the current count value to the caller
    # The function pauses here and saves its state for the next call
    yield count
    # Increment the counter by 1
    count += 1
  # When the loop ends, the generator stops (implicitly returns)

# Use a for loop to iterate through the generator
# Each iteration retrieves the next yielded value from count_up_to(5)
for num in count_up_to(5):
  # Print each number yielded by the generator
  print(num)
```

Generators Saves Memory
Generators are memory-efficient because they generate values on-the-fly instead of storing everything in memory.

For large datasets, generators save memory:

```python
# Define a generator function 'large_sequence' that yields numbers from 0 to n-1
def large_sequence(n):
  # Loop from 0 to n-1
  for i in range(n):
    # Yield each number one at a time
    # The function pauses here and saves its state
    yield i

# Create a generator object for 1 million numbers
# Important: This does NOT create all 1 million numbers in memory immediately
# Instead, it creates a generator that will produce them on demand
gen = large_sequence(1000000)

# Call next() on the generator to get the first yielded value
# This starts execution and runs until the first yield
print(next(gen))  # Output: 0

# Call next() again to get the second yielded value
# Execution resumes after the previous yield and continues until the next yield
print(next(gen))  # Output: 1

# Call next() a third time to get the third yielded value
print(next(gen))  # Output: 2
# This demonstrates how generators can handle large sequences efficiently without using large amounts of memory.
```

```python
# Define a generator function 'simple_gen' that yields three strings one by one
def simple_gen():
  # Yield the string "Emil" and pause execution
  yield "Emil"
  # Yield the string "Tobias" and pause execution again
  yield "Tobias"
  # Yield the string "Linus" and pause execution once more
  yield "Linus"

# Create a generator object by calling the generator function
gen = simple_gen()

# Use next() to get the next value yielded by the generator
# The first call resumes execution, yields "Emil", then pauses
print(next(gen))  # Output: Emil

# The second call resumes execution after the first yield, yields "Tobias"
print(next(gen))  # Output: Tobias

# The third call resumes execution after the second yield, yields "Linus"
print(next(gen))  # Output: Linus

# If next() were called again here, it would raise StopIteration as there are no more yields
print(next(gen)) 
```

```python
# List comprehension - creates a list of squares from 0 to 4 (all at once in memory)
list_comp = [x * x for x in range(5)]
print(list_comp)  # Output: [0, 1, 4, 9, 16]

# Generator expression - creates a generator object that generates squares on demand
gen_exp = (x * x for x in range(5)) # Note the use of parentheses instead of brackets. The parentheses () 
# are the Python syntax shortcut that implicitly includes the generator logic (what yield does in a function).
print(gen_exp)  # Output: <generator object ...> (the generator object itself)

# Convert the generator into a list by iterating through it (consuming the generator)
print(list(gen_exp))  # Output: [0, 1, 4, 9, 16]
# This code demonstrates the difference between list comprehensions and generator expressions.  
```

```python
# Calculate the sum of squares from 0 to 9 using a generator expression
# This does not create an intermediate list, saving memory
total = sum(x * x for x in range(10))

# Print the result
print(total)  # Output: 285
```

The expression x * x for x in range(10) is a generator expression.

It generates the square of each number from 0 to 9 on the fly, one at a time.

The sum() function consumes these values as they are produced and adds them together.

This approach is memory-efficient because the entire list of squares isn't stored in memory.

It's equivalent to sum([x * x for x in range(10)]) but saves space for large ranges.

This is a Pythonic, efficient, and readable way to compute sums of transformed sequences without creating full intermediate lists

```python
# Define a generator function 'fibonacci' that yields Fibonacci numbers indefinitely
def fibonacci():
  # Initialize the first two Fibonacci numbers
  a, b = 0, 1
  # Infinite loop to generate Fibonacci sequence endlessly
  while True:
    # Yield the current number 'a' in the sequence
    yield a
    # Update 'a' and 'b' to the next two Fibonacci numbers
    a, b = b, a + b

# Create a generator object which will produce Fibonacci numbers
gen = fibonacci()

# Use a loop to get and print the first 100 Fibonacci numbers
for _ in range(100):
  # next() retrieves the next value yielded from the generator
  print(next(gen))
```

```python
# Define a generator function that acts like a coroutine waiting to receive data
def echo_generator():
  while True:
    # 'yield' expression pauses here and receives data sent by 'send()'
    received = yield
    # Print the received data
    print("Received:", received)

# Create the generator object
gen = echo_generator()

# Prime the generator by advancing to the first 'yield'
next(gen)

# Send data to the generator; it resumes at the yield and 'received' gets the value
gen.send("Hello")  # Prints: Received: Hello
gen.send("World")  # Prints: Received: World
```

```python
# Define a generator function 'my_gen' with a try-finally block
def my_gen():
  try:
    # Yield values one by one
    yield 1
    yield 2
    yield 3
  finally:
    # This block will always execute when the generator is closed or exhausted
    print("Generator closed")

# Create a generator object
gen = my_gen()

# Get the first yielded value (1) from the generator
print(next(gen))

# Close the generator explicitly
# This triggers the 'finally' block inside the generator
gen.close()
```
