```python
# Removing stopwords from text
text = "The quick brown fox is jumping over the lazy dog while it is sleeping under a tree"

# 1. Tokenize the text (turn string into a list of words)
tokens = word_tokenize(text)

# 2. Load the English stopwords
stop_words = set(stopwords.words('english'))

# 3. Filter out the stopwords
## Go through every word and keep it only if it isn't a stopword.
filtered_text = [word for word in tokens if word.lower() not in stop_words]

print(f"Original ({len(tokens)} words): {tokens}\n")
print(f"Filtered ({len(filtered_text)} words): {filtered_text}")
```

### Stemming - 
cutting words down to a common root-like form.

```python
# Import porterstemmer
from nltk.stem.porter import PorterStemmer
stemmer = PorterStemmer()
word1, word2 = "cars", "formality"
print(stemmer.stem(word1), stemmer.stem(word2))
```

```python
words = ['play', 'playing', 'played', 'plays']
words

# get stemmed words
[stemmer.stem(word) for word in words]
```

Key point: Stemming is basically cutting words down to their root-like form, and it doesn't always produce a proper English word.

#### Lemmatization
Tries to find the actual meaningful base word.

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

word1 = "study"
word2 = "studies"
word3 = "studying"
word4 = "studious"

#lemmatized words
print(lemmatizer.lemmatize(word1), lemmatizer.lemmatize(word2),lemmatizer.lemmatize(word3), lemmatizer.lemmatize(word4))

# compare output with stemming
print(stemmer.stem(word1), stemmer.stem(word2), stemmer.stem(word3), stemmer.stem(word4))
```

Think of it this way:

Stemming = chopping 

Lemmatization = looking up in a dictionary

#### Main Parts of Speech

POS	Meaning	Example

NN	Noun	phone, car, book

VB	Verb	run, eat, go

JJ	Adjective	good, beautiful, big

RB	Adverb	quickly, slowly, very

PRP	Pronoun	I, he, she, it

DT	Determiner	the, a, an

IN	Preposition	in, on, at

CC	Conjunction	and, but, or

CD	Number	one, 10, 25

UH	Interjection	wow, oh, hey

NLTK has more detailed tags than just noun/verb/adjective.

NN   → noun
NNS  → plural noun
VB   → base verb
VBD  → past tense verb
VBG  → verb ending in -ing
JJ   → adjective
JJR  → comparative adjective
JJS  → superlative adjective
RB   → adverb

### NER (Named Entity Recognition)

```python
# NER (Named Entity Recognition using Spacy)
sentence="Warren Edward Buffett is an American investor, business tycoon, philanthropist, and the chairman and CEO of Berkshire Hathaway."

# sentence = "Apple is looking at buying U.K. startup for $1 billion"

doc = nlp(sentence)

for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

####  In the output, the first column specifies the entity, the next two columns the start and end characters within the sentence/document, and the final column specifies the category.

PERSON → person
ORG → organization/company
GPE → geographical location
DATE → date
MONEY → money
NORP → nationality/religious/political group
