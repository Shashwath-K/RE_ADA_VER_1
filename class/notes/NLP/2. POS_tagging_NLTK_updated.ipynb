# Chunking

`grammar = "NP: {<DT>?<JJ>*<NN>}"`

Meaning:

Optional determiner (DT)

Zero or more adjectives (JJ)

Followed by noun (NN)

```python

grammar = "NP: {<DT>?<JJ>*<NN>}"
chunk_parser = nltk.RegexpParser(grammar)
chunk_tree = chunk_parser.parse(pos_tags)

print(chunk_tree)
```

```python
# Extarct Noun Phrase from tree

for subtree in chunk_tree.subtrees():
    if subtree.label() == 'NP':
        print(subtree)
```

## Identify the Verb phrase



```python
sentence="The boy is playing  football"
words = nltk.word_tokenize(sentence)
pos_tags = nltk.pos_tag(words)
```

```python
grammar = "VP: {<VB*><NP|PP>*}"
chunk_parser = nltk.RegexpParser(grammar)
chunk_tree = chunk_parser.parse(pos_tags)

print(chunk_tree)
```
