```python
import pandas as pd
# Read the data
df = pd.read_csv("height_weight_data.csv")
df.head(10)
```

```python
X = df[['HT1']]
y = df['WT']
m = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(m)
```

```python
# Emperical modelling and predictive modelling.
# Build the LR class. Build the fit and __init__ method
class LR:
    def __init__(self):
        # Initialize the weights/ slope as None
        self.m = None

    def fit(self, x_train, y_train):
        # Ensure x_train is a 2D array(N rows, 1 column)
        # and y_train is in a 1D array
        X = np.array(x_train).reshape(-1,1)
        y = np.array(y_train)

        # Calculate the slope using the Normal Equation : (X^T * X) ^(-1) * X^T *y

        self.m = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
    def predict(self, X_test):
        # Ensure the test data is also 2D for the math to align
        X_test = np.array(X_test).reshape(-1,1)

        if self.m is None:
            raise ValueError("Not fitted yet")
        
        # y = m* x
        return X_test @ self.m
```

```python
# Dummy Data: Height vs Weight (Assuming a direct relationship through the origin)
heights = [150, 160, 170, 180]
weights = [55, 60, 64, 70]

# Instantiate and train the model
model = LR()
model.fit(heights, weights)

print(f"Calculated Slope (m): {model.m[0]}") 
# Output will be the optimized slope

# Predict the weight of someone who is 175 cm tall
predicted_weight = model.predict([175])
print(f"Predicted Weight for 175cm: {predicted_weight[0]}")
```
