```python
# Import the Iris dataset from sklearn
from sklearn.datasets import load_iris

# Import Decision Tree module from sklearn
from sklearn import tree

# Import pandas library for data handling
import pandas as pd

# Load the Iris dataset into memory
iris = load_iris()

# Create a Decision Tree Classifier object
clf = tree.DecisionTreeClassifier()

# Train (fit) the Decision Tree model using
# iris.data as input features and iris.target as output labels
clf = clf.fit(iris.data, iris.target)

# Import Graphviz library for visualizing the Decision Tree
import graphviz

# Convert the trained Decision Tree into DOT format
# feature_names = names of input features
# class_names = names of target classes
# filled=True = color the nodes based on class
# rounded=True = make node corners rounded
# special_characters=True = allow special characters in labels
dot_data = tree.export_graphviz(
    clf,
    out_file=None,
    feature_names=iris.feature_names,
    class_names=iris.target_names,
    filled=True,
    rounded=True,
    special_characters=True
)

# Create a Graphviz object from the DOT data
graph = graphviz.Source(dot_data)

# Display the Decision Tree graph
graph
```

```python
# ============================================================
#              DECISION TREE CLASSIFICATION - TITANIC
# ============================================================

# ===========================
# 1. Import Required Libraries
# ===========================

import numpy as np                      # Numerical operations
import pandas as pd                     # Data manipulation
import seaborn as sns                   # Data visualization
from matplotlib import pyplot as plt    # Plotting graphs

# Load Titanic dataset
titanic = sns.load_dataset('titanic')

# Display first 5 rows
titanic.head()

# Display multiple outputs in Jupyter Notebook
from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"

# Display plots inside notebook
%matplotlib inline


# ===========================
# 2. Data Preprocessing
# ===========================

# Drop unnecessary columns
titanic.drop(
    columns=[
        'who',
        'adult_male',
        'embark_town',
        'alone',
        'alive',
        'class',
        'deck'
    ],
    inplace=True
)

# Replace missing values in embarked column with mode
titanic['embarked'] = titanic['embarked'].fillna(
    titanic['embarked'].mode()[0]
)

# Replace missing values in age column with median
titanic['age'] = titanic['age'].fillna(
    titanic['age'].median()
)

# Rename columns
titanic = titanic.rename(columns={
    "sex": "gender",
    "sibsp": "siblings",
    "parch": "parents_child"
})

# Display first five rows
titanic.head()

# Check dataset information
titanic.info()


# ===========================
# 3. Encode Categorical Variables
# ===========================

# Convert categorical variables into numerical values
for x in titanic.columns:
    if titanic[x].dtype == "object":
        titanic[x] = pd.Categorical(titanic[x]).codes

# Display encoded dataset
titanic.head()


# ===========================
# 4. Import Machine Learning Libraries
# ===========================

from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn import tree


# ===========================
# 5. Create Features and Target
# ===========================

# Features
x = titanic.drop(['survived'], axis=1)

# Target
y = titanic['survived']

# Display shapes
x.shape
print()
y.shape


# ===========================
# 6. Split Training and Testing Data
# ===========================

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=999
)

# Display shapes
x_train.shape
y_train.shape
x_test.shape
y_test.shape


# ===========================
# 7. Create Decision Tree Model
# ===========================

mytree = tree.DecisionTreeClassifier(
    min_samples_leaf=50,
    min_samples_split=75,
    max_depth=3,
)


# ===========================
# 8. Train the Model
# ===========================

mytree.fit(x_train, y_train)


# ===========================
# 9. Make Predictions
# ===========================

predicted = mytree.predict(x_test)

# Display predictions
predicted


# ===========================
# 10. Evaluate Model
# ===========================

# Print Classification Report
print(metrics.classification_report(y_test, predicted))


# ===========================
# 11. Confusion Matrix
# ===========================

# Generate confusion matrix
df_confusion = metrics.confusion_matrix(
    y_test,
    predicted
)

# Display confusion matrix
df_confusion


# ===========================
# 12. Plot Confusion Matrix
# ===========================

sns.heatmap(
    df_confusion,
    cmap='Blues',
    xticklabels=['Prediction No', 'Prediction Yes'],
    yticklabels=['Actual No', 'Actual Yes'],
    annot=True,
    fmt='d'
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()


# ===========================
# 13. Calculate Accuracy
# ===========================

accuracy = metrics.accuracy_score(
    y_test,
    predicted
)

print("Accuracy :", accuracy)


# ===========================
# 14. Visualize Decision Tree
# ===========================

# Install graphviz (Run only once)
!pip install graphviz

import graphviz

# Export tree
dot_data = tree.export_graphviz(
    mytree,
    out_file=None,
    feature_names=x.columns,
    class_names=['0', '1'],
    filled=True,
    rounded=True,
    special_characters=True
)

# Create graph
graph = graphviz.Source(dot_data)

# Display tree
graph

# Save tree as PNG image
graph.render("Titanic")
```
